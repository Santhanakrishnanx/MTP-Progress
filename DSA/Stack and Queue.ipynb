# Implement Stack using array 

```python
class Stack:
    def __init__(self):
        self.items = []
    # TC : O(1)
    def is_empty(self):
        return len(self.items) == 0
    # TC : O(1)
    def push(self,item):
        self.items.append(item)
    # TC : O(1)
    def pop(self):
        if self.is_empty():
            print("cannot pop, stack is empty")
        x = self.items.pop()
        return x
    # TC : O(1)
    def top(self):
        if self.is_empty():
            print("cannot top, stack is empty")
        return self.items[-1]
    # TC : O(1)
    def size(self):
        return len(self.items)

stack = Stack()
stack.push(5)
stack.push(10)
stack.push(15)
print(f"Poped item = {stack.pop()}")
```

# Implement Queue using arrays

```python
class Queue:
    def __init__(self):
        self.items = []
    # TC : O(1)
    def is_empty(self):
        return len(self.items) == 0
    # TC : O(1)
    def enqueue(self,items):
        self.items.append(items)
    # TC : O(n)
    def dequeue(self):
        if len(self.items) == 0:
            print("Dequeue from empty queue")
            return
        x = self.items.pop(0)
        return x
    # TC : O(1)
    def front(self):
        if len(self.items) == 0:
            print("Cannot peek, queue is empty")
            return
        return self.items[0]
    # TC : O(1)
    def rear(self):
        if len(self.items) == 0:
            print("Cannot read, queue is empty")
            return
        return self.items[-1]
    # TC : O(1)
    def size(self):
        return len(self.items)

queue = Queue()
queue.enqueue(23)
queue.enqueue(5)
queue.enqueue(15)
print("Size: " ,queue.size())
print("Front: ",queue.front())
print("Rear: ",queue.rear())
```

# Implement Queue using Two Stack 

```python
class StackQueue:
    def __init__(self):
        self.st1 = []
        self.st2 = []

    def push(self,x):
        # transfer all elements from stack 1 to stack 2
        while self.st1:
            self.st2.append(self.st1.pop())
        # insert element to stack 1
        self.st1.append(x)
        # transfer all elements from stack 2 to stack 1
        while self.st2:
            self.st1.append(self.st2.pop())

    def pop(self):
        if not self.st1:
            print("Stack is empty")
            return -1 
        # pop top element of stack 1
        top = self.st1.pop()
        return top

    def peek(self):
        if not self.st1:
            print("Stack is empty")
            return -1 
        return self.st1[-1]

    def is_empty(self):
        return not self.st1
        
queue = StackQueue()
queue.push(5)
queue.push(10)
queue.push(15)
print(f"Poped item = {queue.pop()}")
print(f"Poped item = {queue.pop()}")
print(f"Poped item = {queue.pop()}")
```

# Implement stack using Doubly Linked List 

```python
class Node:
    def __init__(self,x):
        self.prev = None
        self.val = x
        self.next = None

class StackDLL:
    def __init__(self):
        self.head = None
        self.tail = None

    def push(self,x):
        temp = Node(x)
        if self.head == None:
            self.head = temp
            self.tail = temp
            return
        if self.tail == self.head:
            self.head.next = temp
            temp.prev = self.head
            self.tail = temp
            return
        self.tail.next = temp
        temp.prev = self.tail
        self.tail = temp
        return

    def peek(self):
        if self.tail == None:
            print("Cannot peek, an empty stack")
            return
        return self.tail.val

    def pop(self):
        if self.tail == None:
            print("Cannot pop, an empty stack")
            return
        temp = self.tail.val
        self.tail = self.tail.prev
        if self.tail == None:
            return temp
        self.tail.next = None
        return temp

stack = StackDLL()
stack.push(5)
stack.push(10)
stack.push(15)
print(f"Poped item = {stack.pop()}")
print(f"Poped item = {stack.pop()}")
print(f"Poped item = {stack.pop()}")
```

## Infix to Postfix Convertion
- Time Complexity : O(N) + O(N)
- Space Complexity : O(N) + O(N)

```python
class Solution:
    def precedence(self,ch):
        if ch == "+" or ch == "-":
            return 1
        if ch == "*" or ch == "/":
            return 2
        if ch == "^":
            return 3
        return 0
    def InfixtoPostfix(self,s):
        stack = []
        result = []
        for char in s:
            if ("a" <= char <="z") or ("A" <= char <="Z") or ("0" <= char <="9"):
                result.append(char)
            # if ( , then push it to stack
            elif char == "(":
                stack.append(char)
            # if ) , then pop until ( is encountered
            elif char == ")":
                while stack and stack[-1] != "(":
                    result.append(stack.pop())
                stack.pop()
            # if char is an operator
            else:
                while stack and self.precedence(stack[-1]) >= self.precedence(char):
                    result.append(stack.pop())
                stack.append(char)
        #pop remaining operators 
        while stack:
            result.append(stack.pop())
        return "".join(result)

sol = Solution()
y = "a+b*(c^d-e)"
print(y)
x = sol.InfixtoPostfix(y)
print(x)
```

## Infix to Prefix Convertion
- Time Complexity : O(6N)
- Space Complexity : O(N) + O(N)

```python
class Solution:
    def precedence(self,ch):
        if ch == "+" or ch == "-":
            return 1
        if ch == "*" or ch == "/":
            return 2
        if ch == "^":
            return 3
        return 0
    def InfixtoPrefix(self,s):
        # reverse the infix  
        s = s[::-1]
        s = s.replace("(","temp").replace(")","(").replace("temp",")")
        stack = []
        result = []
        # infix to postfix
        for char in s:
            if ("a" <= char <="z") or ("A" <= char <="Z") or ("0" <= char <="9"):
                result.append(char)
            # if ( , then push it to stack
            elif char == "(":
                stack.append(char)
            # if ) , then pop until ( is encountered
            elif char == ")":
                while stack and stack[-1] != "(":
                    result.append(stack.pop())
                stack.pop()
            # if char is an operator
            else:
                while stack and self.precedence(stack[-1]) > self.precedence(char):
                    result.append(stack.pop())
                stack.append(char)
        #pop remaining operators 
        while stack:
            result.append(stack.pop())
        #reverse the answer
        return "".join(result[::-1])
        
sol = Solution()
y = "a+b*(c^d-e)"
print(y)
x = sol.InfixtoPrefix(y)
print(x)
```

## Postfix to Infix Convertion
- Time Complexity : O(N)
- Space Complexity : O(N)

```python
class Solution:
    def PostfixtoInfix(self,s):
        stack = []
        for char in s:
            # if character is an operand push it to stack
            if char.isalnum(): # alpha numeric
                stack.append(char)
            else:
                # pop two operands
                operand2 = stack.pop()
                operand1 = stack.pop()
                # compine operands with operator 
                new_exp = f"({operand1}{char}{operand2})"
                stack.append(new_exp)
        # final element in the stack is the result 
        return stack[-1]
        
sol = Solution()
y = "abcd^e-*+"
print(y)
x = sol.PostfixtoInfix(y)
print(x)
```

# Next Greater Element

- Monotonic stack means storing the element in a specific order

```python
def next_greater(arr):
    result,stack = [],[]
    for i in range(len(arr) - 1,-1,-1):
        while stack and stack[-1] <= arr[i]:
            stack.pop()
        if stack == []:
            stack.append(arr[i])
            result.append(-1)
            continue
        result.append(stack[-1])
        stack.append(arr[i])
    return result[::-1]
arr = [4,12,5,3,1,2,5,3,1,2,4,6]
temp = next_greater(arr)
print("next greater element: ")
for i in range(len(arr)):
    print(f"{arr[i]} - > {temp[i]}")
```

#### LeetCode 225 - Implement Stack using Queues

**Question:**

Implement a Last-In-First-Out (LIFO) stack using only queue operations. The stack should support `push`, `pop`, `top`, and `empty`.

- `push(x)`: Push element `x` onto the stack.
- `pop()`: Remove and return the top element of the stack.
- `top()`: Return the top element without removing it.
- `empty()`: Return `True` if the stack is empty; otherwise, return `False`.

**Example:**

- Input: `["MyStack", "push", "push", "top", "pop", "empty"]` with values `[[], [1], [2], [], [], []]`
- Output: `[null, null, null, 2, 2, false]`

**Algorithm:**

**Brute Force:**

- Use a queue to store the elements in the order they are inserted.
- Implement `push(x)` by appending the element to the back of the queue.
- Implement `pop()` by removing the front element of the queue.
- Implement `top()` by accessing the front element of the queue.
- This approach follows FIFO order rather than LIFO order, so it does not correctly implement a stack without additional operations to rearrange the elements.

**Optimal Approach:**

- Use a single queue implemented with `deque`.
- Maintain the queue so that its front always represents the top of the stack.
- In `push(x)`, append the new element to the back of the queue.
- After inserting the element, rotate the previous elements behind it so that the newly inserted element moves to the front.
- Perform this rotation using `len(self.queue) - 1` iterations.
- In each iteration, remove the front element using `popleft()` and append it to the back using `append()`.
- For example, if the queue contains `[1, 2]` and we push `3`, appending gives `[1, 2, 3]`. Rotating the first two elements produces `[3, 1, 2]`.
- In `pop()`, remove and return the front element using `popleft()`, because the front represents the stack's top.
- In `top()`, return the front element using `self.queue[0]` without removing it.
- In `empty()`, return whether the queue contains zero elements.
- Each push rearranges the existing elements to preserve LIFO order, while pop and top operate directly on the front.

**Complexity:**

**Push:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space

**Pop:** Time Complexity: O(1)  Space Complexity: O(1)

**Top:** Time Complexity: O(1)  Space Complexity: O(1)

**Empty:** Time Complexity: O(1)  Space Complexity: O(1)

**Common Mistakes:**

- Appending the new element without rotating the queue, which leaves the oldest element at the front.
- Rotating `len(self.queue)` times instead of `len(self.queue) - 1` times.
- Using `pop()` on the queue instead of `popleft()` to remove the front element.
- Forgetting that the queue front must always represent the stack top.
- Removing the element in `top()` instead of only returning it.
- Forgetting to handle an empty queue before calling `popleft()` or accessing index `0`.
- Returning a string for an empty stack when the expected method return type is an integer. For this problem, operations are normally called only when valid; if handling empty cases explicitly, use a consistent approach.

**Recognition Pattern:**

- Implement a LIFO data structure using a FIFO queue.
- Think: Queue → Append New Element → Rotate Previous Elements → Front Becomes Stack Top.

**Key Observation:**

A queue naturally follows FIFO order, whereas a stack follows LIFO order. By rotating the queue after every push, the newest element is moved to the front. This makes `pop()` and `top()` O(1), at the cost of O(N) time for each push.

**Code:**

```python
from collections import deque
class MyStack:

    def __init__(self):
        self.queue = deque()

    def push(self, x: int) -> None:
        self.queue.append(x)
        # after append at last then rotate n-1 elements before
        for _ in range(len(self.queue) - 1):
            self.queue.append(self.queue.popleft())

    def pop(self) -> int:
        if len(self.queue) == 0:
            return "Stack is empty"
        return self.queue.popleft()

    def top(self) -> int:
        if len(self.queue) == 0:
            return "Stack is empty"
        return self.queue[0]

    def empty(self) -> bool:
        return len(self.queue) == 0
        
# Your MyStack object will be instantiated and called as such:
# obj = MyStack()
# obj.push(x)
# param_2 = obj.pop()
# param_3 = obj.top()
# param_4 = obj.empty()
```

#### LeetCode 232 - Implement Queue using Stacks

**Question:**

Implement a First-In-First-Out (FIFO) queue using only stack operations. The queue should support `push`, `pop`, `peek`, and `empty`.

- `push(x)`: Add element `x` to the back of the queue.
- `pop()`: Remove and return the front element of the queue.
- `peek()`: Return the front element without removing it.
- `empty()`: Return `True` if the queue is empty; otherwise, return `False`.

**Example:**

- Input: `["MyQueue", "push", "push", "peek", "pop", "empty"]` with values `[[], [1], [2], [], [], []]`
- Output: `[null, null, null, 1, 1, false]`

**Algorithm:**

**Brute Force:**

- Use a single stack to store the queue elements.
- Push new elements onto the stack.
- To remove the front element, transfer all elements to another temporary stack until the oldest element reaches the top.
- Remove the oldest element and transfer the remaining elements back to the original stack.
- This approach requires rearranging elements for each pop operation and does not provide an efficient implementation with only one stack.

**Optimal Approach:**

- Use two stacks, `st1` and `st2`, to implement the queue.
- Maintain `st1` so that the front of the queue is always at the top of the stack.
- Use `st2` as a temporary stack to reverse the order of elements during insertion.

- **Push Operation:**
  - Transfer all elements from `st1` to `st2` by repeatedly popping from `st1` and appending to `st2`.
  - This reverses the order of the existing elements.
  - Push the new element `x` onto `st1`.
  - Transfer all elements from `st2` back to `st1`.
  - The new element is now at the bottom of `st1`, while the oldest element remains at the top.
  - For example, if `st1 = [2, 1]` with `1` at the top, pushing `3` transfers the elements to `st2`, inserts `3`, and restores the stack as `[3, 2, 1]` with `1` at the top.

- **Pop Operation:**
  - Check whether `st1` is empty.
  - If it is empty, return `None`.
  - Otherwise, remove and return the top element using `st1.pop()`.
  - Since the oldest element is maintained at the top, this correctly implements FIFO removal.

- **Peek Operation:**
  - Check whether `st1` is empty.
  - If it is empty, return `None`.
  - Otherwise, return `st1[-1]` without removing the element.

- **Empty Operation:**
  - Return `True` if `st1` is empty; otherwise, return `False`.
  - Since all queue elements are maintained in `st1`, checking this stack is sufficient.

**Complexity:**

**Push:** Time Complexity: O(N)  Space Complexity: O(N)

**Pop:** Time Complexity: O(1)  Space Complexity: O(1)

**Peek:** Time Complexity: O(1)  Space Complexity: O(1)

**Empty:** Time Complexity: O(1)  Space Complexity: O(1)

**Common Mistakes:**

- Forgetting that the top of `st1` must represent the front of the queue.
- Transferring elements in the wrong order and reversing the FIFO behavior.
- Forgetting to transfer the elements back from `st2` after inserting the new element.
- Using `st1.pop(0)` instead of `st1.pop()`.
- Removing the element in `peek()` instead of simply returning it.
- Forgetting to handle an empty queue before calling `pop()` or accessing `st1[-1]`.
- Assuming that every operation takes O(1) time when this implementation makes `push()` O(N).

**Recognition Pattern:**

- Implement a FIFO data structure using LIFO stacks.
- Think: Two Stacks → Reverse Existing Elements → Insert New Element → Restore Order → Stack Top Represents Queue Front.

**Key Observation:**

A stack reverses the order of elements when they are transferred to another stack. By moving all existing elements to `st2`, inserting the new element into `st1`, and restoring the elements, the oldest element remains at the top of `st1`. This makes `pop()`, `peek()`, and `empty()` O(1), while `push()` takes O(N) time.

**Code:**

```python
class MyQueue:

    def __init__(self):
        self.st1 = []
        self.st2 = []
    def push(self, x: int) -> None:
        while self.st1:
            self.st2.append(self.st1.pop())
        self.st1.append(x)
        while self.st2:
            self.st1.append(self.st2.pop())

    def pop(self) -> int:
        if not self.st1:
            return None
        return self.st1.pop()
        
    def peek(self) -> int:
        if not self.st1:
            return None
        return self.st1[-1]
        
    def empty(self) -> bool:
        return not self.st1
        
# Your MyQueue object will be instantiated and called as such:
# obj = MyQueue()
# obj.push(x)
# param_2 = obj.pop()
# param_3 = obj.peek()
# param_4 = obj.empty()
```

#### LeetCode 20 - Valid Parentheses

**Question:**

Given a string `s` containing only the characters `'('`, `')'`, `'{'`, `'}'`, `'['`, and `']'`, determine whether the input string is valid.

- Every opening bracket must be closed by the same type of bracket.
- Opening brackets must be closed in the correct order.
- Every closing bracket must have a corresponding opening bracket.

**Example:**

- Input: `s = "()[]{}"`
- Output: `True`
- Explanation: Every opening bracket is closed by its corresponding closing bracket in the correct order.

**Algorithm:**

**Brute Force:**

- Repeatedly search for adjacent matching pairs such as `()`, `[]`, and `{}`.
- Remove each matching pair from the string.
- Continue until no more matching pairs can be removed.
- If the string becomes empty, return `True`.
- Otherwise, return `False` because some brackets remain unmatched or incorrectly ordered.
- Repeated string searches and removals can take O(N²) time.

**Optimal Approach:**

- Use a Stack to track opening brackets that have not yet been matched.
- Traverse the string from left to right, processing one character at a time.
- If the stack is empty, push the current character because there is no unmatched opening bracket available to match it. This is safe only if the current character is an opening bracket; a closing bracket at this point makes the string invalid.
- If the current character is an opening bracket such as `(`, `[`, or `{`, push it onto the stack.
- If the current character is a closing bracket, check whether the stack contains the corresponding opening bracket at the top.
- If the top of the stack matches the current closing bracket, pop the opening bracket because the pair is now complete.
- If the top does not match the current closing bracket, return `False` because the brackets are incorrectly nested or mismatched.
- Continue processing until every character has been examined.
- After traversal, return `True` only if the stack is empty. Any remaining opening brackets mean that the string is invalid.

**Complexity:**

**Brute Force:** Time Complexity: O(N²)  Space Complexity: O(N)

**Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(N)

**Common Mistakes:**

- Checking only whether the stack is empty at the end without rejecting an unmatched closing bracket during traversal.
- Forgetting to push opening brackets onto the stack.
- Popping the stack when the closing bracket does not match the opening bracket at the top.
- Checking only whether the brackets exist without checking their nesting order.
- Forgetting to check whether the stack is empty after processing the entire string.
- Returning `True` for an empty stack at the beginning of an iteration without validating whether the current character is an opening bracket.

**Recognition Pattern:**

- Validate matching pairs of brackets with correct nesting order.
- Think: Opening Bracket → Push → Closing Bracket → Match Stack Top → Pop → Final Stack Must Be Empty.

**Key Observation:**

The most recently opened bracket must be closed first. A stack naturally follows this Last-In-First-Out (LIFO) order, so every closing bracket must match the opening bracket at the top. A mismatch immediately makes the string invalid, and a valid string must leave the stack empty after traversal.

**Code:**

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        for i in range(len(s)):
            if stack == []:
                stack.append(s[i])
            elif stack[-1] == "(":
                if s[i] == ")":
                    stack.pop()
                else:
                    stack.append(s[i])
            elif stack[-1] == "[":
                if s[i] == "]":
                    stack.pop()
                else:
                    stack.append(s[i])
            elif stack[-1] == "{":
                if s[i] == "}":
                    stack.pop()
                else:
                    stack.append(s[i])
        if stack == []:
            return True
        return False
```

- Return the minimum element found.

- Use a stack that stores a pair of values for every element: `[value, minimum_so_far]`.
- The first value represents the actual element pushed onto the stack.
- The second value represents the minimum element among all elements currently in the stack, including the newly pushed element.
- In `push(value)`, check whether the stack is empty.
- If the stack is empty, append `[value, value]` because the first element is also the minimum.
- If the stack is not empty, retrieve the previous minimum from `self.stack[-1][1]`.
- Calculate the new minimum using `min(self.stack[-1][1], value)`.
- Append `[value, min_el]` to the stack so that the minimum at every level is stored alongside its corresponding value.
- For example, pushing `-2`, `0`, and `-3` produces `[[-2, -2], [0, -2], [-3, -3]]`.

- In `top()`, return `self.stack[-1][0]`, which contains the actual top element.
- In `getMin()`, return `self.stack[-1][1]`, which contains the minimum element for the entire current stack.
- Handle empty-stack cases before accessing the last element.
- Since the current minimum is stored with every element, no traversal is required to retrieve it.

**Complexity:**

**Brute Force:** Time Complexity: Push O(1), Pop O(1), Top O(1), GetMin O(N)  Space Complexity: O(N)

**Optimal Approach:** Time Complexity: Push O(1), Pop O(1), Top O(1), GetMin O(1)  Space Complexity: O(N)

**Common Mistakes:**

- Storing only the actual value without storing the minimum at that position.
- Calculating the minimum using only the current value instead of comparing it with the previous minimum.
- Forgetting that the first element is also the initial minimum.
- Returning `self.stack[-1][0]` in `getMin()` instead of `self.stack[-1][1]`.
- Returning the stored minimum instead of the actual value in `top()`.
- Forgetting that popping a pair automatically restores the previous minimum.
- Accessing the last element when the stack is empty.

**Recognition Pattern:**

- Design a stack that supports retrieving the minimum element in O(1) time.
- Think: Stack → Store Value + Minimum So Far → Push Updates Minimum → Pop Restores Previous Minimum.

**Key Observation:**

Instead of recalculating the minimum whenever `getMin()` is called, store the minimum value alongside every element. Each pair remembers the minimum at that stack position. When an element is removed, the previous minimum becomes available automatically from the new top pair, allowing all operations to run in O(1) time.

**Code:**

- Explanation: The next greater element of the first `1` is `2`. The element `2` has no greater element, and the final `1` finds `2` after wrapping around.

**Algorithm:**

**Brute Force:**

- Iterate through every index `i` in the array.
- For each index, check the next elements one by one, considering the array circularly.
- Use `(i + j) % n` to calculate the index of each candidate, where `n` is the array length.
- Examine at most `n - 1` other elements to avoid checking the current element again.
- Stop when an element greater than `nums[i]` is found and store it in the result.
- If no greater element exists after checking the remaining elements, store `-1`.
- Repeat this process for every index and return the result.

**Optimal Approach:**

- Use a Monotonic Decreasing Stack to find the next greater element for every position.
- Since the array is circular, simulate traversing the array twice by iterating from index `2 * n - 1` down to `0`.
- Use `i % n` to map each iteration to a valid index in the original array.
- The first virtual traversal prepares the stack with candidate elements from the circular continuation. The second traversal calculates the answers for the original array positions.
- Maintain a stack of values in decreasing order from bottom to top. The stack contains potential next greater elements for the current position.
- For every iteration, access the current element using `nums[i % n]`.
- While the stack is not empty and its top element is less than or equal to the current element, pop it.

- If the stack is not empty, its top contains the nearest greater candidate to the right, considering the circular traversal. Append this value to `result`.
- If the stack is empty, no greater element exists, so append `-1`.
- Push the current element onto the stack so that it can act as a candidate for elements to its left.
- Since the original positions are processed from right to left, answers are appended in reverse index order.
- Reverse `result` using `result[::-1]` to restore the original array order.
- Return the final result.

**Complexity:**

**Brute Force:** Time Complexity: O(N²)  Space Complexity: O(1) auxiliary space

**Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(N)

**Common Mistakes:**

- Traversing the array only once and missing greater elements that appear after wrapping around.
- Forgetting to use `i % n` to access valid indices during the circular traversal.
- Popping only elements smaller than the current element instead of elements smaller than or equal to it.
- Recording results during both virtual traversals instead of only when `i < n`.
- Forgetting that results are generated in reverse index order.
- Returning `result` without reversing it.
- Pushing indices while the implementation expects stack values, or confusing the two approaches.
- Forgetting to return `-1` when the stack is empty.

**Recognition Pattern:**

- Find the next greater element in a circular array.
- Think: Circular Array → Two Traversals → Monotonic Decreasing Stack → Modulo Indexing → Reverse Results.

**Key Observation:**

A circular array allows elements near the beginning to serve as next greater elements for positions near the end. Traversing the array twice simulates this circular behavior without modifying the original array. The monotonic stack removes unnecessary candidates, and each element is pushed and popped at most once, giving O(N) time complexity.

**Code:**

```python
class Solution:
    def nextGreaterElements(self, nums: list[int]) -> list[int]:
        stack, result = [],[]
        n = len(nums)
        for i in range(2*n-1 , -1, -1):
            while len(stack) != 0 and stack[-1] <= nums[i % n]:
                stack.pop()
            if i < n:
                if len(stack) != 0:
                    result.append(stack[-1])
                else:
                    result.append(-1)
            stack.append(nums[i % n])
        return result[::-1]
```

Given an array `arr`, find the sum of the minimum element of every contiguous subarray. Return the answer modulo `10^9 + 7`.

**Example:**

- Output: `17`
- Explanation: The minimums of all contiguous subarrays add up to `17`.

**Algorithm:**

**Brute Force:**

- Generate every possible contiguous subarray by fixing a starting index `i`.
- Initialize the minimum element to infinity or the first element of the subarray.
- Extend the subarray by moving the ending index `j` from `i` to the end of the array.
- Update the minimum element whenever a smaller element is encountered.
- Add the minimum of each subarray to the total sum.
- Apply modulo `10^9 + 7` to keep the result within the required range.
- Return the final sum.

**Optimal Approach:**

- Use a Monotonic Increasing Stack to calculate the contribution of every element as the minimum of multiple subarrays.
- Store indices in the stack rather than element values because the indices are required to calculate the number of subarrays.
- Traverse the array from left to right using `range(len(arr) + 1)`.
- The extra iteration at `i == len(arr)` acts as a sentinel boundary. It forces all remaining stack elements to be processed without accessing `arr[i]` at the out-of-bounds index.
- For each index `i`, check whether the stack is non-empty and whether the current element is smaller than or equal to the element at the stack top.
- If the condition holds, pop the top index into `temp`. The popped element can now be processed because its right boundary has been found.
- The comparison `arr[stack[-1]] >= arr[i]` ensures that duplicate minimum values are handled consistently by assigning the current equal element as the right boundary.
- Determine the left boundary using `stack[-1]` after popping. If the stack is empty, use `-1` because no previous smaller element remains.
- Set the right boundary to the current index `i`, which represents the next smaller-or-equal element.
- Calculate the number of subarrays in which `arr[temp]` is the minimum:

  `Count = (temp - left) × (right - temp)`

- The distance `temp - left` gives the number of possible starting positions, while `right - temp` gives the number of possible ending positions.
- Multiply the count by `arr[temp]` to calculate the contribution of this element:

  `Contribution = Count × arr[temp]`

- Apply modulo `10^9 + 7` to the contribution and update `sum_min` using modulo arithmetic.
- Continue popping while the stack condition holds because multiple elements may find their right boundaries at the same current index.
- After processing the required elements, push the current index onto the stack. At the sentinel iteration, this index is `len(arr)`, but it is never accessed as an array element because the loop terminates immediately afterward.
- Return `sum_min`, which contains the sum of all subarray minimums modulo `10^9 + 7`.

**Complexity:**

**Brute Force:** Time Complexity: O(N²)  Space Complexity: O(1) auxiliary space

**Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(N)

**Common Mistakes:**

- Storing element values instead of indices in the stack.
- Using incorrect boundaries for the previous smaller and next smaller-or-equal elements.
- Forgetting that the number of subarrays is `(temp - left) × (right - temp)`.
- Forgetting to process the remaining stack elements using the sentinel boundary.
- Accessing `arr[i]` when `i == len(arr)` without first checking the sentinel condition.
- Mishandling duplicate elements by using inconsistent comparison conditions.
- Applying modulo only at the end instead of controlling the accumulated sum during processing.
- Forgetting to multiply the number of subarrays by the actual minimum value `arr[temp]`.

**Recognition Pattern:**

- Find the sum of minimum elements across all contiguous subarrays.
- Think: Monotonic Increasing Stack → Previous Smaller → Next Smaller-or-Equal → Contribution Counting.

**Key Observation:**

Instead of generating every subarray, calculate how many subarrays have each element as their minimum. The distances to the previous smaller element and the next smaller-or-equal element determine its contribution. The monotonic stack finds these boundaries efficiently, allowing the answer to be calculated in O(N) time.

**Code:**

```python
class Solution:
    def sumSubarrayMins(self, arr: list[int]) -> int:
        mod = 10**9 + 7 
        stack = []
        sum_min = 0
        # arr + 1 to force loop to process all the elements in stack 
        for i in range(len(arr) + 1):
            while stack and (i == len(arr) or arr[stack[-1]] >= arr[i]):
                temp = stack.pop()
                left = stack[-1] if stack else -1
                right = i
                count = (temp - left)*(right - temp) % mod
                sum_min += (count * arr[temp]) % mod
                sum_min %= mod
            stack.append(i)
        return sum_min
```

- `get(key)`: Return the value of the key if it exists; otherwise, return `-1`.
- `put(key, value)`: Update the value if the key exists. Otherwise, insert the key-value pair.
- If the cache exceeds its capacity, evict the least recently used key.
- Both operations must run in O(1) average time.

**Example:**

- Input: `["LRUCache", "put", "put", "get", "put", "get", "put", "get", "get", "get"]` with values `[2, [1, 1], [2, 2], [1], [3, 3], [2], [4, 4], [1], [3], [4]]`
- Output: `[null, null, null, 1, null, -1, null, -1, 3, 4]`
- Explanation: When the cache reaches capacity, the least recently used key is removed before a new key is inserted.

**Algorithm:**

**Brute Force:**

- Store key-value pairs in a list while maintaining their usage order.
- Whenever `get(key)` is called, search the list for the key.
- If the key exists, return its value and move it to the most recently used position.
- Whenever `put(key, value)` is called, search for the key.
- If the key exists, update its value and mark it as recently used.
- Otherwise, insert the new key-value pair at the most recently used position.
- If the cache exceeds its capacity, remove the least recently used key from the opposite end.
- Searching for keys and rearranging elements can take O(N) time per operation.

**Optimal Approach:**

- Use a Hash Map and a Doubly Linked List to implement the LRU cache with O(1) average-time operations.
- The Hash Map, `self.cache`, stores each key and a reference to its corresponding linked-list node.
- The Doubly Linked List maintains the usage order of the nodes.
- Maintain `self.head` as the most recently used node and `self.tail` as the least recently used node.
- Each `Node` stores the `key`, `val`, `next`, and `prev` attributes. Storing the key allows the cache entry to be removed directly during eviction.

- **Move To Head Operation:**
  - If the node is already the head, return immediately because it is already the most recently used node.
  - If the node has a previous node, connect that previous node to the node's next node using `temp.prev.next = temp.next`.
  - If the node has a next node, update its previous pointer using `temp.next.prev = temp.prev`.
  - If the node is the tail, update `self.tail` to `temp.prev` because the node is being detached from the end.
  - Set `temp.prev = None` because the node will become the new head.
  - Connect the node to the current head using `temp.next = self.head`.
  - Update the previous pointer of the old head using `self.head.prev = temp`.
  - Finally, assign `self.head = temp` to mark the node as the most recently used.

- **Get Operation:**
  - Check whether the requested key exists in `self.cache`.
  - If the key does not exist, return `-1`.
  - Otherwise, retrieve its node directly from the Hash Map.
  - Call `move_to_head(temp)` because accessing the key makes it the most recently used node.
  - Return `temp.val` without searching the linked list.

- **Put Operation:**
  - First, check whether the key already exists in the Hash Map.
  - If it exists, retrieve the corresponding node and update its value using `temp.val = value`.
  - Move the updated node to the head because updating an existing key makes it the most recently used.
  - If the key does not exist, create a new node using `Node(key, value)`.
  - Before inserting the new node, check whether the cache has reached its capacity.
  - If the cache is full, remove the least recently used key from the Hash Map using `del self.cache[self.tail.key]`.
  - If the head and tail refer to the same node, the cache contains only one node. Set both `self.head` and `self.tail` to `None`.
  - Otherwise, move the tail pointer to the previous node and set the new tail's `next` pointer to `None`.
  - If the linked list is empty, set both `self.head` and `self.tail` to the new node.
  - Otherwise, connect the new node before the current head and update the head's previous pointer.
  - Update `self.head` to the new node and store the key-node pair in `self.cache`.

- **Why O(1) Operations Work:**
  - The Hash Map provides direct access to a node without searching the linked list.
  - The Doubly Linked List allows a node to be detached and reinserted using a constant number of pointer updates.
  - The tail always identifies the least recently used node, so eviction does not require traversal.

**Complexity:**

**Brute Force:** Time Complexity: O(N) per operation  Space Complexity: O(N)

**Optimal Approach:** Time Complexity: O(1) average per operation  Space Complexity: O(C)

- `N` = Number of elements currently in the cache
- `C` = Cache capacity

**Common Mistakes:**

- Forgetting to update both `next` and `prev` pointers when detaching a node.
- Forgetting to update the tail when moving the last node to the head.
- Forgetting to delete the evicted key from the Hash Map.
- Incorrectly handling the single-node case during eviction.
- Forgetting to move an existing node to the head after `get` or `put`.
- Forgetting to update the old head's `prev` pointer when inserting a new head.
- Removing the tail node without disconnecting its previous node's `next` pointer.
- Failing to maintain the invariant that the head is the most recently used node and the tail is the least recently used node.

**Recognition Pattern:**

- Design a cache with O(1) average-time lookup, insertion, updating, and eviction.
- Think: Hash Map → Direct Node Access → Doubly Linked List → Reordering → LRU Eviction.

**Key Observation:**

The Hash Map provides O(1) average-time access to nodes, while the Doubly Linked List maintains their usage order. Moving a node to the head marks it as recently used, and removing the tail evicts the least recently used node. Because every operation requires only a constant number of Hash Map operations and pointer updates, both `get` and `put` run in O(1) average time.

**Code:**

```python
class Node:
    def __init__(self, key, val):
        self.key, self.val = key, val
        self.next = None
        self.prev = None

class LRUCache:
    def __init__(self, capacity: int):
        self.cap = capacity
        self.cache = {}
        self.head = self.tail = None

    def move_to_head(self, temp):
        if temp == self.head:
            return
        if temp.prev:  # Detach the node from its current position
            temp.prev.next = temp.next
        if temp.next:
            temp.next.prev = temp.prev
        if temp == self.tail:  # If it was the tail, update the tail
            self.tail = temp.prev
        temp.prev = None  # Insert at the head
        temp.next = self.head
        self.head.prev = temp
        self.head = temp
        return

    def get(self, key: int) -> int:
        if key not in self.cache:
            return -1
        temp = self.cache[key]
        self.move_to_head(temp)
        return temp.val

    def put(self, key: int, value: int) -> None:
        if key in self.cache:
            temp = self.cache[key]
            temp.val = value
            self.move_to_head(temp)
            return
        temp = Node(key, value)
        if len(self.cache) == self.cap: # Evict the least recently used node if full
            del self.cache[self.tail.key]
            if self.head == self.tail:
                self.head = self.tail = None
            else:
                self.tail = self.tail.prev
                self.tail.next = None
        if self.head is None: # Insert the new node at the head
            self.head = self.tail = temp
        else:
            temp.next = self.head
            self.head.prev = temp
            self.head = temp
        self.cache[key] = temp
        return
```

#### LeetCode 2104 - Sum of Subarray Ranges

**Question:**

Given an integer array `nums`, the range of a subarray is the difference between its maximum and minimum element.

Return the sum of the ranges of all contiguous subarrays of `nums`.

**Example:**

- Input: `nums = [1,2,3]`
- Output: `4`
- Explanation: The subarrays are `[1]`, `[2]`, `[3]`, `[1,2]`, `[2,3]`, and `[1,2,3]`. Their ranges are `0`, `0`, `0`, `1`, `1`, and `2`, respectively. The total is `4`.

**Algorithm:**

**Brute Force:**

- Generate every possible contiguous subarray by fixing a starting index `i`.
- Initialize the minimum and maximum values to `nums[i]`.
- Extend the subarray by moving the ending index `j` from `i` to the end of the array.
- Update the minimum using `min_min = min(min_min, nums[j])` and the maximum using `max_max = max(max_max, nums[j])`.
- Calculate the range of the current subarray as `max_max - min_min`.
- Add the range to the total sum.
- Repeat for every starting index and return the total.
- This approach takes O(N²) time because there are O(N²) contiguous subarrays, and each range can be updated in O(1) time as the subarray grows.

**Optimal Approach:**

- Use two Monotonic Stacks to calculate the sum of subarray minimums and the sum of subarray maximums independently.
- Maintain `stack1` for minimum contributions and `stack2` for maximum contributions.
- Store indices instead of values because indices are required to calculate how many subarrays contain each element as their minimum or maximum.
- Initialize `sum_min = 0` and `sum_max = 0`.
- Traverse the array using `range(len(nums) + 1)`. The extra iteration acts as a sentinel boundary that processes the remaining elements in both stacks.
- For minimum contributions, use a monotonic increasing stack. While `stack1` is not empty and the current element is smaller than or equal to the element at the top index, pop that index into `temp`.
- The popped element's next smaller-or-equal boundary is the current index `i`.
- After popping, the new stack top gives the previous smaller element's index. If the stack is empty, use `-1`.
- Calculate the number of subarrays where `nums[temp]` is the minimum:

  `Count = (temp - left) × (right - temp)`

- The distance `temp - left` gives the number of possible starting positions, while `right - temp` gives the number of possible ending positions.
- Calculate the element's contribution as `count * nums[temp]` and add it to `sum_min`.
- For maximum contributions, use a monotonic decreasing stack. While `stack2` is not empty and the current element is greater than or equal to the element at the top index, pop that index into `temp`.
- The popped element's next greater-or-equal boundary is the current index `i`.
- After popping, the new stack top gives the previous greater element's index. If the stack is empty, use `-1`.
- Calculate the number of subarrays where `nums[temp]` is the maximum using the same formula:

  `Count = (temp - left) × (right - temp)`

- Multiply this count by `nums[temp]` and add the contribution to `sum_max`.
- Continue popping from each stack while its corresponding condition holds because multiple elements can reach their boundaries at the same index.
- Push the current index onto both stacks after processing the contributions.
- At the sentinel iteration, the conditions short-circuit before accessing `nums[i]`, so all remaining indices are processed safely.
- Finally, return `sum_max - sum_min`, which equals the sum of the ranges of all contiguous subarrays.

**Complexity:**

**Brute Force:** Time Complexity: O(N²)  Space Complexity: O(1) auxiliary space

**Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(N)

**Common Mistakes:**

- Calculating only the minimum contributions or only the maximum contributions.
- Using element values instead of indices in the stacks.
- Using incorrect comparison conditions when popping from the stacks.
- Forgetting to calculate the number of subarrays using `(temp - left) × (right - temp)`.
- Forgetting to process the remaining elements using the sentinel boundary.
- Accessing `nums[i]` when `i == len(nums)` without checking the sentinel condition first.
- Returning `sum_min - sum_max` instead of `sum_max - sum_min`.
- Mishandling duplicate elements by using inconsistent comparison conditions.

**Recognition Pattern:**

- Find the sum of the maximum-minus-minimum difference across all contiguous subarrays.
- Think: Monotonic Increasing Stack → Subarray Minimums + Monotonic Decreasing Stack → Subarray Maximums → Difference of Contributions.

**Key Observation:**

Instead of generating every subarray, calculate how many subarrays have each element as their minimum and maximum. Monotonic stacks determine the boundaries and contribution of each element in O(N) time. Subtracting the sum of subarray minimums from the sum of subarray maximums gives the required answer.

**Code:**

```python
class Solution:
    def subArrayRanges(self, nums: list[int]) -> int:
        stack1,stack2 = [],[]
        sum_min,sum_max = 0,0
        for i in range(len(nums) + 1):
            while stack1 and (i == len(nums) or nums[stack1[-1]] >= nums[i]):
                temp = stack1.pop()
                left = stack1[-1] if stack1 else -1
                right = i
                count = (temp - left)*(right - temp) 
                sum_min += (count * nums[temp])
            while stack2 and (i == len(nums) or nums[stack2[-1]] <= nums[i]):
                temp = stack2.pop()
                left = stack2[-1] if stack2 else -1
                right = i
                count = (temp - left)*(right - temp) 
                sum_max += (count * nums[temp])  
            stack1.append(i)
            stack2.append(i)
        return sum_max - sum_min
```

#### LeetCode 901 - Online Stock Span

**Question:**

Design a `StockSpanner` class that calculates the stock span for each day's price.

The stock span is the number of consecutive days, ending on the current day, for which the stock price was less than or equal to the current day's price.

**Example:**

- Input:
  - `next(100)`
  - `next(80)`
  - `next(60)`
  - `next(70)`
  - `next(60)`
  - `next(75)`
  - `next(85)`
- Output: `[1, 1, 1, 2, 1, 4, 6]`
- Explanation:
  - Price `100` → Span = `1`
  - Price `80` → Span = `1` because `100` is greater than `80`.
  - Price `60` → Span = `1` because `80` is greater than `60`.
  - Price `70` → Span = `2` because `60` and `70` are consecutive days with prices less than or equal to `70`.
  - Price `60` → Span = `1` because `70` is greater than `60`.
  - Price `75` → Span = `4` because `60`, `70`, `60`, and `75` are consecutive prices less than or equal to `75`.
  - Price `85` → Span = `6` because the previous six prices are less than or equal to `85`.

**Algorithm:**

- Use a **monotonic decreasing stack** to calculate the stock span efficiently.
- Store each entry as `[price, index]`, where:
  - `price` is the stock price for that day.
  - `index` is the day's zero-based index.
- Maintain the stack so that prices are in strictly decreasing order from bottom to top.
- Use `self.index` to track the current day.

**Brute Force:**

- For each new stock price, start from the current day and move backward through all previous prices.
- Count consecutive days whose prices are less than or equal to the current price.
- Stop when a previous price is greater than the current price or when there are no previous days left.
- Return the total count as the stock span.

**Optimal Approach:**

- **Step 1: Initialize the stack and index**
  - `self.stack = []` stores `[price, index]` pairs for potential previous greater prices.
  - `self.index = 0` tracks the current day's zero-based index.

- **Step 2: Remove smaller or equal prices**
  - While the stack is not empty and `self.stack[-1][0] <= price`, pop the top element.
  - These previous prices are less than or equal to the current price, so they cannot prevent the current price from spanning across them.
  - Removing them allows the current price to inherit their span indirectly through the remaining previous greater price.

- **Step 3: Calculate the span**
  - If the stack is empty:
    - There is no previous price greater than the current price.
    - Therefore, the current price spans all days from day `0` through the current day.
    - Calculate the span using `self.index + 1`.
  - Otherwise:
    - The top element represents the nearest previous day with a price strictly greater than the current price.
    - Calculate the span using `self.index - self.stack[-1][1]`.
    - This counts all days after that previous greater price, including the current day.

- **Step 4: Push the current price**
  - Append `[price, self.index]` to the stack.
  - The current price becomes a potential previous greater price for future days.

- **Step 5: Move to the next day**
  - Increment `self.index` by `1`.
  - Return the calculated span.

**Complexity:**

- **Brute Force:** Time Complexity: O(N²) in the worst case  Space Complexity: O(1) auxiliary space, excluding stored input history.
- **Optimal Approach:** Time Complexity: O(1) amortized per `next()` call and O(N) total for N calls  Space Complexity: O(N) for the stack.
- Each price is pushed onto the stack once and popped at most once, so the total number of stack operations across N calls is O(N).

**Common Mistakes:**

- Using `self.stack[-1][0] < price` instead of `self.stack[-1][0] <= price` leaves equal prices on the stack, even though equal prices must be included in the span.
- Using `self.index - self.stack[-1][1] + 1` when the stack is non-empty incorrectly adds one to the span. The current formula already counts the current day.
- Returning `self.index` when the stack is empty misses the current day. Use `self.index + 1`.
- Storing only prices instead of both prices and indices makes it difficult to calculate the distance to the nearest previous greater price.
- Forgetting to increment `self.index` causes incorrect spans for subsequent calls.
- Forgetting that the stack stores `[price, index]` pairs can lead to accessing the wrong element when checking prices or calculating spans.

**Recognition Pattern:**

- Problems involving consecutive previous elements satisfying a comparison condition.
- Problems that ask for the distance to the nearest previous greater element.
- Problems involving stock spans, next or previous greater elements, and monotonic stacks.
- Use a monotonic decreasing stack when smaller or equal elements can be removed permanently to find the nearest previous strictly greater element.

**Key Observation:**

- The stock span is determined by the nearest previous price that is strictly greater than the current price.
- Every price less than or equal to the current price is popped because it cannot act as the blocking price for the current span.
- If no greater price remains, the span is `self.index + 1`; otherwise, it is the difference between the current index and the nearest previous greater price's index.
- The monotonic stack avoids repeatedly scanning previous days, giving O(N) total time across all calls.

**Code:**

```python
class StockSpanner:
    def __init__(self):
        self.stack = []
        self.index = 0
        
    def next(self, price: int) -> int:
        while self.stack and self.stack[-1][0] <= price:
            self.stack.pop()
        if not self.stack:
            temp = self.index + 1
        else:
            temp = self.index - self.stack[-1][1]
        self.stack.append([price,self.index])
        self.index += 1
        return temp
        
# Your StockSpanner object will be instantiated and called as such:
# obj = StockSpanner()
# param_1 = obj.next(price)
```

#### LeetCode 239 - Sliding Window Maximum

**Question:**

Given an integer array `nums` and an integer `k`, find the maximum value in every contiguous sliding window of size `k`.

**Example:**

- Input: `nums = [1,3,-1,-3,5,3,6,7]`, `k = 3`
- Output: `[3,3,5,5,6,7]`
- Explanation:
  - Window `[1,3,-1]` → Maximum = `3`
  - Window `[3,-1,-3]` → Maximum = `3`
  - Window `[-1,-3,5]` → Maximum = `5`
  - Window `[-3,5,3]` → Maximum = `5`
  - Window `[5,3,6]` → Maximum = `6`
  - Window `[3,6,7]` → Maximum = `7`

**Algorithm:**

- Use a **monotonic decreasing deque** to efficiently find the maximum of each sliding window.
- Store **indices** in the deque rather than the actual values. This allows us to identify and remove indices that fall outside the current window.
- The indices in the deque are maintained in increasing order, while their corresponding values are maintained in decreasing order.

**Optimal Approach:**

- Initialize an empty `result` list to store the maximum of every window.
- Initialize a deque named `queue` to store indices of potential maximum elements.

- Iterate through every index `i` in `nums`.

- **Step 1: Remove indices outside the current window**
  - The current window starts at index `i - k + 1`.
  - If `queue[0] < i - k + 1`, remove the front index using `popleft()`.
  - This ensures that the deque contains only indices belonging to the current window.

- **Step 2: Maintain the decreasing monotonic deque**
  - While the deque is not empty and `nums[i] >= nums[queue[-1]]`, remove the index at the back using `pop()`.
  - These elements are smaller than or equal to the current element.
  - Since the current element is newer and at least as large, the removed elements can never be the maximum of any future window containing the current element.
  - This maintains decreasing values from the front to the back of the deque.

- **Step 3: Insert the current index**
  - Append index `i` to the back of the deque.
  - The front of the deque now contains the index of the maximum element in the current window, once the window reaches size `k`.

- **Step 4: Record the maximum**
  - If `i + 1 < k`, the first complete window has not formed yet, so continue to the next iteration.
  - Otherwise, append `nums[queue[0]]` to `result`.
  - The front index points to the maximum value in the current window.

- Return `result` after processing all elements.

**Brute Force:**

- Iterate through every possible window of size `k`.
- For each window, find its maximum using `max(nums[i:i+k])`.
- Append the maximum to the result list.
- This approach repeatedly examines overlapping elements, resulting in a time complexity of O(N × K).

**Complexity:**

- **Brute Force:** Time Complexity: O(N × K)  Space Complexity: O(N) for the result, excluding the temporary window.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(K) auxiliary space for the deque, excluding the result.
- Each index is added to the deque once and removed at most once, so the total deque operations across the entire array are O(N).

**Common Mistakes:**

- Storing values instead of indices makes it difficult to determine whether an element has moved outside the current window.
- Using `queue[0] <= i - k + 1` for expiration incorrectly removes an index that is still inside the window. The correct condition is `queue[0] < i - k + 1`.
- Forgetting to remove smaller elements from the back breaks the monotonic property and can cause unnecessary elements to remain in the deque.
- Appending a maximum before the first complete window forms produces extra results. The condition `if i + 1 < k: continue` prevents this.
- Using `pop()` at the front instead of `popleft()` is incorrect for a deque because `pop()` removes the last element.
- Forgetting that the deque stores indices means `nums[queue[0]]` is required to retrieve the maximum value.

**Recognition Pattern:**

- Problems involving the maximum or minimum of every fixed-size sliding window.
- Problems where repeatedly scanning each window would be inefficient.
- Problems where elements can be discarded permanently because a newer element dominates them.
- Use a monotonic deque when you need both efficient expiration of old elements and efficient removal of dominated elements.

**Key Observation:**

- If a newer element is greater than or equal to an older element, the older element can never be the maximum in any future window that contains the newer element.
- Therefore, removing dominated elements from the back preserves all possible maximum candidates.
- The front of the deque always holds the index of the current window's maximum.
- Each element enters and leaves the deque at most once, giving an overall O(N) solution.

**Code:**

```python
class Solution:
    def maxSlidingWindow(self, nums: list[int], k: int) -> list[int]:
        result = []
        queue = collections.deque()
        for i in range(len(nums)):
            if queue and queue[0] < i - k + 1:
                queue.popleft()
            while queue and nums[i] >= nums[queue[-1]]:
                queue.pop()
            queue.append(i)
            if i + 1 < k:
                continue
            result.append(nums[queue[0]])
        return result
```
