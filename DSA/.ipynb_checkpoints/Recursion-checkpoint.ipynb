#### LeetCode 50 - Pow(x, n)

**Question:**

- Implement `myPow(x, n)`, which calculates \(x^n\).
- Handle positive, zero, and negative exponents.
- The solution must efficiently handle very large values of `n`.

**Example:**

- Input: `x = 2.00000, n = 10`
- Output: `1024.00000`
- Explanation: \(2^{10} = 1024\)

- Input: `x = 2.10000, n = 3`
- Output: `9.26100`
- Explanation: \(2.1^3 = 9.261\)

- Input: `x = 2.00000, n = -2`
- Output: `0.25000`
- Explanation: \(2^{-2} = \frac{1}{2^2} = 0.25\)

**Algorithm:**

**Brute Force:**

- Initialize `result = 1`.
- If `n > 0`, multiply `result` by `x` exactly `n` times.
- If `n == 0`, return `1`.
- If `n < 0`, multiply by `x` exactly `abs(n)` times and return the reciprocal.
- This approach performs one multiplication for every unit of the exponent.

**Optimal Approach:**

- Use **Binary Exponentiation (Fast Power)** with recursion and repeated squaring.
- If `n < 0`, calculate the positive power and return its reciprocal: `1 / myPow(x, -n)`.
- If `n == 0`, return `1`, because any nonzero number raised to zero equals one.
- If `n == 1`, return `x`, because any number raised to one equals itself.
- Recursively calculate `half = myPow(x, n // 2)`.
- If `n` is even, return `half * half`.
- If `n` is odd, return `x * half * half`.
- Each recursive call approximately halves the exponent, reducing the number of operations significantly.

**Complexity:**

- Brute Force Time: O(|n|)
- Brute Force Space: O(1) auxiliary space.
- Optimal Time: O(log |n|)
- Optimal Space: O(log |n|) due to the recursion call stack.

**Common Mistakes:**

- Using repeated multiplication for very large exponents, resulting in O(|n|) time.
- Forgetting that a negative exponent requires taking the reciprocal.
- Forgetting the base case `n == 0`, which must return `1`.
- Using `n / 2` instead of `n // 2`, which can convert the exponent to a float in Python 3.
- Using the wrong rule for odd exponents. The correct rule is \(x^n = x \times (x^{\lfloor n/2 \rfloor})^2\) for positive odd `n`.
- Assuming the recursive method uses O(1) space. Each recursive call occupies stack space, so the space complexity is O(log |n|).
- Not considering the special case `x == 0` with a negative exponent, which is mathematically undefined.

**Recognition Pattern:**

- The problem involves calculating a power with a potentially huge exponent.
- Repeated multiplication is too slow.
- The exponent can be repeatedly divided by two.
- Recognize **Binary Exponentiation / Fast Power / Repeated Squaring**.

**Key Observation:**

- For an even exponent: \(x^n = (x^{n/2})^2\).
- For an odd positive exponent: \(x^n = x \times (x^{\lfloor n/2 \rfloor})^2\).
- For a negative exponent: \(x^{-n} = \frac{1}{x^n}\).
- Example: \(2^{16} = (2^8)^2 = ((2^4)^2)^2\).
- The exponent decreases as `n → n/2 → n/4 → n/8`, giving O(log |n|) time.
- The iterative version can achieve O(log |n|) time and O(1) auxiliary space, while your recursive version uses O(log |n|) stack space.

**Code:**

```python
#Optimal (Recursion Method)
class Solution(object):
    def myPow(self, x, n):
        """
        :type x: float
        :type n: int
        :rtype: float
        """
        if n < 0:
            return 1 / (self.myPow(x,-(n)))
        if n == 1:
            return x
        if n == 0:
            return 1
        half = self.myPow(x,n//2)
        if n % 2 == 0:
            return half * half
        else:
            return x * half * half
             
# Brute Force
class Solution(object):
    def myPow(self, x, n):1
        result = 1
        if n > 0:
            for i in range(n):
                result *= x
        elif n == 0:
            return 1
        else:
            y = -(n)
            for i in range(y):
                result *= x
            result = 1/result
        return result
```

#### LeetCode 1922 - Count Good Numbers

**Question:**

- A digit string of length `n` is considered good if the digits at even indices are even digits and the digits at odd indices are prime digits.
- Even indices can contain `0, 2, 4, 6, 8`, giving 5 choices.
- Odd indices can contain `2, 3, 5, 7`, giving 4 choices.
- Return the total number of good digit strings of length `n`, modulo \(10^9 + 7\).

**Example:**

- Input: `n = 1`
- Output: `5`
- Explanation: Only index `0` exists, so it can contain any of the 5 even digits.

- Input: `n = 4`
- Output: `400`
- Explanation: There are 2 even indices and 2 odd indices. The answer is \(5^2 \times 4^2 = 400\).

**Algorithm:**

**Brute Force:**

- Generate every possible digit string of length `n`.
- Check whether every even index contains an even digit and every odd index contains a prime digit.
- Count the strings that satisfy both conditions.
- This approach is infeasible for large `n` because it considers up to \(10^n\) strings.

**Optimal Approach:**

- Count the number of even and odd indices separately.
- If `n` is even:
  - `even_count = n // 2`
  - `odd_count = n // 2`
- If `n` is odd:
  - `even_count = n // 2 + 1`
  - `odd_count = n // 2`
- Every even index has 5 choices, and every odd index has 4 choices.
- Apply the multiplication principle: `result = POW(5, even_count) * POW(4, odd_count)`.
- Use recursive fast exponentiation to calculate powers efficiently.
- If the exponent is `0`, return `1`.
- If the exponent is `1`, return the base.
- If the exponent is odd, reduce it by one and multiply the base by the recursive result.
- If the exponent is even, square the base and halve the exponent.
- Use modulo \(10^9 + 7\) to keep intermediate values manageable.
- Return the final product modulo \(10^9 + 7\).

**Complexity:**

- Brute Force Time: O(10ⁿ × n), because each candidate string may require checking up to `n` positions.
- Brute Force Space: O(n) for storing a candidate string, excluding storage for generated candidates.
- Optimal Time: O(log n), because fast exponentiation is performed twice.
- Optimal Space: O(log n) due to the recursive call stack.

**Common Mistakes:**

- Incorrectly counting even and odd positions when `n` is odd. Index `0` is even, so there is one more even position.
- Using 4 choices for even indices or 5 choices for odd indices.
- Forgetting that the number of combinations is the product of the choices at each position.
- Forgetting to apply modulo to the final product.
- Applying modulo inconsistently during recursive exponentiation.
- In the even-exponent branch, your code squares `base` but does not immediately apply modulo to the squared value. This is still mathematically correct because the recursive calls eventually apply modulo, but applying modulo to the squared base keeps intermediate values smaller.
- In the odd-exponent branch, reducing the result modulo \(10^9 + 7\) is important to control intermediate values.
- Your `POW` method is logically correct for nonnegative exponents, but its modulo handling can be made more consistent.

**Recognition Pattern:**

- The problem asks for the number of strings satisfying independent conditions at each position.
- Each position has a fixed number of valid choices.
- The total number of combinations is the product of the choices at all positions.
- The exponent can be extremely large, making ordinary repeated multiplication inefficient.
- Recognize **Combinatorics + Modular Exponentiation + Binary Exponentiation**.

**Key Observation:**

- For a string of length `n`, the number of even indices is \(\lceil n/2 \rceil\), and the number of odd indices is \(\lfloor n/2 \rfloor\).
- Therefore, the answer is:

  \[
  5^{\lceil n/2 \rceil} \times 4^{\lfloor n/2 \rfloor}
  \pmod{10^9+7}
  \]

- When `n` is odd, index `0` contributes an additional factor of `5`.
- Fast exponentiation uses the identities:
  - Even exponent: \(a^n = (a^2)^{n/2}\).
  - Odd exponent: \(a^n = a \times a^{n-1}\).
- **Important optimization:** In the even branch, return `self.POW((base * base) % MOD, expo // 2)` to keep the recursive base reduced modulo `MOD`.
- The result can be computed in O(log n) time without generating any strings.

**Code:**

```python
class Solution:
    def POW(self,base,expo):
        if expo == 0:
            return 1
        if expo == 1:
            return base
        if expo % 2 == 1:
            return (base * self.POW(base,expo - 1)) % (10**9 + 7)
        else:
            base = base % (10**9 + 7)
            return (self.POW(base*base,expo // 2))

    def countGoodNumbers(self, n: int) -> int:
        result = 1
        even_count,odd_count = 0,0
        if n % 2 == 0:
            even_count,odd_count = n // 2, n // 2
        else:
            even_count,odd_count = n // 2 + 1, n // 2
        result = self.POW(5,even_count) * self.POW(4,odd_count)
        return result % (10**9 + 7)
```

#### LeetCode 22 - Generate Parentheses

**Question:**

- Given an integer `n`, generate all combinations of `n` pairs of well-formed parentheses.
- Every opening parenthesis must have a corresponding closing parenthesis.
- A closing parenthesis cannot appear before its matching opening parenthesis.
- Return all valid combinations.

**Example:**

- Input: `n = 3`
- Output: `["((()))", "(()())", "(())()", "()(())", "()()()"]`
- Explanation: There are 5 valid combinations of 3 pairs of parentheses.

**Algorithm:**

**Brute Force:**

- Generate every possible string of length `2 * n` using `(` and `)`.
- For each string, check whether the number of opening and closing parentheses is equal.
- Scan the string from left to right, ensuring the number of closing parentheses never exceeds the number of opening parentheses.
- Add valid strings to the result.
- This approach generates \(2^{2n} = 4^n\) possible strings before filtering, making it inefficient.

**Optimal Approach:**

- Use **Backtracking** to construct only valid parenthesis combinations.
- Initialize an empty `result` list to store valid combinations.
- Initialize an empty `stack` list to build the current combination.
- Define `backtrack(open_par, close_par)` to track the number of opening and closing parentheses used.
- If `open_par == close_par == n`, a valid combination is complete:
  - Join the characters in `stack` into a string.
  - Append the string to `result`.
  - Return from the recursive call.
- If `open_par < n`:
  - Append `(` to `stack`.
  - Recursively call `backtrack(open_par + 1, close_par)`.
  - Remove the last character using `stack.pop()` to backtrack.
- If `close_par < open_par`:
  - Append `)` to `stack`.
  - Recursively call `backtrack(open_par, close_par + 1)`.
  - Remove the last character using `stack.pop()` to backtrack.
- Start the recursion with `backtrack(0, 0)`.
- Return `result`.

**Complexity:**

- Brute Force Time: O(2²ⁿ × n), because there are \(2^{2n}\) candidate strings and each may take O(n) time to validate.
- Brute Force Space: O(n) auxiliary space for generating and checking one candidate, excluding stored results.
- Optimal Time: O(Cₙ × n), where \(C_n = \frac{1}{n+1}\binom{2n}{n}\) is the nth Catalan number. Each valid combination takes O(n) time to construct.
- Optimal Auxiliary Space: O(n) for the recursion stack and the current combination, excluding the output.
- Output Space: O(Cₙ × n) to store all valid combinations.

**Common Mistakes:**

- Allowing more than `n` opening parentheses.
- Adding a closing parenthesis when `close_par >= open_par`, which can create an invalid prefix.
- Forgetting the base case where both counters equal `n`.
- Forgetting `stack.pop()` after each recursive call, which causes characters from one branch to affect another.
- Appending `stack` directly to `result` instead of `"".join(stack)`. Since `stack` is modified during backtracking, store a completed string.
- Using `close_par < n` as the only condition for adding `)`. This does not prevent a closing parenthesis from appearing before an opening parenthesis.
- Generating all possible strings first and filtering them instead of pruning invalid branches during recursion.

**Recognition Pattern:**

- The problem asks for all valid combinations rather than a single answer.
- Each decision changes the current partial solution.
- Some choices are allowed only when specific constraints are satisfied.
- Invalid partial solutions can be avoided before completing them.
- Recognize **Backtracking + Constraint Pruning + Catalan Numbers**.

**Key Observation:**

- A valid combination must contain exactly `n` opening and `n` closing parentheses.
- At every point in the string, the number of closing parentheses must not exceed the number of opening parentheses.
- Therefore:
  - Add `(` only when `open_par < n`.
  - Add `)` only when `close_par < open_par`.
- These two conditions guarantee that every generated combination is valid, so no separate validation step is required.
- `stack.append()` makes a choice, the recursive call explores it, and `stack.pop()` undoes the choice.
- Backtracking explores all valid paths while skipping invalid ones.

**Code:**

```python
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        result = []
        stack = []
        def backtrack(open_par,close_par):
            if open_par == close_par ==  n:
                result.append("".join(stack))
                return 
            if open_par < n:
                stack.append("(")
                backtrack(open_par + 1, close_par)
                stack.pop()
            if close_par < open_par:
                stack.append(")")
                backtrack(open_par, close_par + 1)
                stack.pop()
        backtrack(0,0)
        return result
```

#### LeetCode 39 - Combination Sum

**Question:**

- Given an array of distinct positive integers `candidates` and an integer `target`, find all unique combinations of candidates whose sum equals `target`.
- You may choose the same candidate an unlimited number of times.
- Return all valid combinations in any order.

**Example:**

- Input: `candidates = [2, 3, 6, 7], target = 7`
- Output: `[[2, 2, 3], [7]]`
- Explanation:
  - `2 + 2 + 3 = 7`
  - `7 = 7`
  - Each candidate can be reused multiple times.

**Algorithm:**

**Brute Force:**

- Generate possible combinations of candidates, allowing repeated selections.
- Calculate the sum of each combination.
- If the sum equals `target`, add the combination to `result`.
- Discard combinations whose sum exceeds `target`.
- This approach can explore many unnecessary combinations without systematically avoiding duplicate search paths.

**Optimal Approach:**

- Use **Backtracking** to explore combinations while tracking the remaining target.
- Initialize an empty `result` list to store valid combinations.
- Define `calculateSum(i, aim, stack)`:
  - `i` represents the current candidate index.
  - `aim` represents the remaining sum needed.
  - `stack` stores the current combination.
- If `aim == 0`:
  - A valid combination has been found.
  - Append `list(stack)` to `result` to store an independent copy.
  - Return from the recursive call.
- If `aim < 0`, return because the current combination exceeds the target.
- **Include the current candidate:**
  - Append `candidates[i]` to `stack`.
  - Recursively call `calculateSum(i, aim - candidates[i], stack)`.
  - Keep the same index `i` because the candidate can be reused.
  - Remove the candidate using `stack.pop()` to backtrack.
- **Exclude the current candidate:**
  - Recursively call `calculateSum(i + 1, aim, stack)` to move to the next candidate.
- Start with `calculateSum(0, target, [])`.
- Return `result`.

**Complexity:**

- Brute Force Time: Exponential, because many candidate combinations may need to be explored.
- Brute Force Space: Depends on the number and length of generated combinations.
- Optimal Time: Exponential in the worst case. For positive integer candidates, the search tree can be large; a commonly used upper-bound style estimate is O(Nᵀ), where `N` is the number of candidates and `T` is the target when candidates are positive integers. Actual complexity depends on candidate values and the number of valid combinations.
- Optimal Auxiliary Space: O(T / m), where `m` is the smallest candidate, for the recursion depth and current combination.
- Output Space: Depends on the number of valid combinations and their lengths.

**Common Mistakes:**

- Calling `calculateSum(i + 1, ...)` after including a candidate. This would prevent reusing that candidate.
- Forgetting `stack.pop()` after the include branch, causing the current combination to contain choices from previous recursive paths.
- Appending `stack` directly to `result` instead of `list(stack)`. The stack is modified later, so a copy is required.
- Continuing when `aim < 0`, which wastes time exploring an invalid combination.
- Returning immediately after excluding a candidate, which can prevent other combinations from being explored.
- Advancing the index after both branches. The include branch must retain the same index to allow repeated use.
- Not handling `aim == 0` as a terminal condition, which can lead to unnecessary recursion or invalid results.

**Recognition Pattern:**

- The problem asks for all combinations that satisfy a target sum.
- Candidates can be reused an unlimited number of times.
- Each decision has two possibilities: include the current candidate or skip it.
- Invalid paths can be pruned when the remaining target becomes negative.
- Recognize **Backtracking + Include/Exclude Recursion + Pruning**.

**Key Observation:**

- `aim` tracks how much of the target remains.
- If `aim == 0`, the current combination is valid.
- If `aim < 0`, the current path cannot produce a valid combination because all candidates are positive.
- The include branch keeps the same index: `calculateSum(i, aim - candidates[i], stack)`.
- The exclude branch advances the index: `calculateSum(i + 1, aim, stack)`.
- Keeping the index unchanged for inclusion permits unlimited reuse, while advancing it for exclusion avoids generating the same combination in different orders.
- The include/exclude structure explores all valid combinations without needing to sort the candidates.

**Code:**

```python
class Solution:
    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        result = []
        def calculateSum(i,aim,stack):
            if aim == 0:
                result.append(list(stack))
                return
            if aim < 0:
                return
            if i < len(candidates) and aim >= 0:
                stack.append(candidates[i])
                calculateSum(i,aim - candidates[i],stack)
                stack.pop()
            if i != len(candidates):
                calculateSum(i + 1,aim,stack)
        calculateSum(0,target,[])
        return result
```

#### LeetCode 40 - Combination Sum II

**Question:**

- Given a collection of candidate numbers `candidates` and a target integer `target`, find all unique combinations where the chosen numbers sum to `target`.
- Each number in `candidates` may be used at most once in each combination.
- The solution must not contain duplicate combinations, even if `candidates` contains duplicate values.
- Return all unique combinations in any order.

**Example:**

- Input: `candidates = [10, 1, 2, 7, 6, 1, 5], target = 8`
- Output: `[[1, 1, 6], [1, 2, 5], [1, 7], [2, 6]]`
- Explanation:
  - `1 + 1 + 6 = 8`
  - `1 + 2 + 5 = 8`
  - `1 + 7 = 8`
  - `2 + 6 = 8`
  - Each array element can be used only once per combination, and duplicate combinations are avoided.

**Algorithm:**

**Brute Force:**

- Generate every possible subset of `candidates`.
- Calculate the sum of each subset.
- If the sum equals `target`, add the subset to the result.
- Use a set or another deduplication method to remove duplicate combinations.
- This approach explores up to \(2^n\) subsets and may generate many duplicate combinations.

**Optimal Approach:**

- Use **Backtracking with Sorting and Duplicate Skipping**.
- Initialize an empty `result` list.
- Sort `candidates` so duplicate values are adjacent and larger values can be used to prune the search.
- Define `calculateResult(i, aim, stack)`:
  - `i` is the starting index for the current recursive call.
  - `aim` is the remaining target.
  - `stack` stores the current combination.
- If `aim == 0`:
  - A valid combination is found.
  - Append `list(stack)` to `result`.
  - Return.
- If `aim < 0`, return because the current combination exceeds the target.
- Iterate through candidates using `for j in range(i, len(candidates))`.
- Skip duplicate values at the same recursion depth using `if j > i and candidates[j] == candidates[j - 1]: continue`.
- If `candidates[j] > aim`, break because the candidates are sorted and later values will be even larger.
- Append `candidates[j]` to `stack`.
- Recursively call `calculateResult(j + 1, aim - candidates[j], stack)`.
- Use `j + 1` so the same array element cannot be reused.
- Remove the last element using `stack.pop()` to backtrack.
- Start with `calculateResult(0, target, [])`.
- Return `result`.

**Complexity:**

- Brute Force Time: O(2ⁿ × n) if every subset is generated and its sum is calculated directly.
- Brute Force Space: O(n) auxiliary space for a subset, excluding stored results.
- Optimal Time: O(2ⁿ × n) as a conservative bound including combination copying; actual work depends on pruning and duplicate values.
- Optimal Auxiliary Space: O(n) for the recursion stack and current combination, excluding output.
- Output Space: Depends on the number of unique valid combinations and their lengths.

- Using `candidates[j]` in the recursive call without advancing the index. This allows the same element to be reused.
- Forgetting to sort `candidates` before skipping duplicates or using the early-break optimization.
- Using `if j > 0` instead of `if j > i` for duplicate skipping. The check must be relative to the current recursion depth.
- Forgetting `stack.pop()` after the recursive call.
- Appending `stack` directly to `result` instead of `list(stack)`.
- Using `arr[i] > target` instead of `candidates[j] > aim`. In your code, `arr` is undefined, and the check should use the current candidate and remaining target.
- Comparing against the original `target` rather than the remaining `aim`, which can prevent correct pruning.
- Breaking when a candidate exceeds `aim` only works because the candidates are sorted.
- Forgetting that duplicates are skipped at the same recursion depth, not across every level. Repeated values can still be used in a combination if they come from different array positions.

**Recognition Pattern:**

- The problem asks for unique combinations that sum to a target.
- Each candidate can be used only once.
- The input may contain duplicate values.
- Sorting enables duplicate skipping and early termination.
- Recognize **Backtracking + Sorting + Duplicate Pruning + Single-Use Elements**.

- `if j > i and candidates[j] == candidates[j - 1]: continue` prevents selecting the same value as the first choice multiple times at the same recursion depth.
- Sorting allows the condition `if candidates[j] > aim: break`, because all later candidates will also exceed the remaining target.
- The corrected pruning condition is `if candidates[j] > aim: break`.
- Unlike Combination Sum I, the recursive call advances to `j + 1` because elements cannot be reused.
- Duplicate values are allowed in the input, but duplicate combinations must not appear in the output.

#### LeetCode 78 - Subsets

**Question:**

- Given an integer array `nums` containing unique elements, return all possible subsets (the power set).
- The solution must not contain duplicate subsets.
- The order of the subsets in the output does not matter.
- An empty subset and the original array must both be included.

**Example:**

- Input: `nums = [1, 2, 3]`
- Output: `[[], [1], [2], [1, 2], [3], [1, 3], [2, 3], [1, 2, 3]]`
- Explanation: Every element can either be included in or excluded from a subset. With 3 elements, there are \(2^3 = 8\) possible subsets.

**Algorithm:**

**Brute Force:**

- Each element has two choices: include it in the subset or exclude it.
- Use recursion to explore both choices for every element.
- Maintain a temporary list `stack` to store the current subset.
- Include the current element and recursively process the next index.
- Remove the element using `stack.pop()` to backtrack.
- Recursively process the next index without including the element.
- When `i == len(nums)`, append a copy of `stack` to `result`.
- Return all generated subsets.

**Optimal Approach:**

- Use **Bit Manipulation** to represent every subset as a binary number.
- Let `n = len(nums)`.
- Calculate the total number of subsets using `total_subset = 1 << n`, which equals \(2^n\).
- Iterate through every integer from `0` to `total_subset - 1`.
- Treat each integer's binary representation as a selection mask for the elements.
- For each number, initialize an empty list `temp`.
- Iterate through every bit position `i` from `0` to `n - 1`.
- Check whether the `i`th bit is set using `num & (1 << i)`.
- If the bit is nonzero, append `nums[i]` to `temp`.
- After checking all bit positions, append `temp` to `result`.
- Return `result`.

**Complexity:**

- Brute Force Time: O(n × 2ⁿ), because there are \(2^n\) subsets and each subset may require up to O(n) work to copy.
- Brute Force Auxiliary Space: O(n) for the recursion stack and current subset, excluding output.
- Bit Manipulation Time: O(n × 2ⁿ), because every one of the \(2^n\) masks checks all `n` bit positions.
- Bit Manipulation Auxiliary Space: O(n) for the temporary subset, excluding output.
- Output Space: O(n × 2ⁿ), because all \(2^n\) subsets are stored and each can contain up to `n` elements.

**Common Mistakes:**

- Using `n ^ 2` instead of `1 << n`. The `^` operator performs bitwise XOR, not exponentiation.
- Iterating only up to `n` instead of `2^n`, which misses most subsets.
- Checking `num & (1 << i)` incorrectly by comparing it with `1`. The result can be a power of two other than `1`; checking whether it is nonzero is correct.
- Forgetting the empty subset, represented by the mask `0`.
- Forgetting to append each generated subset to `result`.
- Appending the same mutable list repeatedly in the recursive approach instead of appending `stack[:]` or `list(stack)`.
- Confusing a bit position with an array index. Bit `i` determines whether `nums[i]` is included.
- Assuming bit manipulation has O(1) total time. Generating all subsets necessarily requires exponential output.

**Recognition Pattern:**

- The problem asks for all possible subsets or combinations where every element can be included or excluded.
- There are exactly two choices for each element.
- The number of results is \(2^n\).
- Recognize **Power Set + Bitmasking + Binary Representation**.
- Recursion with include/exclude decisions is an alternative pattern.

**Key Observation:**

- Every element has two choices: include or exclude.
- For `n` elements, the total number of subsets is \(2^n\).
- The expression `1 << n` calculates \(2^n\) by shifting the binary value `1` left by `n` positions.
- Each bit represents one array element:
  - Bit `1`: include the corresponding element.
  - Bit `0`: exclude the corresponding element.
- Example: `nums = [1, 2, 3]`, mask `5` has binary representation `101`.
- The first and third bits are set, so the corresponding subset is `[1, 3]`.
- Mask `0` generates `[]`, while mask `2ⁿ - 1` generates the complete array.
- Both your bit manipulation solution and recursive solution generate the same power set, but the bit manipulation version uses binary masks instead of recursive include/exclude decisions.

**Code:**

```python
# Bit Manipulation Solution
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        result = []
        n = len(nums)
        total_subset = 1 << n
        for num in range(total_subset):
            temp = []
            for i in range(n):
                if num & (1 << i) != 0:
                    temp.append(nums[i])
            result.append(temp)
        return result
        
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        result = []
        def powerSet(stack,i):
            if i == len(nums):
                result.append(stack[:])
                return
            stack.append(nums[i])
            powerSet(stack,i + 1)
            stack.pop()
            powerSet(stack,i + 1)
        powerSet([],0)
        return result           
```

#### LeetCode 90 - Subsets II

**Question:**

- Given an integer array `nums` that may contain duplicates, return all possible subsets (the power set).
- The solution must not contain duplicate subsets.
- The order of the subsets in the output does not matter.
- Each element can be selected at most once in a subset.

**Example:**

- Input: `nums = [1, 2, 2]`
- Output: `[[], [1], [1, 2], [1, 2, 2], [2], [2, 2]]`
- Explanation: Although the array contains two occurrences of `2`, duplicate subsets must not appear in the result.

**Algorithm:**

**Brute Force:**

- Generate every possible subset using recursion or bit manipulation.
- For each element, choose whether to include or exclude it.
- Store every generated subset.
- Convert each subset into a tuple or another hashable representation and use a set to remove duplicate subsets.
- This approach generates up to \(2^n\) subsets before deduplication, so it can perform unnecessary work when duplicates are present.

**Optimal Approach:**

- Use **Backtracking with Sorting and Duplicate Skipping**.
- Sort `nums` so duplicate values appear next to each other.
- Define `powerSet(i, stack)`:
  - `i` is the current index.
  - `stack` stores the current subset.
- If `i == len(nums)`:
  - Append a copy of `stack` to `result`.
  - Return.
- **Include the current element:**
  - Append `nums[i]` to `stack`.
  - Recursively call `powerSet(i + 1, stack)`.
  - Remove the element using `stack.pop()` to backtrack.
- **Skip duplicate elements:**
  - While the next element exists and equals the current element, increment `i`.
  - This skips repeated values at the current recursion level when exploring the exclusion branch.
- **Exclude the current value:**
  - Call `powerSet(i + 1, stack)` to continue without selecting that value.
- Start with `powerSet(0, [])`.
- Return `result`.

**Complexity:**

- Brute Force Time: O(n × 2ⁿ) for generating and copying all subsets, plus deduplication overhead.
- Brute Force Auxiliary Space: O(n) for the recursion stack and current subset, excluding stored subsets.
- Optimal Time: O(n × 2ⁿ) in the worst case, including the work required to copy the generated subsets. Sorting takes O(n log n).
- Optimal Auxiliary Space: O(n) for the recursion stack and current subset, excluding output.
- Output Space: O(n × U), where `U` is the number of unique subsets produced.

**Common Mistakes:**

- Forgetting to sort `nums` before checking adjacent duplicates.
- Skipping duplicates before exploring the include branch, which can prevent valid subsets containing multiple copies of a value.
- Forgetting `stack.pop()` after exploring the include branch.
- Appending `stack` directly to `result` instead of `stack[:]`, which would store references to a list that keeps changing.
- Using a duplicate-skipping condition that accesses `nums[i + 1]` without checking that `i + 1 < len(nums)`.
- Skipping duplicates at every recursion level instead of only while exploring the exclusion branch. Multiple copies of the same value must still be selectable.
- Forgetting the base case, which should append the current subset exactly once when all elements have been processed.

**Recognition Pattern:**

- The problem asks for all possible subsets.
- The input contains duplicate values.
- Duplicate subsets must be removed.
- Sorting makes equal values adjacent.
- Recognize **Backtracking + Sorting + Duplicate Skipping**.

**Key Observation:**

- Sorting transforms `nums = [2, 1, 2]` into `[1, 2, 2]`.
- When choosing whether to include a value, the include branch advances by one index, allowing another occurrence of that value to be included.
- When excluding a value, the `while` loop skips all adjacent duplicates so the same subset is not generated again through equivalent exclusion choices.
- For example, `[2, 2]` is valid because the two `2`s come from different array positions.
- Your approach correctly handles duplicates by skipping repeated values only in the exclusion branch.
- The total number of unique subsets is at most \(2^n\), and when all elements are distinct, your algorithm generates exactly \(2^n\) subsets.

**Code:**

```python
class Solution:
    def subsetsWithDup(self, nums: List[int]) -> List[List[int]]:
        result = []
        def powerSet(i,stack):
            if i == len(nums):
                result.append(stack[:])
                return
            stack.append(nums[i]) 
            powerSet(i + 1,stack)
            stack.pop()
            while i + 1 < len(nums) and nums[i] == nums[i + 1]:
                i += 1
            powerSet(i + 1,stack)
        nums.sort()
        powerSet(0,[])
        return result
```

#### LeetCode 216 - Combination Sum III

**Question:**

- Find all valid combinations of `k` numbers that add up to `n`.
- Use only numbers from `1` to `9`.
- Each number can be used at most once.
- Return all valid combinations without duplicates.

**Example:**

- Input: `k = 3, n = 7`
- Output: `[[1, 2, 4]]`
- Explanation: The only combination of 3 distinct numbers from `1` to `9` that sums to `7` is `[1, 2, 4]`.

- Input: `k = 3, n = 9`
- Output: `[[1, 2, 6], [1, 3, 5], [2, 3, 4]]`
- Explanation: Each combination contains exactly 3 distinct numbers whose sum is `9`.

**Algorithm:**

**Brute Force:**

- Generate all subsets of numbers from `1` to `9`.
- Keep only subsets containing exactly `k` numbers.
- Calculate the sum of each subset.
- Add a subset to `result` if its sum equals `n`.
- This explores up to \(2^9\) subsets and checks their sizes and sums.

**Optimal Approach:**

- Use **Backtracking** to build combinations incrementally.
- Initialize an empty `result` list.
- Define `backtrack(i, total, stack)`:
  - `i` is the smallest number allowed for the next selection.
  - `total` is the sum of the numbers currently selected.
  - `stack` stores the current combination.
- If `len(stack) > k` or `total > n`, return because the current path cannot produce a valid combination.
- If `len(stack) == k` and `total == n`:
  - Append a copy of `stack` to `result`.
  - Return.
- Iterate through numbers from `i` to `9`.
- Calculate `temp = total + j`.
- Append `j` to `stack`.
- Recursively call `backtrack(j + 1, temp, stack)` so each number is used at most once and combinations are generated in increasing order.
- Remove `j` using `stack.pop()` to backtrack.
- Start with `backtrack(1, 0, [])`.
- Return `result`.

**Complexity:**

- Brute Force Time: O(9 × 2⁹), because each subset may require checking its size and sum.
- Brute Force Auxiliary Space: O(9), excluding the output.
- Optimal Time: O(9 × 2⁹) as a loose upper bound, since the search explores subsets of the nine available numbers. The actual search is smaller because of pruning.
- Optimal Auxiliary Space: O(9), because the recursion depth and current combination cannot exceed nine elements.
- Output Space: O(k × U), where `U` is the number of valid combinations.

**Common Mistakes:**

- Starting from `0` instead of `1`, even though the allowed numbers are `1` through `9`.
- Calling `backtrack(j, ...)` instead of `backtrack(j + 1, ...)`, which would allow the same number to be selected again.
- Forgetting `stack.pop()` after the recursive call, causing elements to remain in subsequent branches.
- Appending `stack` directly to `result` instead of `stack[:]`.
- Checking only `total == n` without ensuring `len(stack) == k`.
- Forgetting to stop when `len(stack) > k`.
- Continuing when `total > n`. Since all remaining candidates are positive, the sum cannot decrease.
- Forgetting that combinations must be unique and that choosing numbers in increasing order avoids different permutations of the same combination.

**Recognition Pattern:**

- The problem asks for combinations with an exact number of elements and an exact target sum.
- Each number can be used only once.
- The candidate range is small and fixed.
- Invalid partial combinations can be pruned early.
- Recognize **Backtracking + Fixed Combination Size + Target Sum + Pruning**.

**Key Observation:**

- The combination must satisfy two conditions simultaneously: `len(stack) == k` and `total == n`.
- The recursive call uses `j + 1`, ensuring that numbers are selected in strictly increasing order and cannot be reused.
- Because all candidates are positive, `total > n` is a safe pruning condition.
- Your code is logically correct. One optimization is to stop the loop when `total + j > n`, since every later candidate is larger and would also exceed the target.
- You can also prune when `len(stack) + (10 - i) < k`, because there would not be enough remaining numbers to complete the combination.

**Code:**

```python
class Solution:
    def combinationSum3(self, k: int, n: int) -> List[List[int]]:
        result = []
        def backtrack(i,total,stack):
            if len(stack) > k or total > n:
                return
            if len(stack) == k and total == n:
                result.append(stack[:])
                return
            for j in range(i,10):
                temp = total + j
                stack.append(j)
                backtrack(j + 1,temp,stack)
                stack.pop()
        backtrack(1,0,[])
        return result
```

#### LeetCode 17 - Letter Combinations of a Phone Number

- Repeat this process for all digits.
- Store every completed combination in `result`.
- This approach becomes cumbersome as the number of digits increases because the number of combinations grows exponentially.

**Optimal Approach:**

- Use **Backtracking** to build each combination one character at a time.
- Create `num_map` to map each digit from `2` to `9` to its corresponding letters.
- Initialize an empty `result` list to store completed combinations.
- Define `subset(i, stack)`:

**Complexity:**

- Optimal Auxiliary Space: O(n) for the recursion stack and current combination, excluding output.
- Output Space: O(n × 4ⁿ) as a worst-case bound for storing all combinations.

- Each input character maps to a collection of possible choices.
- One character must be selected for every input digit.
- Every choice at the current position combines with all valid choices from later positions.

- Each digit represents a set of possible letters.
- A valid combination contains exactly one letter from each digit's mapping.

- `stack.append(ch)` makes a choice, the recursive call explores it, and `stack.pop()` removes it before the next choice.
- **Important correction:** Add an early check in `letterCombinations` to return `[]` when `digits` is empty. This matches the problem's expected behavior.

**Code:**

```python
class Solution:
    def letterCombinations(self, digits: str) -> List[str]:
        result = []
        num_map = {
            2 : "abc",
            3 : "def",
            4 : "ghi",
            5 : "jkl",
            6 : "mno",
            7 : "pqrs",
            8 : "tuv",
            9 : "wxyz" 
        }
        def subset(i,stack):
            if i >= len(digits):
                result.append("".join(stack))
                return
            for ch in num_map[int(digits[i])]:
                stack.append(ch)
                subset(i + 1, stack)
                stack.pop()
        subset(0,[])
        return result
```

#### LeetCode 79 - Word Search

**Question:**

- Given an `m × n` grid of characters `board` and a string `word`, return `True` if the word exists in the grid.
- The word can be constructed from letters of sequentially adjacent cells.
- Adjacent cells are horizontal or vertical neighbors.
- The same cell cannot be used more than once in a single word path.

**Example:**

- Input:
  - `board = [["A","B","C","E"], ["S","F","C","S"], ["A","D","E","E"]]`
  - `word = "ABCCED"`
- Output: `True`
- Explanation: The word can be formed by moving through adjacent cells: `A → B → C → C → E → D`.

- Input:
  - `board = [["A","B","C","E"], ["S","F","C","S"], ["A","D","E","E"]]`
  - `word = "ABCB"`
- Output: `False`
- Explanation: Forming this word would require reusing a cell, which is not allowed.

**Algorithm:**

**Brute Force:**

- Start from every cell in the board.
- Explore every possible path of adjacent cells.
- Check whether the characters along a path match the word.
- Prevent cells from being reused within a path.
- Return `True` if any path matches the entire word; otherwise, return `False`.
- Without pruning, the number of paths can grow exponentially.

**Optimal Approach:**

- Use **Depth-First Search (DFS) + Backtracking**.
- Calculate the number of rows and columns in `board`.
- Define `Solve(i, j, idx)`:
  - `(i, j)` represents the current cell.
  - `idx` represents the index of the character currently being matched in `word`.
- If `idx == len(word)`, return `True` because every character has been matched.
- Return `False` if:
  - The current coordinates are outside the board.
  - `board[i][j]` does not match `word[idx]`.
- Save the current character in `temp`.
- Mark the current cell as visited by setting `board[i][j] = "#"`.
- Recursively search the four neighboring cells:
  - Down: `Solve(i + 1, j, idx + 1)`
  - Up: `Solve(i - 1, j, idx + 1)`
  - Right: `Solve(i, j + 1, idx + 1)`
  - Left: `Solve(i, j - 1, idx + 1)`
- Use logical `or` to return success as soon as one direction finds the remaining word.
- Restore the original character using `board[i][j] = temp` before returning.
- Start the search from every cell in the board.
- If any starting cell produces `True`, return `True`; otherwise, return `False`.

**Complexity:**

- Brute Force Time: Exponential, as all possible paths may need to be explored.
- Optimal Time: O(R × C × 4 × 3ᴸ⁻¹), where `R` and `C` are the board dimensions and `L` is the word length. After the first move, the immediately previous cell cannot be revisited, leaving at most three new directions per step.
- A simpler upper bound is O(R × C × 4ᴸ).
- Auxiliary Space: O(L) for the recursion stack, excluding the board.
- The algorithm temporarily modifies the board but restores every visited cell before returning.

**Common Mistakes:**

- Forgetting to check whether the coordinates are outside the board.
- Checking `board[i][j]` before validating the coordinates, which can cause an index error.
- Forgetting to compare the current cell with `word[idx]`.
- Incrementing `idx` incorrectly, causing characters to be skipped or repeated.
- Forgetting to mark a cell as visited, which can allow the same cell to be reused in a path.
- Forgetting to restore the original character after exploring neighboring cells.
- Marking a cell as visited permanently, which can interfere with searches from other paths.
- Searching only in two directions instead of all four.
- Accessing `board[0]` when the board is empty. Your code assumes the board has at least one row and one column; an empty-board check makes it more robust.

**Recognition Pattern:**

- The problem asks whether a sequence can be formed by traversing adjacent cells.
- Each move depends on the current position.
- A cell cannot be reused within a path.
- Failed paths must be undone so other possibilities can be explored.
- Recognize **Grid DFS + Backtracking + Visited-State Management**.

**Key Observation:**

- Every starting cell is a possible first character of the word.
- `idx` tracks the next character that must be matched.
- Marking a cell with `"#"` temporarily prevents revisiting it in the current path.
- Restoring the cell after recursive exploration allows other paths to use it.
- The base case `idx == len(word)` must be checked before attempting to access `word[idx]`.
- Your code correctly checks the four directions, prevents cell reuse, restores the board, and stops early when a valid path is found.
- **Important edge case:** If `board` is empty or its first row is empty, return `False` before calculating `len(board[0])`.

**Code:**

```python
class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        rows,columns = len(board),len(board[0])
        def Solve(i,j,idx):
            if idx == len(word):
                return True 
            if i < 0 or j < 0 or i >= rows or j >= columns or board[i][j] != word[idx]:
                return False
            temp = board[i][j]
            board[i][j] = "#"
            found = (Solve(i + 1, j, idx + 1) or 
                    Solve(i - 1, j, idx + 1) or
                    Solve(i, j + 1, idx + 1) or
                    Solve(i, j - 1, idx + 1))
            board[i][j] = temp
            return found
        for i in range(rows):
            for j in range(columns):
                if Solve(i,j,0):
                    return True
        return False
```

#### LeetCode 131 - Palindrome Partitioning

**Question:**

- Given a string `s`, partition it so that every substring in the partition is a palindrome.
- Return all possible palindrome partitions of `s`.
- A palindrome is a string that reads the same forward and backward.

**Example:**

- Input: `s = "aab"`
- Output: `[["a", "a", "b"], ["aa", "b"]]`
- Explanation:
  - `["a", "a", "b"]` is valid because each substring is a palindrome.
  - `["aa", "b"]` is valid because both `"aa"` and `"b"` are palindromes.
  - Partitions containing a non-palindrome substring are not allowed.

**Algorithm:**

**Brute Force:**

- Generate every possible way to split the string into substrings.
- Check whether every substring in each partition is a palindrome.
- If all substrings are palindromes, add that partition to `result`.
- Otherwise, discard the partition.
- A string of length `n` has \(2^{n-1}\) possible ways to place cuts between adjacent characters, so checking every partition can be expensive.

**Optimal Approach:**

- Use **Backtracking + Palindrome Checking** to generate only valid partitions.
- Initialize `result` to store all valid partitions and `temp` to store the current partition.
- Define `solve(string, i, temp, result)`:
  - `i` is the starting index of the remaining substring.
  - `temp` contains the palindrome substrings selected so far.
- If `i == len(string)`:
  - The entire string has been partitioned.
  - Append a copy of `temp` to `result`.
  - Return.
- Iterate through possible ending indices `j` from `i` to `len(string) - 1`.
- Check whether `string[i:j+1]` is a palindrome using `ispalindrome(string, i, j)`.
- If it is a palindrome:
  - Append `string[i:j+1]` to `temp`.
  - Recursively call `solve(string, j + 1, temp, result)` to partition the remaining suffix.
  - Remove the substring using `temp.pop()` to backtrack.
- The `ispalindrome` method compares characters at both ends and moves inward until the indices meet or cross.
- Start the recursion with `solve(s, 0, temp, result)`.
- Return `result`.

**Complexity:**

- Brute Force Time: Exponential in the number of possible partitions, with additional time needed to validate each partition.
- Optimal Time: O(n × 2ⁿ) as a commonly used upper bound, including palindrome checks and constructing the output partitions.
- Auxiliary Space: O(n) for the recursion stack and current partition, excluding output.
- Output Space: O(n × 2ⁿ) in the worst case, because there can be \(2^{n-1}\) partitions and each partition stores up to `n` characters across its substrings.

**Common Mistakes:**

- Forgetting the base case `i == len(string)`, which indicates that a complete partition has been formed.
- Appending `temp` directly to `result` instead of `temp[:]`, because backtracking modifies `temp`.
- Forgetting `temp.pop()` after the recursive call, causing substrings from one branch to affect another.
- Checking the entire string for a palindrome instead of checking only the candidate substring from `i` to `j`.
- Using an incorrect substring boundary. Python slicing `string[i:j+1]` includes `i` but excludes `j+1`, so it represents characters from `i` through `j`.
- Forgetting to move to `j + 1` after selecting a palindrome, which can cause the selected substring to be processed again.
- Allowing a substring that is not a palindrome to enter the current partition.
- In your code, `x = len(string) + 1` is valid because `ispalindrome` rejects `j >= len(string)`, but it performs one unnecessary iteration. Using `range(i, len(string))` is cleaner.

**Recognition Pattern:**

- The problem asks for all possible ways to partition a string.
- Every selected substring must satisfy a condition.
- The next recursive call processes the remaining suffix.
- Invalid choices can be rejected before recursion.
- Recognize **Backtracking + String Partitioning + Palindrome Validation**.

**Key Observation:**

- At every index `i`, try every possible substring starting at `i`.
- Only select a substring if it is a palindrome.
- Once a valid substring is selected, recursively partition the remaining string starting at `j + 1`.
- For `"aab"`, the first choices include `"a"`, `"aa"`, and `"aab"`. The last choice is rejected because `"aab"` is not a palindrome.
- Each recursive path represents one complete partition, and backtracking allows other valid partitions to be explored.
- Your approach is logically correct for ordinary nonempty strings. The loop can be simplified to `range(i, len(string))`, and your palindrome checker already handles the inclusive ending index correctly.

**Code:**

```python
class Solution:
    def solve(self,string, i,temp,result):
            if i == len(string):
                result.append(temp[:])
                return 
            x = len(string) + 1
            for j in range(i,x):
                if self.ispalindrome(string,i,j):
                    temp.append(string[i : j+1])
                    self.solve(string, j + 1,temp,result)
                    temp.pop()
            return

    def ispalindrome(self,string,i,j):
        while j < len(string) and i <= j:
            if string[i] != string[j]:
                return False
            i += 1
            j -= 1
        return True

    def partition(self, s: str) -> List[List[str]]:
        result,temp = [],[]
        self.solve(s,0,temp,result)
        return result
```

#### LeetCode 51 - N-Queens

**Question:**

- The N-Queens problem asks you to place `n` queens on an `n × n` chessboard so that no two queens attack each other.
- Queens cannot share the same row, column, or diagonal.
- Return all distinct solutions, where each solution is represented as a list of strings.
- Each string represents one row, with `"Q"` representing a queen and `"."` representing an empty cell.

**Example:**

- Input: `n = 4`
- Output: `[[".Q..", "...Q", "Q...", "..Q."], ["..Q.", "Q...", "...Q", ".Q.."]]`
- Explanation:
  - There are two valid arrangements for a 4 × 4 chessboard.
  - Each arrangement places four queens without any two queens attacking each other.

**Algorithm:**

**Brute Force:**

- Generate every possible arrangement of `n` queens on the board.
- Check whether any two queens share a row, column, or diagonal.
- Add each valid arrangement to the result.
- This approach explores a huge number of board configurations and performs unnecessary work checking invalid arrangements.

**Optimal Approach:**

- Use **Backtracking + Column-by-Column Placement**.
- Initialize an `n × n` board containing only `"."`.
- Define `solve(col, board, ans, n)` to place queens column by column.
- If `col == n`:
  - All `n` queens have been placed successfully.
  - Append `list(board)` to `ans` to save the current arrangement.
  - Return.

- For each position `(row, col)`, call `issafe(row, col, board, n)`.

  - **Lower-left diagonal:** Move diagonally downward and left.

- If the position is safe:
  - Place `"Q"` at `board[row][col]`.
  - Recursively call `solve(col + 1, board, ans, n)`.
  - Restore the cell to `"."` after returning from recursion.
- Start the search with `solve(0, board, result, n)`.
- Return `result`.

**Complexity:**

- Brute Force Time: Extremely large because all possible queen placements must be considered before validation.
- Optimal Time: O(n!), as a common upper-bound estimate for backtracking with one queen per column and one queen per row. The diagonal checks add O(n) work per attempted placement, so a more explicit upper bound is O(n × n!).
- Auxiliary Space: O(n) for recursion depth, excluding the board and output.
- Board Space: O(n²) for storing the chessboard.
- Output Space: Depends on the number of valid solutions, with O(n²) space required per solution.

**Common Mistakes:**

- Forgetting the base case `col == n`, which indicates that all queens have been placed.
- Placing more than one queen in the same column.
- Checking only rows and columns while forgetting both diagonals.
- Checking the right side even though the algorithm places queens from left to right.
- Forgetting to restore the board after a recursive call, which causes one branch to affect another.
- Appending `board` directly to `ans` rather than `list(board)`. A copy is necessary because the board is modified during backtracking.
- Forgetting to check whether a position is safe before placing a queen.
- Incorrectly restoring a cell using the wrong index.
- Returning the result from the recursive helper is unnecessary in each branch because `ans` is modified in place. Your code works because the base case appends to `ans` and the outer call returns the same list.

**Recognition Pattern:**

- The problem asks for all valid arrangements satisfying multiple constraints.
- Each decision places one queen in a column.
- Invalid partial arrangements can be rejected before exploring further.
- Every choice must be undone before trying another position.
- Recognize **Backtracking + Constraint Checking + N-Queens**.

**Key Observation:**

- Place exactly one queen in each column, moving from left to right.
- Since each new queen is placed in a later column, only the left side of the board needs to be checked.
- The upper-left diagonal detects conflicts with queens above the current row.
- The lower-left diagonal detects conflicts with queens below the current row.
- The leftward row scan detects conflicts in the same row.
- If all three checks pass, the current position is safe.
- Your algorithm correctly checks these three directions and restores the board after each recursive call.
- The comment about slicing is correct: `board[row][:col]` preserves the cells before the current column, and `board[row][col + 1:]` preserves the cells after it.

**Code:**

```python
class Solution:
    def solve(self,col,board,ans,n):
        if col == n:
            ans.append(list(board))
            return
        for row in range(n):
            if self.issafe(row,col,board,n):
                board[row] = board[row][:col] + "Q" + board[row][col+1:]
                # Start at index cols and take everything until the end.
                self.solve(col + 1,board,ans,n)
                board[row] =board[row][:col] + "." + board[row][col+1:]
        return ans

    def issafe(self,row,col,board,n):
        duprow,dupcol = row,col
        # check upper diagonal
        while row >= 0 and col >= 0:
            if board[row][col] == "Q":
                return False
            row -= 1
            col -= 1
        col,row = dupcol,duprow
        # check lower diagonal
        while row < n and col >= 0:
            if board[row][col] == "Q":
                return False
            row += 1
            col -= 1
        col,row = dupcol,duprow
        # check columns
        while col >= 0:
            if board[row][col] == "Q":
                return False
            col -= 1
        return True

    def solveNQueens(self, n: int) -> List[List[str]]:
        result = []
        board = ["." * n for i in range(n)]
        result = self.solve(0,board,result,n)
        return result
```

#### LeetCode 37 - Sudoku Solver

**Question:**

- Given a partially filled 9 × 9 Sudoku board, fill the empty cells so that the board becomes a valid Sudoku.
- Each row must contain the digits `1` through `9` without repetition.
- Each column must contain the digits `1` through `9` without repetition.
- Each 3 × 3 sub-box must contain the digits `1` through `9` without repetition.
- Empty cells are represented by `"."`.
- Modify the board in place. The problem guarantees that the given puzzle has a unique solution.

**Example:**

- Input: A partially filled 9 × 9 Sudoku board.
- Output: The same board filled with digits so that all Sudoku constraints are satisfied.
- Explanation: Every empty cell is assigned a digit that does not violate the row, column, or 3 × 3 box constraints.

**Algorithm:**

**Brute Force:**

- Traverse the board row by row and column by column.
- When an empty cell is found, try each digit from `1` to `9`.
- Use `isvalid(board, row, col, s)` to check whether the digit can be placed.
- The validity check scans:
  - The current column.
  - The current row.
  - The corresponding 3 × 3 box.
- If the digit is valid, place it in the cell and recursively solve the remaining board.
- If the recursive call fails, restore the cell to `"."` and try another digit.
- If every empty cell is filled successfully, return `True`.
- This approach repeatedly scans rows, columns, and boxes, even when the same information was checked in earlier recursive calls.

**Optimal Approach:**

- Use **Backtracking + Hash Sets for Constraint Tracking**.
- Create three collections of sets:
  - `rows[r]` stores digits already present in row `r`.
  - `cols[c]` stores digits already present in column `c`.
  - `boxes[b]` stores digits already present in the corresponding 3 × 3 box.
- Initialize these sets by traversing all existing cells in the board.
- Calculate the box index using:

  `box_index = (row // 3) * 3 + (col // 3)`

- Define `solve(row, col)` to process cells sequentially.
- If `row == 9`, return `True` because the entire board has been solved.
- If `col == 9`, move to the next row by calling `solve(row + 1, 0)`.
- If the current cell is already filled, move to the next column.
- Otherwise, try digits from `1` to `9`.
- For each digit, check whether it is absent from the current row, column, and box sets.
- If the digit is valid:
  - Place it in `board[row][col]`.
  - Add it to the corresponding row, column, and box sets.
  - Recursively call `solve(row, col + 1)`.
- If recursion succeeds, return `True` immediately.
- If recursion fails:
  - Restore the cell to `"."`.
  - Remove the digit from all three sets.
  - Try the next digit.
- If no digit works, return `False`.
- Start with `solve(0, 0)`. The board is modified in place, so no separate result needs to be returned.

**Complexity:**

- Brute Force Time: O(9ᴱ), where `E` is the number of initially empty cells, as a loose worst-case bound. Each candidate assignment also requires O(1) work for a fixed-size 9 × 9 validity check.
- Brute Force Auxiliary Space: O(E) for the recursion stack.
- Optimal Time: O(9ᴱ) as a worst-case upper bound. Set lookups are O(1) on average, but backtracking may still explore many candidate assignments.
- Optimal Auxiliary Space: O(E) for the recursion stack, plus O(81) space for the row, column, and box sets.
- The board is modified in place, so no additional board copy is required.

**Common Mistakes:**

- Confusing row and column checks when validating a candidate digit.
- Forgetting to check the entire 3 × 3 box.
- Calculating the box index incorrectly. The correct formula is `(row // 3) * 3 + (col // 3)`.
- Forgetting to update all three sets after placing a digit.
- Forgetting to remove the digit from all three sets during backtracking.
- Restoring the board cell to the wrong value instead of `"."`.
- Returning `False` too early before trying all digits for an empty cell.
- Forgetting the base case `row == 9`, which indicates that all rows have been completed.
- Forgetting to move to the next row when `col == 9`.
- In the set-based approach, returning immediately after a successful recursive call is essential; otherwise, the solved board could be undone by backtracking.
- In the brute-force approach, the recursive call returns a Boolean. Returning `True` when it succeeds and restoring the cell only when it fails are both necessary.

**Recognition Pattern:**

- The problem asks for a complete arrangement satisfying multiple constraints.
- Every empty position has a limited set of possible values.
- A choice may make a later position impossible, requiring the algorithm to undo it.
- Repeated validity checks can be optimized by maintaining previously used values.
- Recognize **Backtracking + Constraint Satisfaction + Hash-Set Optimization**.

**Key Observation:**

- Sudoku solving is a constraint-satisfaction problem: each empty cell must satisfy row, column, and box restrictions.
- Your brute-force solution checks constraints by scanning the board each time a candidate digit is considered.
- Your optimized solution maintains sets, so checking whether a digit is already used takes O(1) average time.
- The set-based approach must keep the board and the three sets synchronized during both placement and backtracking.
- The base case `row == 9` signals success because recursion processes all nine rows.
- Both approaches use backtracking, but the second reduces repeated validation work.
- **Further optimization:** Choose the empty cell with the fewest valid candidate digits (the minimum-remaining-values heuristic) instead of always processing cells from left to right. This can substantially reduce the search tree for difficult puzzles.

**Code:**

```python
class Solution:
    def solveSudoku(self, board: List[List[str]]) -> None:
        rows = [set() for _ in range(9)]
        cols = [set() for _ in range(9)]
        boxes = [set() for _ in range(9)]
        # Initialize sets based on the initial board state
        for r in range(9):
            for c in range(9):
                val = board[r][c]
                if val != '.':
                    rows[r].add(val)
                    cols[c].add(val)
                    box_index = (r // 3) * 3 + (c // 3)
                    boxes[box_index].add(val)

        def solve(row, col):
            if row == 9:
                return True
            if col == 9:
                return solve(row + 1, 0)
            if board[row][col] != '.':
                return solve(row, col + 1)
            box_index = (row // 3) * 3 + (col // 3)
            for val in map(str, range(1, 10)):
                if val not in rows[row] and val not in cols[col] and val not in boxes[box_index]:
                    board[row][col] = val
                    rows[row].add(val)
                    cols[col].add(val)
                    boxes[box_index].add(val)
                    if solve(row, col + 1):
                        return True
                    # backtrack
                    board[row][col] = '.'
                    rows[row].remove(val)
                    cols[col].remove(val)
                    boxes[box_index].remove(val)
            return False

        solve(0, 0)

# Brute Force:
class Solution:
    def isvalid(self,board,row,col,s):
        # row checking
        for i in range(9):
            if board[i][col] == s:
                return False
        # column checking
        for j in range(9):
            if board[row][j] == s:
                return False
        # 3x3 box checking
        rowstart = 3 * (row // 3)
        colstart = 3 * (col // 3)
        for i in range(3):
            for j in range(3):
                if board[rowstart + i][colstart + j] == s:
                    return False
        return True

    def solveSudoku(self, board: List[List[str]]) -> None:
        for i in range(9):
            for j in range(9):
                if board[i][j] == ".":
                    for s in map(str,range(1,10)):
                        if self.isvalid(board,i,j,s):
                            board[i][j] = s
                            if self.solveSudoku(board):
                                return True
                            board[i][j] = "."
                    return False
        return True
```

#### LeetCode 282 - Expression Add Operators

**Question:**

- Given a string `num` containing only digits and an integer `target`, insert the operators `+`, `-`, and `*` between the digits to create expressions that evaluate to `target`.
- Return all valid expressions that evaluate to the target.
- The digits must remain in their original order.
- Multi-digit operands are allowed, but operands cannot contain leading zeros unless the operand itself is `0`.
- Multiplication must follow normal arithmetic precedence, so it is evaluated before addition and subtraction.

**Example:**

- Input: `num = "123", target = 6`
- Output: `["1+2+3", "1*2*3"]`
- Explanation:
  - `1 + 2 + 3 = 6`
  - `1 * 2 * 3 = 6`

- Input: `num = "232", target = 8`
- Output: `["2+3*2", "2*3+2"]`
- Explanation:
  - `2 + 3 * 2 = 8`
  - `2 * 3 + 2 = 8`

**Algorithm:**

**Brute Force:**

- Generate every possible way to split the digit string into operands.
- Insert `+`, `-`, or `*` between consecutive operands.
- Evaluate each generated expression using the correct operator precedence.
- If the expression evaluates to `target`, append it to `result`.
- Reject operands with leading zeros, such as `"05"`.
- This approach generates an exponential number of expressions and repeatedly evaluates them.

**Optimal Approach:**

- Use **Backtracking + Expression Evaluation with Operator Precedence**.
- Define `Solve(num, target, start, current_val, last_operand, expression, result)`:
  - `start` is the index of the next digit to process.
  - `current_val` is the value of the expression constructed so far.
  - `last_operand` is the signed value of the most recently added operand, including any multiplication chain.
  - `expression` stores the expression constructed so far.
  - `result` stores expressions that evaluate to `target`.
- If `start == len(num)`:
  - Check whether `current_val == target`.
  - If true, append `expression` to `result`.
  - Return.
- Iterate through every possible ending index `i` from `start` to `len(num) - 1`.
- If the operand would have a leading zero, stop extending it. For example, `"05"` is invalid, but `"0"` is valid.
- Extract the candidate operand using `current_num = num[start:i+1]`.
- Convert the operand to an integer using `current_num_val = int(current_num)`.
- **First operand:**
  - If `start == 0`, add the operand without an operator.
  - Initialize `current_val` and `last_operand` to `current_num_val`.
- **Addition:**
  - Add the new operand to `current_val`.
  - Set `last_operand` to the positive value of the new operand.
  - Append `"+" + current_num` to the expression.
- **Subtraction:**
  - Subtract the new operand from `current_val`.
  - Set `last_operand` to the negative value of the new operand.
  - Append `"-" + current_num` to the expression.
- **Multiplication:**
  - Remove the previous signed operand from `current_val`.
  - Add back the previous operand multiplied by the new operand.
  - Update `last_operand` to `last_operand * current_num_val`.
  - Append `"*" + current_num` to the expression.
- Continue recursively until every digit has been processed.
- Return `result`.

**Complexity:**

- Brute Force Time: Exponential, because all possible operand splits and operator combinations may need to be generated and evaluated.
- Brute Force Space: Exponential in the worst case if all generated expressions are stored.
- Optimal Time: O(n × 4ⁿ) as a common upper-bound estimate, where `n` is the length of `num`. There are exponentially many expression choices, and building or storing an expression can take O(n) time.
- Optimal Auxiliary Space: O(n) for recursion depth and the current expression, excluding output.
- Output Space: O(n × R), where `R` is the number of valid expressions returned.

**Common Mistakes:**

- Forgetting to handle the first operand separately, because an expression cannot begin with `+`, `-`, or `*`.
- Allowing operands with leading zeros, such as `"05"` or `"00"`.
- Updating `current_val` with multiplication directly, without correcting for the previous operand.
- Forgetting that multiplication has higher precedence than addition and subtraction.
- Storing `last_operand` as a positive number after subtraction. It must be negative so multiplication correctly handles expressions such as `5-2*3`.
- Updating `last_operand` incorrectly after multiplication, which breaks chains such as `2*3*4`.
- Comparing `current_val` with `target` before consuming every digit.
- Appending an expression before all digits have been processed.
- Forgetting to include the operator in the recursive expression string.
- Returning from the leading-zero check incorrectly. Your `return` is correct here because once an operand starts with `0`, extending it would create an invalid multi-digit operand.

**Recognition Pattern:**

- The problem asks for all expressions formed by inserting operators between digits.
- The order of the digits cannot change.
- Different operator choices produce different expression values.
- Operator precedence must be handled while building the expression.
- Recognize **Backtracking + String Partitioning + Arithmetic State Tracking**.

**Key Observation:**

- The key challenge is multiplication precedence.
- For an expression such as `1 + 2 * 3`, simply adding `2 * 3` to the accumulated value would be incorrect if the previous operand were already included.
- Maintain `last_operand` so the multiplication branch can undo the previous operand and replace it with the multiplied value.
- Example: For `1 + 2 * 3`:
  - Before multiplication, `current_val = 3` and `last_operand = 2`.
  - Remove the previous operand: `3 - 2 = 1`.
  - Add the multiplied operand: `1 + (2 * 3) = 7`.
- For subtraction, store the last operand as negative. For example, `5 - 2 * 3` becomes `5 - 6 = -1`, not `5 - 2 - 6`.
- Your implementation correctly tracks `current_val` and `last_operand`, handles multiplication precedence, and prevents leading-zero operands.

**Code:**

```python
class Solution:
    def Solve(self,num,target,start,current_val,last_operand,expression,result):
        if start == len(num):
            if current_val == target:
                result.append(expression)
            return
        for i in range(start,len(num)):
            if i > start and num[start] == '0':
                return 
            current_num = num[start:i+1]  
            current_num_val = int(current_num)  
            if start == 0:
                self.Solve(num,target,i + 1,current_num_val,current_num_val,current_num,result)
            else:
                self.Solve(num,target,i + 1,current_val + current_num_val,current_num_val,expression + "+" + current_num,result)
                self.Solve(num,target,i + 1,current_val - current_num_val,-current_num_val,expression + "-" + current_num,result)
                self.Solve(num,target,i + 1,current_val - last_operand + last_operand * current_num_val,last_operand * current_num_val,expression + "*" + current_num,result)

    def addOperators(self, num: str, target: int) -> List[str]:
        result  = []
        self.Solve(num,target,0,0,0,"",result)
        return result
```
