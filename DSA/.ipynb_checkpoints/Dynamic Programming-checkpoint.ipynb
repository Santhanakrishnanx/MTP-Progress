#### LeetCode 139 - Word Break

**Question:**

Given a string `s` and a dictionary of strings `wordDict`, return `True` if `s` can be segmented into a sequence of one or more dictionary words. Otherwise, return `False`.

The same dictionary word may be reused multiple times.

**Example:**

- Input: `s = "leetcode"`, `wordDict = ["leet", "code"]`
- Output: `True`
- Explanation:
  - `"leetcode"` can be segmented into `"leet"` + `"code"`.
  - Both words exist in `wordDict`.

- Input: `s = "applepenapple"`, `wordDict = ["apple", "pen"]`
- Output: `True`
- Explanation:
  - `"applepenapple"` can be segmented into `"apple"` + `"pen"` + `"apple"`.

- Input: `s = "catsandog"`, `wordDict = ["cats", "dog", "sand", "and", "cat"]`
- Output: `False`
- Explanation:
  - No valid segmentation can cover the entire string using only dictionary words.

**Algorithm:**

- Use **dynamic programming (DP)** to determine whether each prefix of the string can be segmented into dictionary words.
- `dp[i]` indicates whether the prefix `s[0:i]` can be segmented successfully.
- Initialize `dp[0] = True` because an empty prefix requires no words.
- For every ending index `i`, check possible starting indices `j`.
- If `dp[j]` is `True` and `s[j:i]` belongs to the dictionary, then `dp[i] = True`.
- Use a set for `wordDict` to make dictionary membership checks efficient.
- Limit the search to the maximum dictionary word length because no valid dictionary word can be longer than that.

**Brute Force:**

- Start from the beginning of the string and try every possible prefix that matches a word in `wordDict`.
- If a matching word is found, recursively attempt to segment the remaining suffix.
- Return `True` if any sequence of matching words covers the entire string.
- Return `False` if no possible segmentation succeeds.
- This approach can take exponential time because it may repeatedly explore the same suffixes.

**Optimal Approach:**

- **Step 1: Calculate the maximum dictionary word length**
  - Initialize `maxlen = 0`.
  - Iterate through `wordDict` and update `maxlen` with the length of the longest word.
  - This limits how far backward each DP iteration needs to search.

- **Step 2: Create a set for dictionary lookup**
  - Execute `wordSet = set(wordDict)`.
  - A set provides average O(1) membership checks, avoiding repeated linear searches through the dictionary.

- **Step 3: Initialize the DP array**
  - Let `n = len(s)`.
  - Create `dp = [False] * (n + 1)`.
  - Each position `i` represents the prefix `s[0:i]`.
  - Set `dp[0] = True` because the empty prefix is segmentable.

- **Step 4: Iterate through every prefix**
  - Iterate through `i` from `0` to `n`.
  - Each iteration determines whether the prefix ending at index `i` can be segmented.

- **Step 5: Check possible starting positions**
  - Iterate through `j` from `max(0, i - maxlen)` to `i - 1`.
  - This considers only substrings whose lengths do not exceed the longest dictionary word.
  - The substring `s[j:i]` starts at index `j` and ends just before index `i`.

- **Step 6: Apply the DP transition**
  - Check whether `dp[j]` is `True`.
  - Check whether `s[j:i]` exists in `wordSet`.
  - If both conditions hold, the prefix ending at `i` can be segmented.
  - Set `dp[i] = True` and break because one valid segmentation is sufficient.

- **Step 7: Return the final result**
  - Return `dp[n]`.
  - If `dp[n]` is `True`, the entire string can be segmented into dictionary words; otherwise, it cannot.

**Complexity:**

- **Brute Force:** Time Complexity: Exponential in the worst case  Space Complexity: O(N) auxiliary space for the recursion stack, excluding stored input.
- **Optimal Approach:** Time Complexity: O(N × L²) in the worst case  Space Complexity: O(N + D), excluding the input, where:
  - `N` is the length of `s`.
  - `L` is the maximum length of a dictionary word.
  - `D` is the total number of characters stored across the dictionary words.
- The DP algorithm examines at most `L` possible starting positions per ending position, and each substring operation and hash lookup can take up to O(L) time in the worst case.
- The DP array requires O(N) space, while the set requires O(D) space.

**Common Mistakes:**

- Initializing `dp[0] = False`. The empty prefix must be `True` so that words beginning at index `0` can be recognized.
- Using `s[j:i+1]` instead of `s[j:i]`. The ending index `i` is exclusive in Python slicing.
- Checking only whether `s[j:i]` is in the dictionary without verifying `dp[j]`. The preceding prefix must also be segmentable.
- Iterating over every possible `j` even when the substring exceeds the longest dictionary word. Using `maxlen` avoids unnecessary checks.
- Forgetting to break after setting `dp[i] = True`. The result would remain correct without the break, but the additional checks would be unnecessary.
- Returning `dp[n - 1]` instead of `dp[n]`. The DP array has length `n + 1`, and `dp[n]` represents the entire string.
- Assuming that matching a word is enough. A valid segmentation must cover the entire string without leaving an unmatched prefix or suffix.

**Recognition Pattern:**

- Problems asking whether a string can be divided into valid dictionary words.
- Problems involving valid prefixes and decisions based on previously solved prefixes.
- Problems where a successful state at index `j` can help determine the state at index `i`.
- Use one-dimensional DP when the answer depends on whether earlier prefixes can be segmented successfully.

**Key Observation:**

- `dp[i]` is `True` if at least one valid split exists where the prefix before the split is segmentable and the remaining substring is a dictionary word.
- The transition is `dp[i] = True` whenever `dp[j]` is `True` and `s[j:i]` belongs to `wordSet`.
- The maximum dictionary word length restricts the search to relevant starting positions.
- Since the problem asks whether any valid segmentation exists, finding one valid split is enough to stop checking that position.

**Code:**

```python
class Solution:
    def wordBreak(self, s: str, wordDict: List[str]) -> bool:
        maxlen = 0
        wordSet = set(wordDict)
        for word in wordDict:
            maxlen = max(maxlen,len(word))
        n = len(s)
        dp = [False] * (n + 1) 
        dp[0] = True
        for i in range(n + 1):
            for j in range(max(0, i - maxlen), i):
                if dp[j] and s[j:i] in wordSet:
                    dp[i] = True
                    break
        return dp[n]
```
