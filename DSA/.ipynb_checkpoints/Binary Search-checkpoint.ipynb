## It requires the array to be sorted

- assign low and high as start and end of array which is our entire search space and pick the middle point and compare with target
- if it equal to target then return index
- if it is less than target then make high = mid -1 and then calculate mid for the current search space and perform the same
- if it is greater than the target then make high = mid + 1 and then calculate mid for the current search space and perform the same
- mid = (low + high)/2
- when high < low return - 1
- Binary search takes O(log n) timefor n length array

## Pseudo code

### Iterative Method

```python
function(arr,n,target)
{
    low = 0; high = n-1 ;
    while (low <= high):
    {
        mid = (low + high)/2
        if (arr[mid] == target)
            return mid;
        else if (target > arr[mid] )
            low = mid +1;
        else
            high = mid - 1;
    }
    return -1;
}       
```

### Recursive Method 

```python
function(arr,low,high,target)
{
    if (low > high)
        return -1;
    mid = (low + high)/ 2
    if  (arr[mid] == target)
        return mid;
    else if (target > arr[mid])
        return function(arr,mid + 1, high,target)
    else
        return function(arr,low,mid - 1,target)
}
```

- the time complexity is O(log2 n) becaue each time search psace is divided into half
- to avoid overflow at big array we can use something like mid = low + (high - low )/2

# Lower Bound

### smallest index such that arr[index] >= x then return lower bound = index

- searching extensilvelly one by one will take time complexity of O(n)
- so we use low = 0 and high = n- 1
- create a variable called ans and calculate mid and check arr[mid] >= x
- if the condition is satisfied then upadate the ans but still perform reducing the serach space because we need to find the smallest element
- based on the result move either high = mid - 1 or low = mid + 1
- do this iteratively until high < low then return ans which stores the smallest index 

## Pseudo Code

```python
function(arr,x,n)
{
    low = 0, high = n-1;
    ans = n
    while (low < = high)
    {
        mid = (low + high)/2;
        if (arr[mid] >= x)
        {
            ans = mid;
            high = mid - 1;
        }
        else
        {
            low = mid + 1; 
        }
        
    }
}
```

```python
for c++ code :
lb =  lower_bound(arr.begin(),arr.end(),x) - arr.begin();
```

- time complexity is O(log n)

# First and Last Occurence of x

#### Solution with lower bound and upper bound 

```python
class Solution(object):
    def searchRange(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        result = []
        n = len(nums)
        low,high = 0,n-1
        first,last = -1,-1
        while low <= high :
            mid = (low + high)//2
            if nums[mid] >= target :
                if nums[mid] == target:
                    first = mid
                high = mid - 1
            else:
                low = mid + 1
        low,high = 0,n-1
        while low <= high :
            mid = (low + high)//2
            if nums[mid] > target :
                last = mid
                high = mid - 1
            else:
                low = mid + 1
        if first == -1:
            return [-1, -1]
        elif first != -1 and last == -1:
            return [first,n-1]
        else:
            last = last - 1
        result.append(first)
        result.append(last)
        return result 
```

# Count Occurrences in a Sorted Array

```python
class Solution(object):
    def searchRange(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        result = []
        n = len(nums)
        low,high = 0,n-1
        first,last = -1,-1
        while low <= high :
            mid = (low + high)//2
            if nums[mid] == target :
                first = mid
                high = mid - 1
            elif nums[mid] >= target:
                high = mid - 1
            else:
                low = mid + 1
        low,high = 0,n-1
        while low <= high :
            mid = (low + high)//2
            if nums[mid] == target :
                last = mid
                low = mid + 1
            elif nums[mid] >= target:
                high = mid - 1
            else:
                low = mid + 1
        if first == -1:
            return 0
        
        return last - first + 1
```

# Search in rotated sorted array

- when the sorted array is rotated at a point we have to find the target element in this array
- example like [0,1,2,4,5,6,7] -> [4,5,6,7,0,1,2]

### for normal array like [4,5,6,7,0,1,2]

```python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        n = len(nums)
        low,high = 0,n-1
        while low <= high:
            mid = (low + high)//2
            if nums[mid] == target:
                return mid
            if nums[low] <= nums[mid]:
                if nums[low] <= target and target <= nums[mid]:
                    high = mid - 1
                else:
                    low = mid + 1
            else:
                if nums[mid] <= target and target <= nums[high]:
                    low = mid + 1
                else:
                    high = mid - 1
        return - 1 
```

### for non decreasing array like [2,5,6,0,0,1,2]

```python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: bool
        """
        n = len(nums)
        low,high = 0,n-1
        while low <= high:
            mid = (low+high)//2
            if nums[mid] == target:
                return True
            if nums[low] == nums[mid] == nums[high]:
                low = low + 1
                high = high - 1
                continue
            if nums[low] <= nums [mid]:
                if nums[low] <= target and target <= nums[mid]:
                    high = mid - 1
                else:
                    low = mid + 1
            else:
                if nums[mid] <= target and target <= nums[high]:
                    low = mid + 1
                else:
                    high = mid - 1
        return False
```

# Find minimum in Rotated Sorted Array

- find the sorted half
- Sorted Half may or may not have the answer
- pick the minimum from the sorted half and eliminate that space
- first check whether left side is sorted if yes then pick the minimum element and move to the right space
- if left side is not sorted then pick the minimum element in the right and move to left search space
- keep track of the minimum element in a seperate varible

```python
class Solution(object):
    def findMin(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums)
        low,high = 0, n-1
        result = float('inf')
        while low <= high:
            mid = (low + high)//2  
            if nums[low] <= nums[mid]:
               result = min(result,nums[low])
               low = mid + 1
            else:
                result = min(result,nums[mid])
                high = mid - 1
        return result
        
```

# 	Find out how many times the array is rotated

- single point rotation
- we find the minimum of the array
- along with result we keep track of the index of the minimum element

```python
def rotated(nums):
    n = len(nums)
    low,high = 0, n-1
    result = float('inf')
    index = -1
    while low <= high:
        mid = (low + high)//2  
        if nums[low] <= nums[mid]:
            if nums[low] < result:
                index = low
                result = nums[low]
            low = mid + 1
        else:
            if nums[mid] < result:
                index = mid
                result = nums[mid]
            high = mid - 1
    return index
        
    
arr = [4,5,6,7,0,1,2,3]
rlt = rotated(arr)
print(rlt)
arr = [3,4,5,1,2]
rlt = rotated(arr)
print(rlt)
```

# Single element in a Sorted Array

- In a sorted list every element has a pair except the one element which does not have a pair find the element
- if the problem is to solved within O(log n) then use the binary search
- understand the pattern like when there is repetation of two elements then the index would be even and odd for the same two elements whereas when the left or right subspace contains the single element then this gets disturbed and changed to odd and even look for which side has this pattern then we omit the side which does not have this pattern
- if even and odd pattern is there in the mid then eliminate left space and search in right else move to left space and eliminate right space
- if odd and even pattern is there in the mid then eliminate the right space and search in left else eliminate the left and move to right space
- this method shrink the search space in each iteration
- the edge case are when there is single element in array so we just keep if statement to return that same lement
- the other problems are the 0 and n-1 elements can cause index out of bound error so we push the low and high to 1 and n-2
- so keep two sepearte if conditions to check the borders of the array

```python
class Solution(object):
    def singleNonDuplicate(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums)
        low,high = 1,n-2
        if n == 1:
            return nums[0]
        if nums[0] != nums[1]:
            return nums[0]
        if nums[n-1] != nums[n-2]:
            return nums[n-1]
        while low <= high:
            mid =(low + high)//2
            if nums[mid] == nums[mid + 1]:
                if mid % 2 == 0 and (mid + 1)%2 == 1:
                    low = mid + 1
                else:
                    high = mid - 1
            elif nums[mid] == nums[mid - 1]:
                if (mid- 1)%2 == 0 and mid%2 == 1:
                    low = mid + 1
                else:
                    high = mid - 1
            else:
                return nums[mid]
        
```

# Find peak element

- peak element is  arr[i-1] < arr[i] > arr[i+1]
- an array can have multiple peak element and they mostly expect to return any one of them
- assume in both the start and end element have the adjacent as - infinite so for arr[] = [1,2,3,4,5] or arr[] = [5,4,3,2,1] the peak element is 5 
- the naive approach is too iterate through each element and compare its adjacent elements 

#### Pseudo code for Brute Force:

```python
for (i ==0;i < n; i++){
    if ((i == 0 || arr[i - 1] < arr[i]) &&
        (i == n - 1 || arr[i] > arr[i + 1]))
                return i;
}
```

- this brute force takes O(n) time complexity and O(1) space complexity
- look for whether array have a close tendency to sorted order

- first find the mid like binary search and check whther the mid is a peak element if yes return that element
- if no then check which of the two condition satisfy if arr[i-1] < arr[i]  then omit left space and move to right search space. if it satisfy this arr[i] > arr[i+1] then omit right space and move to left search space.
- introduce two if condition at start to check the start and end for peak elements
- then reduce the search space from 0 , n- 1 to 1 , n-2
- here we use a greedy approach where we pick the first found peak element and return it even though it has multiple peak element

```python
class Solution(object):
    def findPeakElement(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums)
        low,high = 1,n-2
        if n == 1 or nums[0] > nums[1]:
            return 0
        if nums[n-2] < nums[n-1]:
            return n - 1
        while low <= high:
            mid = (low + high)//2
            if nums[mid - 1] < nums[mid] and nums[mid] > nums[mid + 1]:
                return mid 
            elif nums[mid - 1] < nums[mid]:
                low  = mid + 1
            else:
                high = mid - 1 
```
