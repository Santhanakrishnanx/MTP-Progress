**Question 1:**

You are given a list threshold of length N, where,Each robot i has a threshold value threshold[i]. Each robot can be in one of two states: Operating (ON) or Idle (OFF). Let K be the total number of operating robots. A configuration is valid if no robot malfunctions.

A robot malfunction:

- It is ON and
the number of other operating robots is less than its threshold
→ (K - 1) < threshold[i]

- It is OFF and
the number of operating robots is greater than its threshold
→ K > threshold[i]


Return the number of valid configurations such that no robot malfunctions.

**Solution**

- Let K = total number of operating robots.
- For a configuration to be valid, Exactly K robots must have threshold[i] < K.
- So we must count how many K satisfy, count(threshold[i] < K) == K
- Sort the threshold array.
- Try all K from 0 to N.
- Sorting → O(N log N), Checking K → O(N) so Overall → O(N log N)

```python
def count_valid_configurations(threshold):
    """
    Returns the number of valid robot operating configurations
    such that no robot malfunctions.
    
    Assumes at least one robot must be operating.
    """
    
    n = len(threshold)
    threshold.sort()
    
    valid_count = 0
    pointer = 0  # counts how many elements are < K
    
    for K in range(1, n + 1):   # start from 1 instead of 0
        
        while pointer < n and threshold[pointer] < K:
            pointer += 1
        
        if pointer == K:
            valid_count += 1
    
    return valid_count


x = [6,0,3,3,6,7,2,7]
y = [3,0,0,3,3]

print(count_valid_configurations(x))
print(count_valid_configurations(y))  
```

**Question 2:**

You are given a list of strings orders
Each string represents one customer order and contains space-separated product IDs.
Example:

[
  "B07 B08 B09",
  "B07 B08",
  "B08 B09"
]

For each order:
- Consider all unique 2-item combinations of products.Count how many times each pair appears across all orders.
- Return the pair (as a string) that: Has the highest frequency,If multiple pairs have same frequency → return the lexicographically largest pair.
Return format example:

"B08 B09"


Solution:
- Split by space
- Remove duplicates inside a single order
- Generate all 2-item combinations
- Store frequency in dictionary
- Apply tie-breaking rule

```python
def most_frequent_pair(orders):
    freq = {}

    for order in orders:
        items = sorted(set(order.split()))

        for i in range(len(items)):
            for j in range(i + 1, len(items)):
                pair = (items[i], items[j])
                
                if pair in freq:
                    freq[pair] += 1
                else:
                    freq[pair] = 1

    max_freq = max(freq.values())

    candidates = []
    for pair, count in freq.items():
        if count == max_freq:
            candidates.append(pair)
    best_pair = max(candidates)
    return best_pair[0] + " " + best_pair[1]

orders = [
    "B07 B08 B09",
    "B07 B08",
    "B08 B09"
]

print(most_frequent_pair(orders))
```
