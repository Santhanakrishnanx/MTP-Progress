#### LeetCode 1903 - Largest Odd Number in String

**Question:**

Given a string `num` representing a large non-negative integer, return the largest-valued odd integer (as a string) that is a non-empty substring of `num`.

If no odd integer exists, return an empty string `""`.

**Example:**

- Input: `num = "52"`
- Output: `"5"`
- Explanation:
  - The substrings are `"5"`, `"2"`, and `"52"`.
  - `"5"` is odd, while `"2"` and `"52"` are even.
  - Therefore, the largest odd-number substring is `"5"`.

- Input: `num = "4206"`
- Output: `""`
- Explanation:
  - Every digit is even, so no odd-number substring exists.

- Input: `num = "35427"`
- Output: `"35427"`
- Explanation:
  - The last digit is `7`, which is odd.
  - Therefore, the entire string represents an odd number.

**Algorithm:**

- An integer is odd if and only if its last digit is odd.
- To find the largest odd-number substring, scan the string from right to left.
- The first odd digit encountered is the rightmost odd digit.
- Return the substring from the beginning through that digit.
- If no odd digit exists, return an empty string.

**Brute Force:**

- Generate every possible non-empty substring of `num`.
- Check whether each substring ends in an odd digit.
- Among all odd substrings, select the one with the largest numerical value.
- Return the selected substring, or `""` if none exists.
- This approach is inefficient because it generates and checks many unnecessary substrings.

**Optimal Approach:**

- **Step 1: Get the string length**
  - Set `n = len(num)`.
  - This determines the last valid index as `n - 1`.

- **Step 2: Scan from right to left**
  - Use `for i in range(n - 1, -1, -1)`.
  - Start at the last digit and move toward the first digit.
  - Convert the current character to an integer using `x = int(num[i])`.

- **Step 3: Find the rightmost odd digit**
  - Check whether `x % 2 == 1`.
  - If the condition is true, set `flag = 1` and break out of the loop.
  - Since the scan moves backward, the first odd digit found is the rightmost odd digit.

- **Step 4: Handle the absence of odd digits**
  - Initialize `result = ""`.
  - If `flag == 0`, no odd digit was found.
  - Return the empty string.

- **Step 5: Return the largest odd substring**
  - Return `num[:i+1]`.
  - Python slicing excludes the ending index, so `i + 1` includes the odd digit at index `i`.
  - The resulting substring ends with an odd digit and contains the longest possible prefix that satisfies the condition.

**Complexity:**

- **Brute Force:** Time Complexity: O(N³) in a straightforward implementation that generates and checks all substrings  Space Complexity: O(N²) if all substrings are stored.
- **Optimal Approach:** Time Complexity: O(N) in the worst case  Space Complexity: O(1) auxiliary space, excluding the returned substring.
- The optimal approach scans the string at most once. Python creates a new substring for the returned result, which can require O(N) space.

**Common Mistakes:**

- Scanning from left to right and returning the first odd digit. The rightmost odd digit gives the largest valid prefix.
- Returning only the odd digit instead of the prefix ending at that digit.
- Using `num[:i]` instead of `num[:i+1]`. Python slicing excludes the ending index.
- Forgetting to return `""` when all digits are even.
- Converting the entire string to an integer. The input can represent a number too large for ordinary integer types in some languages, and conversion is unnecessary.
- Comparing substrings by length alone without recognizing that the valid substring with the rightmost odd ending digit is the required answer.

**Recognition Pattern:**

- Problems where a number is represented as a string.
- Problems where validity depends only on the final digit.
- Problems asking for the largest valid prefix or substring.
- Use a right-to-left scan when the rightmost position determines whether a prefix is valid.

**Key Observation:**

- A number is odd exactly when its last digit is odd.
- The rightmost odd digit determines the longest prefix that ends in an odd digit.
- Any substring extending beyond that position would end with an even digit, making it invalid.
- Therefore, scanning backward and returning `num[:i+1]` finds the answer in linear time.

**Code:**

```python
class Solution(object):
    def largestOddNumber(self, num):
        """
        :type num: str
        :rtype: str
        """
        n = len(num)
        i = 0
        flag = 0
        for i in range(n-1,-1,-1):
            x = int(num[i])
            if x % 2 == 1:
                flag = 1
                break
        result = ""
        if flag == 0:
            return result
        return num[:i+1]
```

#### LeetCode 1021 - Remove Outermost Parentheses

**Question:**

A valid parentheses string is called primitive if it cannot be split into two non-empty valid parentheses strings.

Given a valid parentheses string `s`, remove the outermost parentheses from every primitive substring and return the resulting string.

**Example:**

- Input: `s = "(()())(())"`
- Output: `"()()()"`
- Explanation:
  - Primitive strings: `"(()())"` and `"(())"`.
  - Remove the outermost parentheses from `"(()())"` to obtain `"()()"`.
  - Remove the outermost parentheses from `"(())"` to obtain `"()"`.
  - Concatenate the results to obtain `"()()()"`.

**Algorithm:**

- Use a depth counter to track the current nesting level of parentheses.
- Every time an opening parenthesis `(` is encountered, increment `depth`.
- Every time a closing parenthesis `)` is encountered, decrement `depth`.
- Skip the opening parenthesis that changes the depth from `0` to `1`.
- Skip the closing parenthesis that changes the depth from `1` to `0`.
- Append every other parenthesis to the result.
- This removes the outermost pair from each primitive without needing a stack.

**Brute Force:**

- Traverse the string and use a stack to collect the characters of the current primitive.
- Increment an opening-parenthesis counter when `(` is encountered.
- Increment a closing-parenthesis counter when `)` is encountered.
- When the two counters become equal and nonzero, a complete primitive has been found.
- Append `stack[1:-1]` to the result to remove the first and last parentheses.
- Reset the stack and counters before processing the next primitive.
- Return the concatenated result.

**Optimal Approach:**

- **Step 1: Handle the empty string**
  - If `s == ""`, return `s`.
  - No processing is required for an empty input.

- **Step 2: Initialize variables**
  - Set `result = ""` to store the answer.
  - Set `depth = 0` to track the nesting level.

- **Step 3: Process each character**
  - Iterate through every character `i` in `s`.
  - Store the previous depth using `prev = depth`.
  - If `i == "("`, increment `depth`.
  - If `i == ")"`, decrement `depth`.

- **Step 4: Skip the outermost opening parenthesis**
  - If `prev == 0` and `depth == 1`, the current character opens a new primitive.
  - Skip it because it is the outermost opening parenthesis.

- **Step 5: Skip the outermost closing parenthesis**
  - If `prev == 1` and `depth == 0`, the current character closes the current primitive.
  - Skip it because it is the outermost closing parenthesis.

- **Step 6: Append all other characters**
  - If neither outermost condition is satisfied, append the current character to `result`.
  - These parentheses belong to the interior of a primitive.

- **Step 7: Return the result**
  - Return `result` after processing the entire string.

**Complexity:**

- **Brute Force:** Time Complexity: O(N)  Space Complexity: O(N) auxiliary space for the stack and result.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space, excluding the output.
- The optimal approach visits each character once and uses only a constant number of tracking variables. The output requires O(N) space.
- In Python, repeated string concatenation can create additional copying overhead. Using a list with `append()` followed by `"".join(result)` preserves the intended O(N) time complexity more reliably.

**Common Mistakes:**

- Removing parentheses based only on their character positions instead of tracking nesting depth.
- Skipping every opening or closing parenthesis. Only the outermost pair of each primitive should be removed.
- Forgetting to save `prev = depth` before updating the depth, which makes it harder to identify the exact transitions.
- Skipping the transitions `0 → 1` and `1 → 0` incorrectly. These are precisely the outermost opening and closing parentheses.
- Forgetting that a primitive ends when the depth returns to zero.
- Assuming a stack is necessary. A depth counter is sufficient because the input is guaranteed to be a valid parentheses string.
- Using `result = result + "".join(i)` unnecessarily. Since `i` is already a single-character string, `result += i` is simpler; a list and `join()` are more efficient for larger inputs.

**Recognition Pattern:**

- Problems involving balanced parentheses and nesting levels.
- Problems requiring removal of the outermost layer of each balanced group.
- Problems where the input is guaranteed to be a valid parentheses string.
- Use a depth counter when only the current nesting level matters and individual opening positions do not need to be stored.

**Key Observation:**

- Each primitive starts when the depth changes from `0` to `1` and ends when it changes from `1` to `0`.
- These two transitions identify exactly the outermost opening and closing parentheses.
- Every other parenthesis belongs to the interior of a primitive and must be preserved.
- Therefore, tracking depth is enough to solve the problem in one pass without a stack.

**Code:**

```python
class Solution(object):
    def removeOuterParentheses(self, s):
        """
        :type s: str
        :rtype: str
        """
        if s == "":
            return s
        result = ""
        depth = 0
        for i in s:
            prev = depth
            if i =="(":
                depth += 1
            if i == ")":
                depth -= 1
            if depth == 1 and prev == 0:
                continue
            if depth == 0 and prev == 1:
                continue
            result = result + "".join(i) 
        return result

# Brute Force
class Solution:
    def removeOuterParentheses(self, s: str) -> str:
        if s == "":
            return s
        stack = []
        result = ""
        start,end = 0,0
        for i in s:
            if i == "(":
                stack.append(i)
                start += 1
            elif i == ")":
                stack.append(i)
                end += 1
            if start != 0 and start == end:
                result = result + "".join(stack[1:-1])
                stack = []
                start,end = 0,0
        return result
```

#### LeetCode 14 - Longest Common Prefix

**Question:**

Write a function to find the longest common prefix string amongst an array of strings.

If there is no common prefix, return an empty string `""`.

**Example:**

- Input: `strs = ["flower", "flow", "flight"]`
- Output: `"fl"`
- Explanation:
  - All strings begin with `"fl"`.
  - The next character differs between `"flower"` and `"flight"`.
  - Therefore, the longest common prefix is `"fl"`.

- Input: `strs = ["dog", "racecar", "car"]`
- Output: `""`
- Explanation:
  - The strings do not share a common starting character.

**Algorithm:**

- **Brute Force:** Find the length of the shortest string, then compare characters at each position across all strings.
- **Optimal Approach:** Sort the strings lexicographically and compare only the first and last strings.
- After sorting, the first and last strings represent the extreme strings in lexicographical order.
- Their longest common prefix is also the longest common prefix shared by the entire array.
- Compare corresponding characters until they differ or one of the two strings ends.

**Brute Force:**

- Initialize `result = ""`.
- Find the length of the shortest string in `strs`.
- Iterate through each character position up to that minimum length.
- For each position, take the character from the first string as the reference.
- Compare that character with the character at the same position in every other string.
- If any character differs, return `result` immediately.
- If all strings have the same character at that position, append it to `result`.
- Return `result` after all positions have been checked.

**Optimal Approach:**

- **Step 1: Handle an empty list**
  - If `strs` is empty, return `""`.
  - This prevents accessing elements that do not exist.

- **Step 2: Sort the strings**
  - Execute `strs.sort()`.
  - Sorting arranges strings in lexicographical order.
  - After sorting, the first and last strings provide the two extreme strings to compare.

- **Step 3: Select the first and last strings**
  - Set `first = strs[0]` and `last = strs[-1]`.
  - The longest common prefix of these two strings is the longest common prefix shared by all strings.

- **Step 4: Compare corresponding characters**
  - Iterate from index `0` to `min(len(first), len(last)) - 1`.
  - If `first[i] != last[i]`, return `"".join(ans)` immediately.
  - The first mismatch marks the end of the common prefix.

- **Step 5: Store matching characters**
  - If the characters match, append `first[i]` to `ans`.
  - The list stores the common prefix characters found so far.

- **Step 6: Return the result**
  - If a mismatch occurs, join `ans` and return the prefix.
  - If the loop completes, the entire shorter string is a prefix of the longer string.
  - Return `"".join(ans)`.

**Complexity:**

- **Brute Force:** Time Complexity: O(N × L)  Space Complexity: O(1) auxiliary space, excluding the output.
- **Optimal Approach:** Time Complexity: O(N log N × L) in the worst case  Space Complexity: O(N × L) for sorting-related storage in a typical implementation, excluding the returned prefix.
- Here, `N` is the number of strings and `L` is the maximum string length.
- The brute-force approach compares at most `L` character positions across `N` strings.
- The sorting approach takes O(N log N) string comparisons, with each comparison potentially examining O(L) characters. Python's sorting implementation may also use additional temporary storage.
- The optimal approach is useful for its simple comparison logic, but it is not always faster than the brute-force approach because sorting has an additional cost.

**Common Mistakes:**

- Forgetting to handle an empty list before accessing `strs[0]`.
- Comparing characters beyond the length of the shorter string, which can cause an index-out-of-range error.
- Returning the entire first string without checking whether the last string shares all its characters.
- Appending characters after the first mismatch instead of stopping immediately.
- Assuming sorting preserves the original order. `strs.sort()` modifies the input list in place.
- Using the minimum string length incorrectly. The comparison must stop when either selected string ends.
- Assuming the sorting approach always has better time complexity than comparing all strings directly.

**Recognition Pattern:**

- Problems involving shared prefixes across multiple strings.
- Problems where the answer must match the beginning of every string.
- Sorting can help reduce comparisons when the lexicographically smallest and largest elements provide useful bounds.
- A direct character-by-character scan is often preferable when modifying or sorting the input is undesirable.

**Key Observation:**

- The longest common prefix shared by all strings must also be shared by the lexicographically smallest and largest strings.
- If those two strings differ at a position, no prefix extending beyond that position can be common to every string.
- Therefore, comparing only the first and last strings after sorting is sufficient to determine the answer.
- The brute-force alternative avoids sorting by comparing each character position across all strings.

**Code:**

```python
class Solution(object):
    def longestCommonPrefix(self, strs):
        """
        :type strs: List[str]
        :rtype: str
        """
        if not strs:
            return ""
        strs.sort()
        first,last = strs[0],strs[-1]
        ans = []
        for i in range(min(len(first),len(last))):
            if first[i] != last[i]:
                return ''.join(ans)
            ans.append(first[i])
        return ''.join(ans)
            
# Brute force
class Solution(object):
    def longestCommonPrefix(self, strs):
        result = ""
        n = float('inf')
        m = len(strs)
        for i in range(m):
            x = len(strs[i])
            if x < n:
                n = x
        for i in range(n):
            count = 0
            x = strs[0][i] 
            for j in range (m):
                if strs[j][i] == x:
                    count += 1
                else:
                    return result
            if count == m:
                result = result + "".join(x)
        return result
```

#### LeetCode 205 - Isomorphic Strings

**Question:**

Given two strings `s` and `t`, determine whether they are isomorphic.

Two strings are isomorphic if the characters in `s` can be replaced to obtain `t`, while preserving the order of characters.

Each character in `s` must map to exactly one character in `t`, and no two different characters in `s` can map to the same character in `t`. A character may map to itself.

**Example:**

- Input: `s = "egg"`, `t = "add"`
- Output: `True`
- Explanation:
  - `e → a`
  - `g → d`
  - Both occurrences of `g` map to `d`, so the mapping is consistent.

- Input: `s = "foo"`, `t = "bar"`
- Output: `False`
- Explanation:
  - The first `o` maps to `a`.
  - The second `o` maps to `r`.
  - The same character cannot map to two different characters.

- Input: `s = "badc"`, `t = "baba"`
- Output: `False`
- Explanation:
  - Different characters in `s` would map to the same character in `t`, violating the one-to-one mapping requirement.

**Algorithm:**

- Use two dictionaries to maintain a one-to-one mapping between characters in `s` and `t`.
- `ref` stores the mapping from characters in `s` to characters in `t`.
- `count` tracks characters in `t` that have already been assigned to a character in `s`.
- For each pair of characters at the same index:
  - If the character from `s` has not been mapped, assign it to the current character from `t`.
  - Ensure the target character from `t` has not already been assigned to a different source character.
  - If the source character already has a mapping, verify that it matches the current target character.
- Return `False` immediately if either mapping rule is violated.
- If every character pair satisfies the rules, return `True`.

**Brute Force:**

- For every pair of positions `i` and `j`, check whether `s[i] == s[j]` if and only if `t[i] == t[j]`.
- If the equality relationship differs for any pair, the strings are not isomorphic.
- If all pairs satisfy the condition, return `True`.
- This approach directly checks whether repeated-character patterns match, but it requires comparing every pair of positions.

**Optimal Approach:**

- **Step 1: Initialize dictionaries**
  - Set `ref = {}` to store mappings from `s` to `t`.
  - Set `count = {}` to track characters already used as mapping targets in `t`.

- **Step 2: Iterate through corresponding characters**
  - Loop through the indices of `s`.
  - At every index `i`, compare `s[i]` with `t[i]`.

- **Step 3: Handle a new source character**
  - If `s[i] not in ref`, the source character has not been mapped before.
  - Assign `ref[s[i]] = t[i]`.
  - Check whether `t[i]` already exists in `count`.
  - If it does, another source character has already claimed this target character, so return `False`.
  - Otherwise, record `count[t[i]] = 1`.

- **Step 4: Validate an existing mapping**
  - If `s[i]` already exists in `ref`, check whether `ref[s[i]] == t[i]`.
  - If they differ, the same source character is mapping to multiple target characters.
  - Return `False` if the mapping is inconsistent.

- **Step 5: Return the result**
  - If the loop completes without finding a violation, return `True`.

**Complexity:**

- **Brute Force:** Time Complexity: O(N²)  Space Complexity: O(1) auxiliary space if character comparisons use no extra data structures and the alphabet is fixed.
- **Optimal Approach:** Time Complexity: O(N) average  Space Complexity: O(1) auxiliary space for a fixed-size character set, or O(K) for K distinct characters.
- The dictionary-based solution processes each character once and performs average O(1) dictionary operations.

**Common Mistakes:**

- Checking only the mapping from `s` to `t`. This allows two different source characters to map to the same target character.
- Forgetting to validate an existing mapping. Every occurrence of a source character must map to the same target character.
- Using a set for source characters without storing their corresponding target characters.
- Returning `True` immediately after finding one valid mapping instead of checking the entire string.
- Forgetting that repeated characters must preserve their pattern in both strings.
- Assuming that `count` stores occurrence frequencies. In this implementation, it only records whether a target character has already been used.
- Not checking that both strings have equal lengths. The problem guarantees equal lengths, but a reusable implementation should validate this if arbitrary inputs are possible.

**Recognition Pattern:**

- Problems involving one-to-one character mappings.
- Problems where repeated characters must preserve the same pattern across two strings.
- Problems requiring both consistency and uniqueness of mappings.
- Use two hash maps when mappings must be valid in both directions.

**Key Observation:**

- Isomorphic strings require a consistent mapping from `s` to `t` and an injective mapping, meaning two different source characters cannot map to the same target character.
- `ref` ensures that each source character always maps to the same target character.
- `count` prevents multiple source characters from claiming the same target character.
- Therefore, checking both conditions at every index is sufficient to establish isomorphism.

**Code:**

```python
class Solution(object):
    def isIsomorphic(self, s, t):
        """
        :type s: str
        :type t: str
        :rtype: bool
        """
        ref = {}
        count = {}
        for i in range(len(s)):
            if s[i] not in ref:
                ref[s[i]] = t[i]
                if t[i] not in count:
                    count[t[i]] = 1
                else:
                    return False
            else:
                if ref[s[i]] != t[i]:
                    return False
        return True
```

#### LeetCode 796 - Rotate String

**Question:**

Given two strings `s` and `goal`, return `True` if and only if `s` can become `goal` after some number of left-shift rotations.

A left rotation moves the first character of a string to the end.

**Example:**

- Input: `s = "abcde"`, `goal = "cdeab"`
- Output: `True`
- Explanation:
  - Rotate `"abcde"` once → `"bcdea"`.
  - Rotate `"bcdea"` again → `"cdeab"`.
  - Therefore, `goal` can be obtained by rotating `s`.

- Input: `s = "abcde"`, `goal = "abced"`
- Output: `False`
- Explanation:
  - No rotation of `"abcde"` produces `"abced"`.

**Algorithm:**

- **Brute Force:** Repeatedly rotate the string one position to the left and check whether it matches `goal`.
- **My Approach:** Concatenate `s` with itself to generate every possible rotation as a substring, then use a sliding window of length `len(goal)` to search for `goal`.
- **Optimal Approach:** Check whether `goal` is a substring of `s + s`, after verifying that both strings have the same length.
- Concatenating `s` with itself contains every possible rotation of `s` as a substring of length `len(s)`.

**Brute Force:**

- First, check whether `s == goal`. If so, return `True`.
- Handle the empty string and verify that both strings have equal lengths.
- Create the first rotation using `s[1:] + s[0]`.
- Compare each rotation with `goal`.
- If a rotation matches `goal`, return `True`.
- Otherwise, continue rotating the string one position to the left.
- Stop when the rotation returns to the original string.
- If no rotation matches, return `False`.

**Optimal Approach:**

- **Step 1: Concatenate the string**
  - Create `temp = s + s`.
  - For example, if `s = "abcde"`, then `temp = "abcdeabcde"`.

- **Step 2: Check the lengths**
  - If `len(s) != len(goal)`, return `False`.
  - Rotations preserve the length of the original string, so strings of different lengths cannot match.

- **Step 3: Search for the target**
  - Check whether `goal` exists inside `s + s`.
  - If it does, return `True`.
  - Otherwise, return `False`.

- **Why does this work?**
  - Every rotation of `s` can be represented by taking a substring of length `len(s)` from `s + s`.
  - For example, `"cdeab"` appears in `"abcdeabcde"`.
  - Therefore, substring membership is sufficient to determine whether `goal` is a valid rotation.

**My Approach:**

- **Step 1: Create the doubled string**
  - Execute `temp = s + "".join(s)`.
  - Since joining the characters of `s` reconstructs `s`, this is equivalent to `temp = s + s`.

- **Step 2: Initialize the sliding window**
  - Set `start = 0` and `end = len(goal)`.
  - The window represents a substring of length `len(goal)`.

- **Step 3: Handle special cases**
  - Return `True` if `s == goal`.
  - Return `False` for an empty `s` that does not match `goal`.
  - Return `False` if the lengths differ.
  - The length check is necessary before searching for rotations.

- **Step 4: Slide the window**
  - Extract `x = temp[start:end]`.
  - If `x == goal`, return `True`.
  - Otherwise, increment both `start` and `end` by `1`.
  - Continue checking while `end < total_len`.

- **Step 5: Return the result**
  - If no window matches `goal`, return `False`.

- **Important improvement:**
  - Your sliding-window solution works, but the condition `while end < total_len` can miss the final possible window when `end == total_len`.
  - For example, if `s = "abc"` and `goal = "cab"`, the matching substring is `temp[2:5]`, which ends exactly at `len(temp)`.
  - The correct condition is `while end <= total_len`.
  - Alternatively, use the simpler substring check `goal in (s + s)` after validating the lengths.

**Complexity:**

- **Brute Force:** Time Complexity: O(N²)  Space Complexity: O(N) for storing rotated strings.
- **My Approach:** Time Complexity: O(N²) in the worst case  Space Complexity: O(N) for the doubled string and temporary substrings.
- **Optimal Approach:** Time Complexity: O(N²) worst case for Python's substring membership check  Space Complexity: O(N) for the doubled string.
- Here, `N` is the length of the input strings.
- The substring-checking approach is the simplest implementation. Its actual performance depends on the string-search algorithm used by the language runtime.

**Common Mistakes:**

- Forgetting to check whether the strings have equal lengths before checking rotations.
- Using only `s + s` without validating the lengths. A shorter `goal` could otherwise match a substring even though it is not a valid rotation.
- Using `while end < total_len` instead of `while end <= total_len`, which skips the final possible window.
- Rotating the string repeatedly when concatenation can represent every rotation at once.
- Forgetting that `s + "".join(s)` is simply `s + s`; the `join()` call is unnecessary.
- Returning `False` immediately when `s != goal`. A different string may still be a valid rotation.
- Mishandling empty strings. Under the standard rotation definition, two empty strings are equal and should return `True`.

**Recognition Pattern:**

- Problems involving cyclic shifts or rotations of strings.
- Problems where a valid answer is a rearrangement caused only by moving a prefix to the end.
- Problems where concatenating a sequence with itself can represent all cyclic starting positions.
- Use `goal in (s + s)` when checking whether one string is a rotation of another.

**Key Observation:**

- Every rotation of `s` is a substring of length `len(s)` in `s + s`.
- Checking equal lengths ensures that a matching substring represents a complete rotation rather than a partial match.
- Therefore, the simplest solution is to verify equal lengths and check whether `goal` occurs in `s + s`.
- This avoids explicitly generating each rotation.

**Code:**

```python
class Solution(object):
    def rotateString(self, s, goal):
        """
        :type s: str
        :type goal: str
        :rtype: bool
        """
        temp = s + "".join(s)
        start,end = 0,len(goal)
        n,total_len = len(s),len(temp)
        if s == goal:
            return True
        if s == "":
            return False
        if len(s) != len(goal):
            return False
        while end < total_len:
            x = temp[start:end]
            if x == goal:
                return True
            start += 1
            end += 1
        return False
        
# simplest optimal Solution
class Solution(object):
    def rotateString(self, s, goal):
        if len(s) != len(goal):
            return False
        return goal in (s + s)
            
# Brute Force
class Solution(object):
    def rotateString(self, s, goal):
        x = s[1:] + s[0]
        if s == goal:
            return True
        if s == "":
            return False
        if len(s) == 1 and len(s) == len(goal):
            return False
        while s != x:
            if x == goal:
                return True
            x = x[1:] + x[0]
        return False
```

#### LeetCode 242 - Valid Anagram

**Question:**

Given two strings `s` and `t`, return `True` if `t` is an anagram of `s`. Otherwise, return `False`.

An anagram is a string formed by rearranging the characters of another string, using every character the same number of times.

**Example:**

- Input: `s = "anagram"`, `t = "nagaram"`
- Output: `True`
- Explanation:
  - Both strings contain the same characters with identical frequencies.
  - The characters can be rearranged to transform one string into the other.

- Input: `s = "rat"`, `t = "car"`
- Output: `False`
- Explanation:
  - The strings contain different characters.
  - Therefore, `t` is not an anagram of `s`.

**Algorithm:**

- Use a hash map to count the frequency of each character in `s`.
- If the strings have different lengths, return `False` immediately.
- Traverse `t` and decrease the corresponding character frequency.
- If a character in `t` does not exist in the map or its frequency is already zero, return `False`.
- If every character in `t` can be matched with an available occurrence, return `True`.

**Brute Force:**

- Sort both strings.
- Compare the sorted strings.
- If they are equal, return `True`; otherwise, return `False`.
- Sorting puts identical characters in the same order, so two anagrams produce identical sorted strings.

**Optimal Approach:**

- **Step 1: Check the lengths**
  - If `len(s) != len(t)`, return `False`.
  - Anagrams must contain exactly the same number of characters.

- **Step 2: Initialize a frequency dictionary**
  - Create `data_s = {}`.
  - This dictionary stores each character from `s` as a key and its frequency as the corresponding value.

- **Step 3: Count characters in `s`**
  - Iterate through every character `x` in `s`.
  - If `x` is not in `data_s`, initialize `data_s[x] = 1`.
  - Otherwise, increment its frequency using `data_s[x] += 1`.

- **Step 4: Match characters from `t`**
  - Iterate through every character `y` in `t`.
  - If `y` does not exist in `data_s`, return `False`.
  - If `data_s[y] == 0`, return `False` because all available occurrences of that character have already been used.
  - Otherwise, decrement its frequency using `data_s[y] -= 1`.

- **Step 5: Return the result**
  - If the entire string `t` is processed without finding an invalid character or exhausted frequency, return `True`.
  - Since the strings have equal lengths, every character occurrence in `s` has been matched exactly once.

**Complexity:**

- **Brute Force:** Time Complexity: O(N log N)  Space Complexity: O(N) for the sorted strings, depending on the sorting implementation.
- **Optimal Approach:** Time Complexity: O(N) average  Space Complexity: O(K), where `K` is the number of distinct characters.
- The optimal approach processes each string once and performs average O(1) dictionary operations per character.
- If the character set is fixed and bounded, the auxiliary space is O(1).

**Common Mistakes:**

- Forgetting to check whether the strings have equal lengths.
- Checking only whether characters exist in both strings without counting their frequencies.
- Forgetting to return `False` when a character's frequency becomes zero before all occurrences in `t` have been processed.
- Decrementing a frequency without first verifying that the character exists in the dictionary.
- Returning `True` immediately after matching one character instead of processing the entire string.
- Assuming that having the same set of characters is sufficient. For example, `"aab"` and `"abb"` contain the same distinct characters but have different frequencies.
- Forgetting that the final frequency values do not need to be checked separately in this implementation. Equal lengths and successful matching of every character in `t` guarantee that all counts are exhausted.

**Recognition Pattern:**

- Problems involving rearrangements of strings.
- Problems where character order does not matter but character frequencies do.
- Problems requiring detection of duplicate or missing character occurrences.
- Use a frequency dictionary when you need to count occurrences and compare two collections of characters efficiently.

**Key Observation:**

- Two strings are anagrams if and only if they have the same character frequencies.
- The dictionary records the available occurrences from `s`.
- Every character in `t` consumes one occurrence from that dictionary.
- If a character is missing or its frequency is exhausted, the strings cannot be anagrams.
- Equal lengths ensure that successfully matching every character in `t` also accounts for every character in `s`.

**Code:**

```python
class Solution(object):
    def isAnagram(self, s, t):
        """
        :type s: str
        :type t: str
        :rtype: bool
        """
        if len(s) != len(t):
            return False
        data_s = {}
        for x in s:
            if x not in data_s:
                data_s[x] = 1
            else:
                data_s[x] += 1
        for y in t:
            if y not in data_s or data_s[y] == 0:
                return False
            else:
                data_s[y] -= 1
        return True

# interview solution
class Solution(object):
    def isAnagram(self, s, t):
        if len(s) != len(t):
            return False

        freq = {}

        for c in s:
            freq[c] = freq.get(c, 0) + 1

        for c in t:
            if c not in freq or freq[c] == 0:
                return False
            freq[c] -= 1

        return True 
```

#### LeetCode 1614 - Maximum Nesting Depth of the Parentheses

**Question:**

Given a valid parentheses string `s`, return its maximum nesting depth.

The nesting depth is the maximum number of parentheses pairs that are nested inside one another at any point in the string.

**Example:**

- Input: `s = "(1+(2*3)+((8)/4))+1"`
- Output: `3`
- Explanation:
  - The substring `(2*3)` has nesting depth `2`.
  - The substring `((8)/4)` contains three nested opening parentheses at its deepest point.
  - Therefore, the maximum nesting depth is `3`.

- Input: `s = "(1)+((2))+(((3)))"`
- Output: `3`
- Explanation:
  - The deepest part, `(((3)))`, contains three nested pairs of parentheses.

**Algorithm:**

- Use a counter to track the current nesting depth.
- Increment the counter whenever an opening parenthesis `(` is encountered.
- Update the maximum depth whenever the counter increases.
- Decrement the counter whenever a closing parenthesis `)` is encountered.
- Ignore all non-parenthesis characters.
- Return the maximum depth recorded.

**Brute Force:**

- Traverse the string and maintain a counter for the current nesting depth.
- Increment the counter for every opening parenthesis.
- Decrement it for every closing parenthesis.
- Store the maximum value reached by the counter.
- Return the maximum value after processing the string.
- This is already a linear-time approach; there is no need to generate substrings or use a stack.

**Optimal Approach:**

- **Step 1: Initialize variables**
  - Set `count = 0` to track the current nesting depth.
  - Set `result = 0` to store the maximum depth encountered.

- **Step 2: Traverse the string**
  - Iterate through every character `i` in `s`.

- **Step 3: Handle opening parentheses**
  - If `i == "("`, increment `count` by `1`.
  - Compare `count` with `result`.
  - If `result < count`, update `result = count`.
  - This ensures that the maximum depth is updated whenever a deeper nesting level is reached.

- **Step 4: Handle closing parentheses**
  - If `i == ")"`, decrement `count` by `1`.
  - This returns the depth to the level outside the closed pair.

- **Step 5: Ignore other characters**
  - Characters such as digits, letters, operators, and spaces do not affect nesting depth.

- **Step 6: Return the answer**
  - Return `result`, which contains the maximum nesting depth encountered during the traversal.

**Complexity:**

- **Brute Force:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space.
- Both approaches process each character exactly once and maintain only two integer variables.

**Common Mistakes:**

- Updating the maximum depth before incrementing the counter for an opening parenthesis. The depth must first increase.
- Forgetting to decrement the counter when encountering a closing parenthesis.
- Updating the maximum depth when processing a closing parenthesis instead of when opening a new nesting level.
- Counting every parenthesis rather than tracking the current nesting level.
- Using a stack unnecessarily when the input is guaranteed to be a valid parentheses string.
- Assuming that the number of opening parentheses in the entire string equals the maximum depth. Sequential, non-nested pairs do not increase the maximum beyond `1`.

**Recognition Pattern:**

- Problems involving nested parentheses or balanced delimiters.
- Problems asking for the maximum nesting level or depth.
- Problems where only the current number of unmatched opening parentheses matters.
- Use a counter instead of a stack when you only need the nesting depth and the input is guaranteed to be valid.

**Key Observation:**

- The current depth equals the number of opening parentheses encountered minus the number of closing parentheses encountered so far.
- Every opening parenthesis increases the depth by one, while every closing parenthesis decreases it by one.
- The maximum depth can only increase when an opening parenthesis is processed.
- Therefore, tracking the current depth and its maximum is sufficient to solve the problem in one pass.

**Code:**

```python
# interview solution
class Solution(object):
    def maxDepth(self, s):
        """
        :type s: str
        :rtype: int
        """
        depth = 0
        ans = 0

        for ch in s:
            if ch == '(':
                depth += 1
                ans = max(ans, depth)
            elif ch == ')':
                depth -= 1

        return ans

# optimal solution
class Solution(object):
    def maxDepth(self, s):
        count,result = 0,0
        for i in s:
            if i == "(":
                count += 1
                if result < count:
                    result = count
            elif i == ")":
                count -= 1
        return result   
```

#### LeetCode 13 - Roman to Integer

**Question:**

Given a Roman numeral string `s`, convert it into an integer.

Roman numerals use the following symbols:

- `I = 1`
- `V = 5`
- `X = 10`
- `L = 50`
- `C = 100`
- `D = 500`
- `M = 1000`

Normally, values are added together. However, when a smaller value appears before a larger value, the smaller value is subtracted instead of added.

**Example:**

- Input: `s = "III"`
- Output: `3`
- Explanation:
  - `I + I + I = 1 + 1 + 1 = 3`.

- Input: `s = "LVIII"`
- Output: `58`
- Explanation:
  - `L + V + I + I + I = 50 + 5 + 1 + 1 + 1 = 58`.

- Input: `s = "MCMXCIV"`
- Output: `1994`
- Explanation:
  - `M = 1000`
  - `CM = 900`
  - `XC = 90`
  - `IV = 4`
  - Total = `1000 + 900 + 90 + 4 = 1994`.

**Algorithm:**

- Create a dictionary that maps every Roman numeral symbol to its integer value.
- Traverse the string from left to right.
- Compare the current symbol's value with the next symbol's value.
- If the current value is smaller than the next value, subtract the current value.
- Otherwise, add the current value.
- For the final symbol, add its value because there is no next symbol to compare.
- Return the accumulated result.

**Brute Force:**

- Identify the six standard subtractive combinations:
  - `IV = 4`
  - `IX = 9`
  - `XL = 40`
  - `XC = 90`
  - `CD = 400`
  - `CM = 900`
- Scan the string and check whether the current symbol and the next symbol form one of these combinations.
- If a subtractive combination is found, add its combined value and skip both symbols.
- Otherwise, add the value of the current symbol and move forward by one position.
- Return the accumulated integer.
- This approach explicitly handles subtractive pairs, but requires additional conditional checks.

**Optimal Approach:**

- **Step 1: Create the value dictionary**
  - Initialize `x` with the integer value of each Roman numeral symbol.
  - This allows each symbol's value to be retrieved using `x[s[i]]`.

- **Step 2: Initialize the result**
  - Set `result = 0`.
  - Convert `s` into a list using `s = list(str(s))`.
  - Since `s` is already a string, this conversion is unnecessary; you can iterate directly over the string.

- **Step 3: Traverse the symbols**
  - Set `n = len(s)`.
  - Iterate through the indices from `0` to `n - 1`.

- **Step 4: Handle the last symbol**
  - If `i == n - 1`, add `x[s[i]]` to `result`.
  - The last symbol has no following symbol, so it cannot form a subtractive pair with a later symbol.

- **Step 5: Handle subtractive notation**
  - If `x[s[i]] < x[s[i + 1]]`, subtract the current symbol's value from `result`.
  - For example, in `"IV"`, subtract `1` because `I` comes before the larger value `V`.

- **Step 6: Handle normal addition**
  - If the current value is greater than or equal to the next value, add the current value to `result`.
  - For example, in `"VI"`, add `5` and then add `1`.

- **Step 7: Return the result**
  - After processing every symbol, return `result`.

**Complexity:**

- **Brute Force:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space, assuming a fixed-size symbol mapping.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space, assuming a fixed-size symbol mapping.
- Both approaches process each symbol a constant number of times.
- The list conversion in your implementation uses O(N) additional space. Removing `s = list(str(s))` avoids this extra allocation.

**Common Mistakes:**

- Adding every Roman numeral value without checking for subtractive notation.
- Subtracting when the current value is greater than the next value instead of when it is smaller.
- Accessing `s[i + 1]` without checking whether `i` is the last index.
- Forgetting to add the final symbol's value.
- Treating every smaller-before-larger pair as valid without relying on the problem's guarantee that the input is a valid Roman numeral.
- Converting the input string into a list unnecessarily.
- Confusing the direction of subtraction: for `"IV"`, subtract `I` and add `V`, producing `-1 + 5 = 4`.

**Recognition Pattern:**

- Problems involving Roman numeral conversion.
- Problems where the current element's contribution depends on the next element.
- Problems where a smaller value before a larger value changes the normal addition rule.
- Use a left-to-right scan with adjacent comparisons when the next symbol determines whether the current value should be added or subtracted.

**Key Observation:**

- Roman numerals follow an addition rule except for subtractive notation.
- If the current symbol is smaller than the next symbol, its value must be subtracted.
- Otherwise, its value is added.
- The last symbol is always added.
- Therefore, comparing each symbol with its next neighbor is sufficient to convert a valid Roman numeral into an integer in linear time.

**Code:**

```python
class Solution(object):
    def romanToInt(self, s):
        """
        :type s: str
        :rtype: int
        """
        x = {
            'I' : 1,
            'V' : 5,
            'X' : 10,
            'L' : 50,
            'C' :100,
            'D' :500,
            'M':1000
        }
        result = 0
        s = list(str(s))
        n = len(s)
        for i in range(n):
            if i == n-1:
                result += x[s[i]]
            elif x[s[i]] < x[s[i+1]]:
                result -= x[s[i]]
            else:
                result += x[s[i]]
        return result
```

#### LeetCode 125 - Valid Palindrome

**Question:**

Given a string `s`, determine whether it is a palindrome after converting all uppercase letters into lowercase letters and removing all non-alphanumeric characters.

A palindrome reads the same forward and backward.

**Example:**

- Input: `s = "A man, a plan, a canal: Panama"`
- Output: `True`
- Explanation:
  - After removing spaces and punctuation and converting to lowercase, the string becomes `"amanaplanacanalpanama"`.
  - The resulting string reads the same forward and backward.

- Input: `s = "race a car"`
- Output: `False`
- Explanation:
  - After preprocessing, the string becomes `"raceacar"`.
  - It does not read the same forward and backward.

- Input: `s = " "`
- Output: `True`
- Explanation:
  - Removing non-alphanumeric characters produces an empty string.
  - An empty string is considered a palindrome.

**Algorithm:**

- **Brute Force:** Build a cleaned string containing only lowercase alphanumeric characters, then compare it with its reverse.
- **Optimal Approach:** Use two pointers, one at the beginning and one at the end of the string.
- Skip non-alphanumeric characters from both ends.
- Compare the lowercase versions of the characters at the two pointers.
- If they differ, return `False`.
- If they match, move both pointers inward.
- If the pointers cross without finding a mismatch, return `True`.

**Brute Force:**

- Initialize an empty string named `result`.
- Traverse every character in `s`.
- If a character satisfies `isalnum()`, convert it to lowercase and append it to `result`.
- Reverse the cleaned string using `result[::-1]`.
- If the reversed string equals the original cleaned string, return `True`.
- Otherwise, return `False`.

**Optimal Approach:**

- **Step 1: Initialize two pointers**
  - Set `n = len(s)`.
  - Set `low = 0` and `high = n - 1`.
  - `low` points to the first character, while `high` points to the last character.

- **Step 2: Continue while the pointers have not crossed**
  - Use `while low <= high`.
  - The loop compares characters from opposite ends of the string.

- **Step 3: Skip non-alphanumeric characters from the left**
  - If `not s[low].isalnum()`, increment `low`.
  - Use `continue` to restart the loop without comparing the current character.

- **Step 4: Skip non-alphanumeric characters from the right**
  - If `not s[high].isalnum()`, decrement `high`.
  - Use `continue` to restart the loop without comparing the current character.

- **Step 5: Compare valid characters**
  - Convert both characters to lowercase using `.lower()`.
  - If `s[low].lower() == s[high].lower()`, the characters match.
  - Move `low` forward and `high` backward.
  - If they differ, return `False` immediately.

- **Step 6: Return the result**
  - If the loop finishes without finding a mismatch, return `True`.
  - This means the string is a palindrome after ignoring case and non-alphanumeric characters.

**Complexity:**

- **Brute Force:** Time Complexity: O(N)  Space Complexity: O(N) for the cleaned string and its reversed representation.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space.
- Here, `N` is the length of the input string.
- The two-pointer approach processes each character at most a constant number of times without creating a cleaned copy of the string.

**Common Mistakes:**

- Forgetting to ignore spaces and punctuation.
- Comparing characters without converting them to the same case.
- Moving both pointers when only one points to a non-alphanumeric character.
- Forgetting `continue` after skipping an invalid character, which could lead to comparing a character that should have been ignored.
- Using `low < high` is also valid for this problem, but `low <= high` additionally checks the middle character when the string has odd length.
- Creating a cleaned string when constant auxiliary space is desired.
- Assuming an empty cleaned string is not a palindrome. It should return `True`.

**Recognition Pattern:**

- Problems involving palindrome validation.
- Problems requiring characters to be compared from opposite ends.
- Problems where some characters must be ignored during comparison.
- Use two pointers when you can process a sequence from both ends without constructing a modified copy.

**Key Observation:**

- A palindrome must have matching characters at symmetric positions.
- Spaces, punctuation, and other non-alphanumeric characters do not participate in the comparison.
- Case differences do not matter because both characters are converted to lowercase.
- The two-pointer approach skips irrelevant characters and compares only meaningful pairs.
- Therefore, the string can be checked in linear time with constant auxiliary space.

**Code:**

```python
class Solution(object):
    def isPalindrome(self, s):
        """
        :type s: str
        :rtype: bool
        """
        n = len(s)
        low,high = 0, n-1
        while low <= high:
            if not s[low].isalnum():
                low += 1
                continue
            if not s[high].isalnum():
                high -= 1
                continue
            if s[low].lower() == s[high].lower():
                low += 1
                high -= 1
            else:
                return False
        return True

# Brute Force Approach
class Solution(object):
    def isPalindrome(self, s):
        n = len(s)
        result = ""
        for i in range(n):
            if s[i].isalnum():
                result += s[i].lower()
        if result[::-1] == result:
            return True
        else:
            return False
```

#### LeetCode 151 - Reverse Words in a String

**Question:**

Given an input string `s`, reverse the order of the words.

A word is a sequence of non-space characters. The output must contain only a single space between adjacent words, with no leading or trailing spaces.

**Example:**

- Input: `s = "  the sky  is blue  "`
- Output: `"blue is sky the"`
- Explanation:
  - Remove the leading and trailing spaces.
  - Reverse the order of the words.
  - Remove extra spaces between words so that only one space separates adjacent words.

- Input: `s = "hello world"`
- Output: `"world hello"`

- Input: `s = "a good   example"`
- Output: `"example good a"`

**Algorithm:**

- **Brute Force:** Extract each word from left to right, store the words in a stack, and then pop or traverse them in reverse order to build the result.
- **Optimal Approach:** Traverse the string from right to left and extract one word at a time.
- Skip spaces before each word.
- Identify the start and end boundaries of the word.
- Append the word to the result, adding a single space between words.
- This avoids storing all the words in a separate stack.

**Brute Force:**

- Initialize an empty string `temp` to build the current word and an empty list `stack` to store words.
- Traverse each character in `s`.
- If the character is not a space, append it to `temp`.
- If the character is a space and `temp` is not empty, append `temp` to `stack` and reset `temp`.
- Ignore consecutive spaces when `temp` is empty.
- After the loop, append the final word if `temp` is not empty.
- Traverse `stack` backward and append each word to `result`, separating adjacent words with a single space.
- Return `result`.

**Optimal Approach:**

- **Step 1: Initialize the result and pointer**
  - Set `result = ""`.
  - Set `i = len(s) - 1` so that traversal starts from the last character.

- **Step 2: Skip trailing or consecutive spaces**
  - While `i >= 0` and `s[i] == " "`, decrement `i`.
  - This skips spaces before searching for the next word.

- **Step 3: Check whether the string has been fully processed**
  - If `i < 0`, break out of the loop.
  - This means there are no more words to extract.

- **Step 4: Record the end of the word**
  - Set `end = i`.
  - This stores the index of the last character of the current word.

- **Step 5: Find the beginning of the word**
  - Move `i` backward while `i >= 0` and `s[i] != " "`.
  - When this loop ends, `i` is either at a space or just before the beginning of the string.
  - The current word occupies the range from `i + 1` through `end`.

- **Step 6: Extract the word**
  - Use `word = s[i + 1:end + 1]`.
  - Python excludes the ending index in slicing, so `end + 1` includes the last character.

- **Step 7: Append the word**
  - If `result != ""`, append a single space to separate the words.
  - Append the extracted word to `result`.
  - Since words are processed from right to left, the resulting string contains the words in reverse order.

- **Step 8: Return the result**
  - Continue until all characters have been processed.
  - Return `result` without leading or trailing spaces.

**Complexity:**

- **Brute Force:** Time Complexity: O(N)  Space Complexity: O(N) for the stack, extracted words, and output.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(N) for the output string, with O(1) auxiliary pointer variables excluding the output.
- Both approaches process the input linearly.
- In Python, repeated string concatenation can introduce additional copying overhead. Building a list of words or characters and joining it at the end can improve performance for large strings.

**Common Mistakes:**

- Forgetting to remove leading and trailing spaces.
- Treating consecutive spaces as separate words.
- Adding multiple spaces between words in the output.
- Using `s[i + 1:end]` instead of `s[i + 1:end + 1]`, which excludes the last character of the word.
- Forgetting to append the final word in the brute-force approach because there may be no trailing space to trigger the word-storage condition.
- Reversing the characters inside each word instead of reversing the order of the words.
- Adding a space before the first word or after the last word.
- Forgetting to check `i >= 0` before accessing `s[i]`, which can cause an index-out-of-range error.

**Recognition Pattern:**

- Problems involving word extraction from strings with irregular spacing.
- Problems requiring words to be reordered while preserving each word's internal character order.
- Problems where scanning from the end naturally produces the required output order.
- Use two nested loops when you need to skip separators and then identify the boundaries of each token.

**Key Observation:**

- Reversing the order of words does not require reversing the entire string character by character.
- By scanning from right to left, each word can be extracted in its original character order and appended directly to the result.
- Skipping spaces before extracting each word automatically handles leading, trailing, and repeated spaces.
- Adding a separator only when the result is non-empty ensures exactly one space between words.
- Therefore, the string can be processed in linear time without a separate stack of words.

**Code:**

```python
# Optimal approach
class Solution(object):
    def reverseWords(self, s):
        """
        :type s: str
        :rtype: str
        """
        result = ""
        i = len(s) - 1
        while i >= 0:
            while i >= 0 and s[i] == " ":
                i -= 1
            if i < 0:
                break
            end = i
            while i >= 0 and s[i] != " ":
                i -= 1
            word = s[i + 1:end + 1]
            if result != "":
                result += " "
            result += word
        return result
            
# Brute Force
class Solution(object):
    def reverseWords(self, s):
        temp = ""
        stack = []
        for i in s:
            if i != " ":
                temp = temp + "".join(i)
            elif i == " ":
                if temp == "":
                    continue
                stack.append(temp)
                temp = ""
        if temp != "":
            stack.append(temp)
        n = len(stack)
        result = ""
        for i in range(n,0,-1):
            result = result + "".join(stack[i - 1])
            if i == 1:
                continue
            else:
                result = result + "".join(" ")
        return result
```

Given a string `s`, sort it in decreasing order based on the frequency of its characters.

The frequency of a character is the number of times it appears in the string.

Return the sorted string. If multiple characters have the same frequency, any order among those characters is acceptable.

**Example:**

- Input: `s = "tree"`

  - Characters with higher frequencies must appear before characters with lower frequencies.

- Input: `s = "cccaaa"`
- Output: `"cccaaa"` or `"aaaccc"`
- Explanation:
  - `'c'` and `'a'` each appear three times.
  - Both groups have the same frequency, so either order is valid.

**Algorithm:**

- Use a hash map to count the frequency of each character.
- Sort the distinct characters by their frequencies in descending order.
- Multiply each character by its frequency to create the required number of repetitions.
- Append the repeated characters to a result list.
- Join the list into a single string and return it.

**Brute Force:**

- Count the frequency of each character using a dictionary.
- Repeatedly find the character with the highest remaining frequency.
- Append that character the required number of times to the result.
- Remove or mark the processed character so that it is not selected again.
- Continue until every distinct character has been processed.
- This approach avoids sorting but can require repeatedly searching for the maximum frequency.

**Optimal Approach:**

- **Step 1: Initialize the frequency dictionary**
  - Set `freq = {}`.
  - The dictionary stores each character as a key and its frequency as the value.

- **Step 2: Count character frequencies**
  - Iterate through every character `ch` in `s`.
  - Use `freq[ch] = freq.get(ch, 0) + 1`.
  - If the character is new, its frequency starts at `1`; otherwise, its frequency increases by `1`.

- **Step 3: Sort characters by frequency**
  - Execute `sorted(freq.items(), key=lambda x: x[1], reverse=True)`.
  - `freq.items()` produces `(character, frequency)` pairs.
  - `key=lambda x: x[1]` sorts according to the frequency.
  - `reverse=True` places the highest frequencies first.

- **Step 4: Build the result**
  - Iterate through each `(ch, cnt)` pair in the sorted list.
  - Calculate `ch * cnt` to repeat the character according to its frequency.
  - Append the repeated string to `result`.

- **Step 5: Return the final string**
  - Execute `"".join(result)`.
  - This combines all repeated-character groups into one string without adding separators.

**Complexity:**

- **Brute Force:** Time Complexity: O(K² + N)  Space Complexity: O(K + N) for the frequency map and output.
- **Optimal Approach:** Time Complexity: O(N + K log K)  Space Complexity: O(N + K) including the output and frequency data.
- Here, `N` is the length of `s` and `K` is the number of distinct characters.
- Counting frequencies takes O(N).
- Sorting the distinct characters takes O(K log K).
- Constructing the result requires O(N) time overall.
- Since `K` cannot exceed `N`, the overall time complexity is O(N log N) in the worst case.

**Common Mistakes:**

- Sorting characters alphabetically instead of sorting by frequency.
- Forgetting `reverse=True`, which would sort frequencies in ascending order.
- Using the character itself as the sorting key instead of its frequency.
- Appending each character only once instead of repeating it `cnt` times.
- Returning the list of repeated strings instead of joining it into a single string.
- Assuming that characters with equal frequencies must appear in a particular order. The problem permits any order for ties.
- Confusing the number of distinct characters `K` with the total number of characters `N`.

**Recognition Pattern:**

- Problems requiring elements to be arranged according to their occurrence counts.
- Problems involving frequency maps followed by sorting.
- Problems where repeated values must be grouped together.
- Use a hash map to count occurrences and sort its entries when the number of distinct elements is manageable.

**Key Observation:**

- The output order depends on character frequency, not the character's alphabetical or original position.
- Once the frequency of each character is known, sorting the distinct characters by descending frequency determines the group order.
- Repeating each character by its frequency ensures that every occurrence from the original string is preserved.
- Both your interview solution and your alternative solution implement the same algorithm; the interview solution is slightly more concise because it unpacks each `(character, frequency)` pair directly in the loop.

**Code:**

```python
# interview solution
class Solution(object):
    def frequencySort(self, s):
        """
        :type s: str
        :rtype: str
        """
        freq = {}
        for ch in s:
            freq[ch] = freq.get(ch, 0) + 1
        result = []
        for ch, cnt in sorted(freq.items(), key=lambda x: x[1], reverse=True):
            result.append(ch * cnt)
        return "".join(result)

# optimal solution
class Solution(object):
    def frequencySort(self, s):
        freq = {}
        for x in s:
            freq[x] = freq.get(x,0) + 1
        result = sorted(freq.items(), key=lambda x:x[1], reverse=True)    
        temp = []
        for i in result:
            x,y = i[0],i[1]
            temp.append(x*y)
        return "".join(temp)         
```

#### LeetCode 8 - String to Integer (atoi)

**Question:**

Implement the `myAtoi(s)` function, which converts a string into a 32-bit signed integer.

The conversion follows these rules:

- Ignore leading whitespace.
- Determine the sign from an optional `+` or `-` character.
- Read consecutive digits until a non-digit character is encountered or the string ends.
- If no digits are found, return `0`.
- Clamp the result to the signed 32-bit integer range: `[-2³¹, 2³¹ - 1]`.

**Example:**

- Input: `s = "42"`
- Output: `42`
- Explanation:
  - The string begins with digits, so the integer is `42`.

- Input: `s = "   -042"`
- Output: `-42`
- Explanation:
  - Leading spaces are ignored.
  - The negative sign is recognized.
  - Leading zeros do not affect the numerical value.

- Input: `s = "4193 with words"`
- Output: `4193`
- Explanation:
  - Conversion stops at the first non-digit character after the digits.

- Input: `s = "words and 987"`
- Output: `0`
- Explanation:
  - The first non-whitespace character is not a digit or sign followed by digits.

- Input: `s = "-91283472332"`
- Output: `-2147483648`
- Explanation:
  - The parsed value is below the minimum 32-bit signed integer.
  - The result is clamped to `-2³¹`.

**Algorithm:**

- Skip leading spaces.
- Determine the sign from an optional `+` or `-`.
- Parse consecutive digits and build the integer one digit at a time.
- After every digit, check whether the signed result exceeds the 32-bit integer bounds.
- Return the clamped value if a bound is reached.
- Stop parsing at the first non-digit character and return the accumulated signed integer.

**Brute Force:**

- Skip leading spaces.
- Read the optional sign.
- Collect consecutive digits into a separate string.
- If no digits are collected, return `0`.
- Convert the digit string into an integer and apply the sign.
- Clamp the result to the 32-bit signed integer range.
- This approach is straightforward in Python, but it constructs an intermediate string and relies on integer conversion rather than parsing the digits incrementally.

**Optimal Approach:**

- **Step 1: Initialize the bounds**
  - Set `INT_MIN = -2**31`.
  - Set `INT_MAX = 2**31 - 1`.
  - These represent the minimum and maximum signed 32-bit integer values.

- **Step 2: Skip leading spaces**
  - Initialize `i = 0`.
  - While `i < len(s)` and `s[i] == ' '`, increment `i`.
  - This moves the pointer to the first non-space character.

- **Step 3: Determine the sign**
  - Initialize `sign = 1`.
  - If the current character is `+` or `-`, set the sign accordingly.
  - Increment `i` to move past the sign.
  - If no sign exists, keep `sign = 1`.

- **Step 4: Initialize recursive parsing**
  - Call `self.helper(s, i, 0, sign)`.
  - The parameters represent:
    - `s`: the input string.
    - `i`: the current character index.
    - `num`: the integer accumulated so far.
    - `sign`: either `1` or `-1`.

- **Step 5: Handle the stopping condition**
  - If `i >= len(s)` or `s[i]` is not a digit, return `sign * num`.
  - This stops parsing at the end of the string or at the first non-digit character.
  - If no digits were parsed, `num` remains `0`, so the result is `0`.

- **Step 6: Build the integer**
  - Calculate `num = num * 10 + int(s[i])`.
  - Multiplying by `10` shifts the existing number one decimal position to the left.
  - Adding the current digit incorporates it into the accumulated number.

- **Step 7: Check the integer bounds**
  - Check whether `sign * num <= INT_MIN`. If true, return `INT_MIN`.
  - Check whether `sign * num >= INT_MAX`. If true, return `INT_MAX`.
  - Otherwise, continue parsing recursively at index `i + 1`.

- **Step 8: Return the result**
  - When parsing stops, return the signed integer.
  - If a bound is reached, return the corresponding clamped limit immediately.

**Complexity:**

- **Brute Force:** Time Complexity: O(N)  Space Complexity: O(N) for the extracted digit string.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(N) auxiliary space in the worst case due to recursion.
- Here, `N` is the length of the input string.
- The recursive approach processes each digit once, but each recursive call adds a stack frame.
- An iterative digit-parsing implementation can achieve O(1) auxiliary space.

**Common Mistakes:**

- Forgetting to skip leading spaces before checking the sign.
- Allowing multiple signs, such as `"--12"` or `"+-12"`, to be interpreted as valid input.
- Continuing to parse digits after encountering a non-digit character.
- Returning `0` whenever a non-digit character is encountered instead of returning the number accumulated so far.
- Forgetting to apply the sign to the parsed number.
- Failing to clamp the result to the signed 32-bit integer range.
- Checking only the positive bound and forgetting that the negative bound is `-2³¹`, whose magnitude is one greater than `2³¹ - 1`.
- Using `>= INT_MAX` as the positive overflow check is safe here because the check occurs after a digit is added: a value equal to `INT_MAX` is clamped to the same value. However, the conventional check is `num > INT_MAX` for the positive case and `num < INT_MIN` for the negative case, or a sign-aware overflow check before multiplying and adding.
- Using recursion for a potentially long digit sequence can cause a recursion-depth error in Python.

**Recognition Pattern:**

- Problems requiring conversion of a string into an integer under strict parsing rules.
- Problems where parsing must stop at the first invalid character.
- Problems involving signs, leading whitespace, and overflow handling.
- Use a digit-by-digit parser when the conversion rules differ from ordinary integer conversion.

**Key Observation:**

- The integer can be constructed incrementally using `num = num * 10 + digit`.
- Leading whitespace and an optional sign must be handled before parsing digits.
- Once a non-digit is encountered, the accumulated number is the final result.
- Overflow must be checked while building the number so that the result never exceeds the permitted 32-bit signed integer range.
- Your recursive helper implements this parsing process, but an iterative loop is generally safer in Python because it avoids recursion-depth limitations.

**Code:**

```python
class Solution:
    def helper(self,s, i, num, sign):
        INT_MIN = -2**31
        INT_MAX = 2**31 - 1
        if i >= len(s) or not s[i].isdigit():
            return sign * num
        num = num * 10 + int(s[i])
        if sign * num <= INT_MIN: return INT_MIN
        if sign * num >= INT_MAX: return INT_MAX
        return self.helper(s, i + 1, num, sign)

    def myAtoi(self,s):
        i = 0
        while i < len(s) and s[i] == ' ':
            i += 1
        sign = 1
        if i < len(s) and (s[i] == '+' or s[i] == '-'):
            sign = -1 if s[i] == '-' else 1
            i += 1
        return self.helper(s, i, 0, sign)
```

#### LeetCode 5 - Longest Palindromic Substring

**Question:**

Given a string `s`, return the longest palindromic substring in `s`.

A palindrome is a string that reads the same forward and backward. A substring must consist of consecutive characters from the original string.

**Example:**

- Input: `s = "babad"`
- Output: `"bab"` or `"aba"`
- Explanation:
  - `"bab"` is a palindrome.
  - `"aba"` is also a palindrome.
  - Both have the maximum length of `3`.

- Input: `s = "cbbd"`
- Output: `"bb"`
- Explanation:
  - `"bb"` is a palindrome of length `2`.

**Algorithm:**

- Use the **expand-around-center** technique to find palindromic substrings.
- Every palindrome has either:
  - A single character as its center, for odd-length palindromes.
  - A gap between two characters as its center, for even-length palindromes.
- For every index `i`, expand around both types of centers.
- While the characters at the left and right pointers match, move the pointers outward.
- When expansion stops, return the palindrome identified by the final boundaries.
- Keep the longest palindrome found so far and return it.

**Brute Force:**

- Generate every possible substring of `s`.
- Check whether each substring is a palindrome by comparing it with its reverse.
- If it is a palindrome and longer than the current result, update the result.
- Return the longest palindrome found.
- There are O(N²) substrings, and checking each substring for palindrome validity can take O(N) time.

**Optimal Approach:**

- **Step 1: Define the expansion helper**
  - `findPalindrome(s, left, right)` expands outward from the given center.
  - The initial pointers determine whether the center is odd-length or even-length.

- **Step 2: Compare the characters**
  - While `left >= 0` and `right < len(s)`, check whether `s[left] == s[right]`.
  - If the characters differ, stop expanding because the substring is no longer a palindrome.
  - Otherwise, decrement `left` and increment `right` to check the next outer pair.

- **Step 3: Return the palindrome**
  - When the loop stops, the pointers have moved one position beyond the valid palindrome boundaries.
  - Return `s[left + 1:right]`.
  - Python slicing excludes the ending index, so this returns the complete palindrome.

- **Step 4: Initialize the answer**
  - Set `result = s[0]`.
  - This assumes the input string is non-empty.
  - For a reusable implementation, handle `s == ""` first by returning `""`.

- **Step 5: Check odd-length palindromes**
  - Iterate through each index `i`.
  - Call `findPalindrome(s, i - 1, i + 1)`.
  - This expands around `s[i]` as the center.
  - If the returned substring is longer than `result`, update `result`.

- **Step 6: Check even-length palindromes**
  - Call `findPalindrome(s, i, i + 1)`.
  - This uses the gap between `s[i]` and `s[i + 1]` as the center.
  - If the returned substring is longer than `result`, update `result`.

- **Step 7: Return the answer**
  - After checking both centers for every index, return `result`.
  - The longest palindrome is found among the expansions from all possible centers.

**Complexity:**

- **Brute Force:** Time Complexity: O(N³)  Space Complexity: O(N) for temporary substrings.
- **Optimal Approach:** Time Complexity: O(N²)  Space Complexity: O(N) including the returned substring, or O(1) auxiliary space excluding the output.
- There are O(N) possible centers, and each expansion can take O(N) time in the worst case.
- The helper does not use an additional data structure proportional to the input size.

**Common Mistakes:**

- Checking only odd-length palindromes. Even-length palindromes such as `"abba"` require a center between two characters.
- Checking only even-length palindromes. Odd-length palindromes such as `"racecar"` require a single-character center.
- Returning `s[left:right]` instead of `s[left + 1:right]`. The pointers move one position beyond the valid palindrome when expansion stops.
- Forgetting that Python slicing excludes the ending index.
- Updating `result` only when the new palindrome is strictly longer. This is valid because any longest palindrome is accepted when multiple answers exist.
- Initializing `result = s[0]` without checking whether the input is empty.
- Confusing a substring with a subsequence. A substring must contain consecutive characters.
- Using the wrong centers: `i - 1, i + 1` for odd-length palindromes and `i, i + 1` for even-length palindromes.

**Recognition Pattern:**

- Problems asking for the longest palindromic substring.
- Problems where a valid sequence expands symmetrically around a center.
- Problems where each character or gap can serve as the center of a candidate solution.
- Use expand-around-center when palindrome validity depends on matching characters on both sides.

**Key Observation:**

- Every non-empty palindrome has a center, either at a character or between two adjacent characters.
- Expanding outward while characters match finds the longest palindrome for that center.
- Checking both odd and even centers covers every possible palindromic substring.
- There are O(N) centers, and each expansion takes at most O(N) time, giving an overall O(N²) solution.
- This is substantially more efficient than generating and validating every substring independently.

**Code:**

```python
class Solution:
    def findPalindrome(self,s,left,right):
        while left >= 0 and right < len(s):
            if s[left] != s[right]:
                break
            else:
                left = left - 1
                right = right + 1
        return s[left + 1:right]

    def longestPalindrome(self, s: str) -> str:
        result = s[0]
        for i in range(len(s)):
            temp = self.findPalindrome(s,i-1,i+1)
            if len(result) < len(temp):
                result = temp
            temp = self.findPalindrome(s,i,i+1)
            if len(result) < len(temp):
               result = temp
        return result
```

#### LeetCode 1781 - Sum of Beauty of All Substrings

**Question:**

The beauty of a string is defined as the difference between the frequency of the most frequent character and the frequency of the least frequent character that appears in the string.

Given a string `s`, return the sum of the beauty of all its substrings.

**Example:**

- Input: `s = "aabcb"`
- Output: `5`
- Explanation:
  - Substring `"aab"`: Frequencies are `a = 2`, `b = 1`. Beauty = `2 - 1 = 1`.
  - Substring `"aabc"`: Frequencies are `a = 2`, `b = 1`, `c = 1`. Beauty = `2 - 1 = 1`.
  - Substring `"aabcb"`: Frequencies are `a = 2`, `b = 2`, `c = 1`. Beauty = `2 - 1 = 1`.
  - Substring `"abcb"`: Frequencies are `a = 1`, `b = 2`, `c = 1`. Beauty = `2 - 1 = 1`.
  - Substring `"bcb"`: Frequencies are `b = 2`, `c = 1`. Beauty = `2 - 1 = 1`.
  - The sum of the beauty values of all substrings is `5`.

**Algorithm:**

- Enumerate every possible substring by fixing its starting index `i`.
- Maintain a frequency dictionary `temp` for the current substring.
- Extend the substring one character at a time by moving the ending index `j`.
- Update the frequency of `s[j]` whenever a character is added.
- Calculate the maximum and minimum frequencies among the characters currently present.
- Add their difference to `result`.
- Return `result` after processing every substring.

**Brute Force:**

- Generate every possible substring of `s`.
- For each substring, create a frequency dictionary by counting its characters.
- Find the maximum and minimum frequencies among the characters present in that substring.
- Calculate the beauty as `maximum frequency - minimum frequency`.
- Add the beauty to the total result.
- Return the total after evaluating all substrings.
- This approach repeatedly rebuilds frequency dictionaries for overlapping substrings, resulting in O(N³) time when substring processing and frequency counting are included.

**Optimal Approach:**

- **Step 1: Initialize variables**
  - Set `n = len(s)`.
  - Set `result = 0` to store the sum of the beauty values of all substrings.

- **Step 2: Fix the starting index**
  - Iterate through `i` from `0` to `n - 1`.
  - Each value of `i` represents the beginning of a new set of substrings.

- **Step 3: Reset the frequency dictionary**
  - Initialize `temp = {}` for every starting index.
  - This dictionary stores character frequencies for substrings beginning at index `i`.

- **Step 4: Extend the substring**
  - Iterate through `j` from `i` to `n - 1`.
  - Add the current character `s[j]` to the frequency dictionary.
  - Use `temp[s[j]] = temp.get(s[j], 0) + 1` to increment its frequency.
  - Because the substring grows one character at a time, previous frequency counts can be reused instead of recalculated.

- **Step 5: Calculate the beauty**
  - Use `max(temp.values())` to find the highest frequency among the characters present.
  - Use `min(temp.values())` to find the lowest frequency among the characters present.
  - Add their difference to `result`.
  - Characters that do not appear in the current substring are not included in the frequency dictionary, so their frequency of zero does not affect the minimum.

- **Step 6: Return the answer**
  - After evaluating all substrings, return `result`.

**Complexity:**

- **Brute Force:** Time Complexity: O(N³)  Space Complexity: O(K) auxiliary space for the frequency map of a substring.
- **Optimal Approach:** Time Complexity: O(N² × K)  Space Complexity: O(K) auxiliary space.
- Here, `N` is the length of `s` and `K` is the number of distinct characters in the current substring.
- There are O(N²) substrings.
- For each substring extension, `max(temp.values())` and `min(temp.values())` each scan up to `K` frequency values.
- Therefore, the general time complexity of your implementation is O(N² × K).
- For a fixed alphabet, such as lowercase English letters where `K <= 26`, this is effectively O(N²).

**Common Mistakes:**

- Rebuilding the frequency dictionary for every substring instead of updating it incrementally.
- Calculating the beauty as `minimum frequency - maximum frequency` instead of `maximum frequency - minimum frequency`.
- Including characters that do not appear in the current substring when calculating the minimum frequency.
- Forgetting to reset `temp` for each new starting index `i`.
- Adding the beauty only after the inner loop finishes. Every substring ending at `j` contributes separately.
- Confusing the number of substrings with the number of distinct characters.
- Assuming the beauty must be positive. A substring with all characters appearing equally often has beauty `0`.

**Recognition Pattern:**

- Problems asking for an aggregate value across all substrings.
- Problems involving character frequencies in each substring.
- Problems where a substring can be extended incrementally while maintaining useful state.
- Use nested loops with a frequency map when each new substring can be formed by adding one character to the previous substring.

**Key Observation:**

- Every substring is uniquely determined by its starting index `i` and ending index `j`.
- Fixing `i` and extending `j` allows the frequency dictionary to be updated in constant average time per added character.
- The beauty can then be calculated from the current maximum and minimum frequencies.
- Reusing the frequency counts avoids rebuilding them for every substring, reducing the solution from O(N³) to O(N² × K).
- With a fixed alphabet of 26 lowercase English letters, the solution runs in O(N²) time.

**Code:**

```python
# Optimal Solution
class Solution:
    def beautySum(self, s: str) -> int:
        n = len(s)
        result = 0
        for i in range(n):
            temp = {}
            for j in range(i, n):
                temp[s[j]] = temp.get(s[j],0) + 1
                result += (max(temp.values()) - min(temp.values()))
        return result
```

#### LeetCode 7 - Reverse Integer

**Question:**

Given a signed 32-bit integer `x`, return the integer obtained by reversing its digits.

If the reversed integer falls outside the signed 32-bit integer range `[-2³¹, 2³¹ - 1]`, return `0`.

Assume the environment does not allow storing 64-bit integers.

**Example:**

- Input: `x = 123`
- Output: `321`
- Explanation:
  - Reversing the digits of `123` produces `321`.

- Input: `x = -123`
- Output: `-321`
- Explanation:
  - Reverse the digits while preserving the negative sign.

- Input: `x = 120`
- Output: `21`
- Explanation:
  - Reversing the digits produces `021`, which is interpreted as `21`.

- Input: `x = 1534236469`
- Output: `0`
- Explanation:
  - The reversed integer exceeds the maximum signed 32-bit integer, so return `0`.

**Algorithm:**

- Initialize `result = 0` to store the reversed number.
- Record whether the input is negative.
- Convert a negative input to its positive magnitude.
- Extract the last digit using `x % 10`.
- Append that digit to the reversed number using `result = result * 10 + tmp`.
- Remove the last digit from `x` using integer division by `10`.
- Repeat until `x` becomes `0`.
- Restore the negative sign if necessary.
- Check the signed 32-bit integer bounds.
- Return `0` if the reversed number is out of range; otherwise, return the reversed number.

**Brute Force:**

- Convert the integer to a string.
- If the number is negative, separate the `-` sign from the digits.
- Reverse the digit string.
- Convert the reversed string back into an integer.
- Restore the negative sign.
- Check whether the result lies within the signed 32-bit integer range.
- Return `0` if it exceeds the bounds; otherwise, return the reversed integer.

**Optimal Approach:**

- **Step 1: Initialize variables**
  - Set `result = 0` to accumulate the reversed digits.
  - Set `neg_flag = 0` to indicate that the input is initially considered non-negative.

- **Step 2: Handle negative input**
  - If `x < 0`, set `neg_flag = 1`.
  - Convert the input to its positive magnitude using `x = 0 - x`.
  - This allows the digit-reversal logic to operate on a non-negative value.

- **Step 3: Handle zero**
  - If `x == 0`, return `0`.
  - Reversing zero still produces zero.

- **Step 4: Extract the last digit**
  - While `x > 0`, calculate `tmp = x % 10`.
  - The modulo operation extracts the last decimal digit.

- **Step 5: Append the digit**
  - Update `result = result * 10 + tmp`.
  - Multiplying by `10` shifts the existing digits left by one decimal position.
  - Adding `tmp` places the extracted digit at the end.

- **Step 6: Remove the processed digit**
  - Execute `x //= 10`.
  - Integer division removes the last digit.
  - Repeat until all digits have been processed.

- **Step 7: Check a non-negative result**
  - If `neg_flag == 0`, check whether `result > 2**31 - 1`.
  - If the result exceeds the maximum signed 32-bit integer, return `0`.
  - Otherwise, return `result`.

- **Step 8: Restore the negative sign**
  - If `neg_flag == 1`, execute `result = 0 - result`.
  - Check whether `result < -(2**31)`.
  - If it is below the minimum signed 32-bit integer, return `0`.
  - Otherwise, return the negative result.

**Complexity:**

- **Brute Force:** Time Complexity: O(D)  Space Complexity: O(D) for the string representation.
- **Optimal Approach:** Time Complexity: O(D)  Space Complexity: O(1) auxiliary space.
- Here, `D` is the number of digits in `x`.
- Each iteration processes one digit, so the algorithm takes linear time in the number of digits.
- The arithmetic approach uses only a constant number of variables.

**Common Mistakes:**

- Forgetting to preserve the sign of a negative input.
- Returning the reversed number without checking the signed 32-bit integer limits.
- Checking only the positive bound and forgetting that the minimum signed 32-bit integer is `-2³¹`, whose magnitude is one greater than `2³¹ - 1`.
- Forgetting to remove the processed digit using `x //= 10`, which would cause an infinite loop.
- Using `result = result + tmp * 10` instead of `result = result * 10 + tmp`.
- Forgetting that trailing zeros become leading zeros after reversal and therefore disappear in the integer representation.
- Assuming the positive and negative bounds are symmetric. The valid range is `[-2147483648, 2147483647]`.
- Checking the overflow only after reversing is appropriate in Python because integers can grow beyond 32 bits, but languages with fixed-width integers may need to check overflow before multiplication and addition.

**Recognition Pattern:**

- Problems involving digit extraction and reconstruction of integers.
- Problems where digits must be processed from right to left.
- Problems requiring sign preservation and overflow handling.
- Use modulo and integer division to process digits without converting the number to a string.

**Key Observation:**

- The last digit of a positive integer can be extracted using `x % 10`.
- Removing the last digit is accomplished using `x // 10`.
- The reversed number is constructed by shifting the current result one decimal position and adding the extracted digit.
- Negative inputs can be handled by reversing the magnitude and restoring the sign afterward.
- The final result must lie within the signed 32-bit integer range, or the function must return `0`.

**Code:**

```python
class Solution(object):
    def reverse(self, x):
        """
        :type x: int
        :rtype: int
        """
        result = 0
        neg_flag = 0
        if x < 0:
            neg_flag = 1
            x = 0 - (x)
        if x == 0:
            return 0
        while x > 0:
            tmp = x % 10
            result = result*10 + tmp
            x //= 10
        if neg_flag == 0:
            if result > (2**31) - 1:
                return 0
            return result
        else:
            result = 0 - result
            if result < 0 - 2**31:
                return 0
            return result
```
