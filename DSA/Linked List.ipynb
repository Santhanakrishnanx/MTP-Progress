## Middle of the Linked List

#### Brute Force
- Time Complexity : O(N + N/2)
- Space Complexity : O(1)

```python
# Definition for singly-linked list.
# class ListNode(object):
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution(object):
    def middleNode(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        curr = head
        count = 0
        while curr is not None:
            curr = curr.next
            count += 1
        target = count//2
        curr,count = head,0
        while curr is not None and count != target:
            curr = curr.next
            count += 1
        head = curr
        return head

```

#### Optimal Solution
- Time Complexity : O(N + N/2)
- Space Complexity : O(1)

```python
# Definition for singly-linked list.
# class ListNode(object):
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution(object):
    def middleNode(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        slow,fast = head,head
        while fast != None and fast.next != None:
            fast = fast.next.next
            slow = slow.next
        return slow
```

## Reverse a Linked List
#### Brute Force
- you should not reverse the values like first value to last and last to first
- Time Complexity : O(2N)
- Space Complexity : O(N)

```python
class Solution(object):
    def reverseList(self, head):
        temp = []
        curr = head
        while curr != None:
            temp.append(curr.val)
            curr = curr.next
        curr = head
        while curr != None:
            x = temp.pop()
            curr.val = x
            curr = curr.next
        return head
```

#### Optimal Solution
- Time Complexity : O(N)
- Space Complexity : O(1)

```python
class Solution(object):
    def reverseList(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        if head is None:
            return head
        prev = None
        temp = head 
        nxt = temp.next
        while temp != None:
            if nxt == None:
                temp.next = prev
                prev = temp
                temp = nxt
            elif prev == None:
                temp.next = None
                prev = temp
                temp = nxt
                nxt = nxt.next
            else:
                temp.next = prev
                prev = temp
                temp = nxt
                nxt = nxt.next
        head = prev
        return head
```

## Detecting Cycle in Linked List

#### Brute Force
- Time Complexity : O(N)
- Spce Complexity : O(N)

```python
class Solution(object):
    def hasCycle(self, head):
        temp = head
        my_set = set()
        while temp != None:
            if temp in my_set:
                return True
            else:
                my_set.add(temp)
                temp = temp.next
        return False
```

#### Optimal Solution
- Time Complexity : O(N)
- Spce Complexity : O(1)

```python
class Solution(object):
    def hasCycle(self, head):
        """
        :type head: ListNode
        :rtype: bool
        """
        if head is None:
            return False
        if head.next == None or head.next.next == None:
            return False
        slow = head
        fast = head.next.next
        while slow != None:
            if fast == None:
                return False
            if fast.next == None or fast.next.next == None:
                return False
            if slow == fast:
                return True
            slow = slow.next
            fast = fast.next.next
        return False
```

#### Optimal Solution
- Find loop exist or not using the slow and fast pointer once found
- reset slow to its next and start moving slow until fast is reached by incrementing a counter
- return the counter value which is the length of the cycle
- Time Complexity : O(N)
- Spce Complexity : O(1)

```python
class Solution(object):
    def hasCycle(self, head):
        """
        :type head: ListNode
        :rtype: bool
        """
        slow,fast = head,head
        while fast != None and fast.next is not None:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                count = 1
                slow = slow.next
                while slow != fast:
                    slow = slow.next
                    count += 1
                return count
        return 0
```

- This works because the distance from the head to the cycle's starting node equals the distance from the first meeting point to the cycle's starting node when moving forward around the cycle.

**Complexity:**

- Brute Force:
  - Time Complexity: O(N), because each node is visited at most once before the cycle is detected.
  - Space Complexity: O(N), because the set may store every node.

- Optimal Approach:
  - Time Complexity: O(N), because the slow and fast pointers traverse the list linearly.
  - Space Complexity: O(1), because only two pointers are used, regardless of the number of nodes.

**Common Mistakes:**

- Checking only whether `fast` is not `None` without checking `fast.next` before moving two steps.
- Returning the first meeting node instead of locating the cycle's starting node.
- Resetting both pointers to the head after the first meeting instead of resetting only `slow`.
- Moving `slow` and `fast` at different speeds during the second phase. Both must move one node at a time.
- Comparing node values instead of node references. Two different nodes can have the same value.
- Forgetting to return `None` when the list has no cycle.
- Using a set when constant auxiliary space is required.

**Recognition Pattern:**

- Problems involving detecting a cycle in a linked list.
- Problems asking for the exact node where a cycle begins.
- Problems requiring constant extra space.
- Use Floyd's Tortoise and Hare algorithm when two pointers can move at different speeds.

**Key Observation:**

- The first meeting of the slow and fast pointers confirms that a cycle exists, but it does not necessarily identify the cycle's starting node.
- After the first meeting, reset one pointer to the head and move both pointers one step at a time. Their next meeting identifies the cycle's entry point.
- The set-based approach detects the cycle by remembering visited nodes, while Floyd's algorithm achieves the same result using O(1) extra space.

**Code:**

```python
# Definition for singly-linked list.
# class ListNode(object):
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution(object):
    def detectCycle(self, head):
        """
        :type head: ListNode
        :rtype: ListNode
        """
        slow,fast = head,head
        while fast != None and fast.next != None:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                slow = head
                while slow != fast:
                    slow = slow.next
                    fast = fast.next
                return slow
        return None

# Brute Force
# Time Complexity : O(N)
# Space Complexity : O(N) 
class Solution(object):
    def detectCycle(self, head):
        temp = head
        my_set = set()
        while temp != None:
            if temp in my_set:
                return temp
            else:
                my_set.add(temp)
                temp = temp.next
        return None
```

#### LeetCode 237 - Delete Node in a Linked List

**Question:**

You are given a node from a singly linked list, and you are not given access to the head of the list. Delete the given node from the linked list.

- The given node is guaranteed not to be the tail node.
- Delete the given node by modifying the linked list in-place.
- Do not return anything.

**Example:**

- Input: `head = [4,5,1,9]`, `node = 5`
- Output: `[4,1,9]`

- Input: `head = [4,5,1,9]`, `node = 1`
- Output: `[4,5,9]`

**Algorithm:**

**Brute Force:**

- Start from the given node.
- Continue traversing until reaching the last node.
- At each step, copy the value of the next node into the current node.
- Store the current node in `prev` and move the pointer to the next node.
- When the traversal reaches the last node, `prev` points to the second-last node.
- Set `prev.next = None` to remove the last node from the linked list.
- This approach shifts the values of all subsequent nodes one position backward until the last node is removed.
- The given node is guaranteed not to be the tail, so the loop executes at least once and `prev` gets initialized.

**Optimal Approach:**

- Since the head of the linked list is unavailable, we cannot traverse from the head to find the previous node.
- Copy the value of the next node into the given node using `node.val = node.next.val`.
- Update the given node's next pointer to skip the next node using `node.next = node.next.next`.
- The next node is effectively removed from the linked list.
- The given node now contains the value that originally belonged to its next node, making it appear as though the given node was deleted.
- This works because the problem guarantees that the given node is not the tail, so `node.next` always exists.

**Complexity:**

- Brute Force:
  - Time Complexity: O(N), where N is the number of nodes from the given node to the tail.
  - Space Complexity: O(1), because only a constant number of pointers are used.

- Optimal Approach:
  - Time Complexity: O(1), because only the given node and its next node are modified.
  - Space Complexity: O(1), because no additional data structures are required.

**Common Mistakes:**

- Trying to access the previous node even though the head is not provided.
- Attempting to delete the given node directly, which is not possible in a singly linked list without access to its previous node.
- Forgetting to update `node.next` after copying the next node's value.
- Using the optimal approach on a tail node, where `node.next` would be `None`. The problem explicitly guarantees that the given node is not the tail.
- In the brute-force approach, forgetting to update `prev.next = None` after reaching the last node.
- Assuming the optimal approach physically removes the original node object. It actually bypasses the next node and overwrites the given node's value.

**Recognition Pattern:**

- Problems where a node must be deleted without access to the head of a singly linked list.
- Problems where the given node is guaranteed not to be the tail.
- In-place linked-list modification problems where copying the next node's value can replace the need for a previous pointer.

**Key Observation:**

- In a singly linked list, deleting a node normally requires access to its previous node so that the previous node's `next` pointer can bypass the target.
- When the head is unavailable, the trick is to copy the next node's value into the given node and skip the next node.
- This achieves the required result in O(1) time and O(1) space.
- The brute-force approach shifts every subsequent value and removes the final node, whereas the optimal approach changes only two pointers/values around the given node.

**Code:**

```python
# Definition for singly-linked list.
# class ListNode(object):
#     def __init__(self, x):
#         self.val = x
#         self.next = None
#Optimal Solution
class Solution(object):
    def deleteNode(self, node):
        """
        :type node: ListNode
        :rtype: void Do not return anything, modify node in-place instead.
        """
        node.val = node.next.val
        node.next = node.next.next

#Brute Force approach     
class Solution(object):
    def deleteNode(self, node):
        while node.next != None:
            node.val = node.next.val
            prev = node
            node = node.next
        prev.next = None     
```

#### LeetCode 25 - Reverse Nodes in k-Group

**Question:**

Given the head of a linked list, reverse the nodes of the list `k` at a time and return the modified list.

- If the number of remaining nodes is less than `k`, leave those nodes unchanged.
- You must modify the links between nodes, not their values.
- Use O(1) extra space.

**Example:**
- Input: `head = [1,2,3,4,5], k = 2`
- Output: `[2,1,4,3,5]`
- Explanation: Reverse each complete group of two nodes. The final node remains unchanged because fewer than two nodes remain.

- Input: `head = [1,2,3,4,5], k = 3`
- Output: `[3,2,1,4,5]`
- Explanation: Reverse the first three nodes. The remaining two nodes stay unchanged.

- Input: `head = [1,2,3,4,5,6], k = 2`
- Output: `[2,1,4,3,6,5]`
- Explanation: Each group of two nodes is reversed.

**Algorithm:**

**Brute Force:**
- Traverse the linked list and store the nodes in an array.
- Divide the array into groups of `k` nodes.
- Reverse each complete group in the array.
- Leave the final group unchanged if it contains fewer than `k` nodes.
- Reconnect the nodes in the resulting order.
- This approach requires extra space and does not satisfy the O(1) auxiliary-space requirement.

**Optimal Approach:**
- Use a helper function `reverseList(head)` to reverse a linked list in place.
- Initialize `temp` to the head of the current group.
- Find the kth node of the current group by traversing at most `k` nodes.
- If fewer than `k` nodes remain, return the current head without reversing the incomplete group.
- Save the node immediately after the group in `nxt`.
- Disconnect the current group by setting `k_node.next = None`.
- Reverse the isolated group using `reverseList(temp)`.
- The returned node becomes the new head of the reversed group, while the original `temp` becomes its tail.
- For the first group, update `head` to the new group head.
- For subsequent groups, connect the previous group's tail to the new group head using `prev.next = temp`.
- Connect the current group's tail to the remaining list using `k_node.next = nxt`.
- Move to the next group and repeat until all complete groups are processed.
- Return the updated head.

**Complexity:**
- **Brute Force:**
  - Time Complexity: O(N)
  - Space Complexity: O(N)
- **Optimal Approach:**
  - Time Complexity: O(N)
  - Space Complexity: O(1)
- Each node is visited a constant number of times. The helper function reverses each group in place without allocating additional data structures.

**Common Mistakes:**
- Reversing the remaining nodes when fewer than `k` nodes exist.
- Failing to verify that a complete group of `k` nodes exists before reversing it.
- Forgetting to save `k_node.next` before disconnecting the group.
- Forgetting either of the two required connections:
  - `prev.next = new_group_head`
  - `current_group_tail.next = next_group_head`
- Confusing the current group's original head, which becomes its tail, with the new head returned by `reverseList()`.
- Advancing `prev` before connecting the previous group's tail to the new group head.
- Resetting the node counter incorrectly when starting a new group.
- In the provided implementation, `prev = k_node` is used to retain the previous group's tail, but its meaning changes across iterations. Using a clearly named `prevGroupTail` makes the logic easier to follow.
- The `reverseList()` helper contains unnecessary branches for the first and last nodes. A standard three-pointer reversal is simpler.
- When `k = 1`, every group contains one node, so the list should remain unchanged.

**Recognition Pattern:**
- Use this pattern when a linked list must be rearranged in fixed-size groups.
- Identify the boundaries of each group before modifying its links.
- Save the remaining list, reverse the current group, and reconnect both ends.
- Use a previous-group tail pointer to stitch successive groups together.
- This pattern applies to linked list problems involving group reversal, segment manipulation, and in-place pointer reconnection.

**Key Observation:**
- Every complete group can be reversed independently.
- The original head of a group becomes its tail after reversal.
- The kth node identifies the group's end, and its original `next` pointer identifies the next group's beginning.
- Saving the next group's head before reversing prevents losing the remaining list.
- The incomplete final group must remain unchanged.
- The optimal approach uses O(1) auxiliary space because it only manipulates pointers.
- The most error-prone part is reconnecting the reversed groups, not reversing the individual group.

**Code:**

```python
# Definition for singly-linked list.
# class ListNode(object):
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution(object):
    def reverseList(self,head):
        if head == None or head.next == None:
            return head
        prev = None
        temp = head
        nxt = temp.next
        while temp != None:
            if nxt == None:
                temp.next = prev
                prev = temp
                temp = nxt
            elif prev == None:
                temp.next = None
                prev = temp
                temp = nxt
                nxt = nxt.next
            else:
                temp.next = prev
                prev = temp
                temp = nxt
                nxt = nxt.next
        head = prev
        return head

    def reverseKGroup(self, head, k):
        """
        :type head: Optional[ListNode]
        :type k: int
        :rtype: Optional[ListNode]
        """
        temp = head
        k_node = None
        while temp != None:
            count =  1
            nxt = None
            prev = k_node
            k_node = temp
            while count != k and k_node != None:
                if k_node.next != None:
                    count += 1
                k_node = k_node.next
            if count != k:
                return head
            if k_node:
                nxt = k_node.next
                k_node.next = None
            k_node = self.reverseList(temp)
            if temp == head:
                head = k_node
                temp,k_node = k_node,temp
                k_node.next = nxt
                temp = k_node.next
                continue
            temp,k_node = k_node,temp
            k_node.next = nxt
            prev.next = temp
            temp = k_node.next
        return head
```
