## Time Complexity

1. rate of increase of time with respect to input size
2. always calculate time complexity interms of worst case time complexity

3. avoid constant values in the notation of time complexity

#### Notation types:
1. big oh is worst case or upper bound
2. theta is average case 
3. omega is best case or lower bound

space complexity has big oh notation and is auxiliary space + input space 
- auxiliary sapce the extra space used to solve the problem
- input space is the space used to store the input

**Problem 5: Print all factors of a number**

Method 1: (Brute Force)
- it takes O(n) time complexity and O(k) space complexity where k is the total number of factors

```python
num = int(input("Enter the number: "))
result = []
for i in range(1, num+1):
    if(num % i == 0):
        result.append(i)

print(result)
```

Method 2:
- based on fact that a number cannot have factorial beyond (num/2)
- it has same time complexity as O(n) and O(k)

```python
num = int(input("Enter the number: "))
result = []
for i in range(1, num // 2 + 1):
    if(num % i == 0):
        result.append(i)
if (num % 2 == 0):
    result.append(int(num/2))
    result.append(num)
print(result)
```

Method 3:(optimal approach)
- it has time complexity of O(sqrt(n)) for finding the factors and O(n*logn) for sorting so total time complexity is

  O(sqrt(n)) + O(n*logn)
- space complexity is O(k)

```python
from math import sqrt
num = int(input("Enter the number: "))
result = []
for i in range(1,int(sqrt(num))+1):
    if(num % i == 0):
        result.append(i)
    if(num // i != i):
        if(num % (num // i) == 0):
            result.append(num // i)
result.sort()
print(result)
```

**Problem 3: Reverse an Array**

```python
def reverse(left,right,num):
    length = len(num)
    if(left >= right):
        return num
    num[left],num[right] = num[right],num[left]
    return reverse(left + 1, right - 1, num)

num = [1,2,3,4,5,6,7,8,9]
print(reverse(2,5,num))
```

**Problem 4: Check if a string is palindrome or not**

```python
def pal(left,right,s):
    if left >= right:
        return True
    if(s[left] == s[right]):
        return pal(left+1,right-1,s)
    else:
        return False
        
s = input("Enter the string:")
length = len(s)
value = pal(0,length-1 ,s)
if(value):
    print("Given String is palindrome")
else:
    print("Given String is not a palindrome")
```

**Problem 5: Find the Fibonacci Number**

```python
def fib(num):
    if (num == 0 or num == 1):
        return num
    return fib(num - 1) + fib(num - 2)

s = int(input("Enter the string:"))
value = fib(s)
str = f"The Fibonacci Number for {s} is : {value}"
print(str)
```

# Sorting

 ### Selection Sort
 - the time complexity is O($n^2$)
 - the space complexity is O(1)

```python
nums = [5,7,8,4,1,6,9,2,3]

def selection_sort(nums):
    n = len(nums)
    for i in range (0,n):
        min_ind = i
        for j in range (i+1, n):
            if nums[j] < nums[min_ind]:
                min_ind = j

        nums[i],nums[min_ind] = nums[min_ind],nums[i]


selection_sort(nums)
print(nums)
```

### Bubble Sort
- Time Complexity is O($n^2$)
- Space Complexity is O(1)

```python
nums = [5,7,8,4,1,6,9,2,3]

def Bubble_sort(nums):
    n = len(nums)
    for i in range (n-1,0,-1):
        for j in range (0,i):
            if nums[j] > nums[j+1]:
                nums[j],nums[j+1] = nums[j+1],nums[j]

Bubble_sort(nums)
print(nums)
```

Best case of Bubble sort

```python
nums = [1,2,9,5,4,10,12,14]

def Bubble_sort_optim(nums):
    n = len(nums)
    for i in range (n-1,0,-1):
        is_swap = False
        for j in range (0,i):
            if nums[j] > nums[j+1]:
                nums[j],nums[j+1] = nums[j+1],nums[j]
                is_swap = True

        if is_swap == False:
            return

Bubble_sort_optim(nums)
print(nums)
```

### Insertion Sort
- Time Complexity is O($n^2$)
- Space Complexity is O(1)

Method 1:

```python
nums = [5,7,8,4,1,6,9,2,3]

def Insertion_sort(nums):
    n = len(nums)
    for i in range (1,n):
        min_el = nums[i]
        pos = i
        if(nums[i] < nums[i - 1]):
            for j in range (i-1,-1,-1):
                if min_el < nums[j]:
                    nums[j+1] = nums[j]
                    pos = j
                else:
                    break
            nums[pos] = min_el

Insertion_sort(nums)
print(nums)
```

Method: 2

```python
nums = [5,7,8,4,1,6,9,2,3]

def Insertion_sort_2(nums):
    n = len(nums)
    for i in range (1,n):
        min_el = nums[i]
        j = i -1
        while j>= 0 and nums[j] > min_el:
            nums[j+1] = nums[j]
            j -= 1
        nums[j+1] = min_el

Insertion_sort_2(nums)
print(nums)
```

### Merge Sort
- Time Complexity is O($n*log$ $n$)
- Space Complexity is O($n$)

```python
nums = [3,1,2,4,1,5,2,6,4]

def merge_sort(nums):
    if len(nums) <= 1:
        return nums
    mid = int(len(nums)/2)
    left_arr = nums[ :mid]
    right_arr = nums[mid: ]
    left = merge_sort(left_arr)
    right = merge_sort(right_arr)
    return merge(left,right)
    
def merge(left,right):
    result = []
    i,j = 0,0
    n,m = len(left),len(right)
    while i < n and j < m:
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    if i < n:
        while i < n:
            result.append(left[i])
            i += 1
    if j < m:
        while j < m:
            result.append(right[j])
            j += 1
    return result

print(nums)
nums = merge_sort(nums)
print(nums)
```

### Quick Sort
Average Case:
- Time Complexity is O($n*log$ $n$)
- Space Complexity is O(1)
  
Worst Case: (when all the elements are same in an n length array)
- Time Complexity is O($n^2$)
- Space Complexity is O(1)

```python
nums = [5,7,8,4,1,6,9,2,3]

def Quick_sort(nums, low, high):
    if low < high:
        p = Partition(nums,low, high)
        Quick_sort(nums,low, p -1)
        Quick_sort(nums,p + 1,high)
    
def Partition(nums, low, high):
    i,j = low,high
    pivot = nums[low]
    while i < j:
        while nums[i] <= pivot and i <= high - 1:
            i += 1
        while nums[j] > pivot and j >= low + 1:
            j -= 1
        if i < j:
            nums[i],nums[j] = nums[j],nums[i]
    nums[low],nums[j] = nums[j],nums[low]
    return j

print(nums,"\n")
Quick_sort(nums, 0, len(nums)-1)
print(nums)
```

**Problem 4: Remove Duplicates from an Sorted array [in place]**

Method 1:
- Time complexity is O($n$)
- space complexity is O($n$)

```python
nums = [1,1,1,2,3,4,4,7,9,9,9,10]

freq_map = {}
for i in range(0,len(nums)):
    if nums[i] not in freq_map:
        freq_map[nums[i]] = 1
map = list( freq_map.keys())

for i in range(0, len(freq_map)):
    nums[i] = map[i]
    
end = len(freq_map) + 1
while end != len(nums):
    nums[end] = 0
    end += 1

print(nums)
```
