# Single element in a Sorted Array

- In a sorted list every element has a pair except the one element which does not have a pair find the element
- if the problem is to solved within O(log n) then use the binary search
- understand the pattern like when there is repetation of two elements then the index would be even and odd for the same two elements whereas when the left or right subspace contains the single element then this gets disturbed and changed to odd and even look for which side has this pattern then we omit the side which does not have this pattern
- if even and odd pattern is there in the mid then eliminate left space and search in right else move to left space and eliminate right space
- if odd and even pattern is there in the mid then eliminate the right space and search in left else eliminate the left and move to right space
- this method shrink the search space in each iteration
- the edge case are when there is single element in array so we just keep if statement to return that same lement
- the other problems are the 0 and n-1 elements can cause index out of bound error so we push the low and high to 1 and n-2
- so keep two sepearte if conditions to check the borders of the array

```python
class Solution(object):
    def singleNonDuplicate(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums)
        low,high = 1,n-2
        if n == 1:
            return nums[0]
        if nums[0] != nums[1]:
            return nums[0]
        if nums[n-1] != nums[n-2]:
            return nums[n-1]
        while low <= high:
            mid =(low + high)//2
            if nums[mid] == nums[mid + 1]:
                if mid % 2 == 0 and (mid + 1)%2 == 1:
                    low = mid + 1
                else:
                    high = mid - 1
            elif nums[mid] == nums[mid - 1]:
                if (mid- 1)%2 == 0 and mid%2 == 1:
                    low = mid + 1
                else:
                    high = mid - 1
            else:
                return nums[mid]
        
```

```python
def sqrt(n,m):
    low,high = 0, m
    result = 1
    while low <= high:
        mid = (low + high)//2
        for i in range(n):
            result = result*mid
            if result > m:
                break
        if result == m:
            return mid
        elif result > m:
            high = mid - 1
        else:
            low = mid + 1
        result = 1 
    return -1

N,M = 10,1000000000000
result = sqrt(N,M)
print(result)
```

# Find the smallest divisor

- Given an array of integers nums and an integer threshold, we will choose a positive integer divisor, divide all the array by it, and sum the division's result. Find the smallest divisor such that the result mentioned above is less than or equal to threshold.
- Each result of the division is rounded to the nearest integer greater than or equal to that element. (For example: 7/3 = 3 and 10/2 = 5).

Solution:
- take low as 1 and high as the max element in array calculate mid an compute the sum of the divisors by each element in array divided by mid(make sure to take the ceilvalue for the answer)
- then if result <= threshold store it in ans variable and reduce the search space to left subspace
- else reduce the search space to right subspace
- Time complexity is (log(max)*N)

```python
class Solution(object):
    def smallestDivisor(self, nums, threshold):
        """
        :type nums: List[int]
        :type threshold: int
        :rtype: int
        """
        n = len(nums)
        low,high = 1 ,max(nums)
        ans = -1
        while low <= high:
            mid = (low + high)//2
            result = 0
            for i in range(n):
                temp = (nums[i] + mid - 1)// mid
                result += temp
            if result <= threshold:
                ans = mid 
                high = mid - 1
            else:
                low = mid + 1
        return ans            
```

```python
class Solution(object):
    def shipWithinDays(self, weights, days):
        """
        :type weights: List[int]
        :type days: int
        :rtype: int
        """
        n = len(weights)
        sum_weight = 0
        for i in range(n):
            sum_weight += weights[i]
        low,high = max(weights),sum_weight
        while low <= high:
            mid = (low + high)//2
            day,sum_day = 1,0
            for i in range(n):
                if sum_day + weights[i] > mid:
                    day += 1
                    sum_day = weights[i]
                else:
                    sum_day += weights[i]
            if day <= days:
                high = mid - 1
            else:
                low = mid + 1
        return low
```

# Split array - largest sum

- Given an integer array nums and an integer k, split nums into k non-empty subarrays such that the largest sum of any subarray is minimized.

- Return the minimized largest sum of the split. A subarray is a contiguous part of the array.

### Binary Search Method`

- take low as maximum in the array and high as the sum of the array then calculate the mid
- then move through the array and check whether the arr[0] <= mid if yes then move to next element and check whether the sum of current and previous element is less than mid do it until the condition fails then just increase the splitted count and start the splitting from current element because till previous element it belong to other split
- now check whether the split is less than or equal to desired k then our minimum result is in left subspace else the minimum we look for is in right subspace
- - the time complexity is O(log (sum- max + 1)* N)

```python
class Solution(object):
    def splitArray(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        n = len(nums)
        if k > n:
            return -1
        ans = 0
        low,high = max(nums),sum(nums)
        while low <= high:
            mid = (low + high)//2
            splited,num_hold = 1,0
            for i in range(n):
                if num_hold + nums[i] <= mid:
                    num_hold += nums[i]
                else:
                    splited += 1
                    num_hold = nums[i]
            if splited <= k:
                ans = mid
                high = mid - 1
            else:
                low = mid + 1
        return ans
        
```

- The overall run time complexity should be O(log (m+n))

- to combine both the array into single array and then find the median for this newly created array

```python
class Solution(object):
    def findMedianSortedArrays(self, nums1, nums2):
        """
        :type nums1: List[int]
        :type nums2: List[int]
        :rtype: float
        """
        overall = []
        i,j = 0,0
        n1,n2 = len(nums1),len(nums2)
        while i < n1 and j < n2:
            if nums1[i] < nums2[j]:
                overall.append(nums1[i])
                i += 1
            else:
                overall.append(nums2[j])
                j += 1
        while i < n1:
            overall.append(nums1[i])
            i += 1
        while j < n2:
            overall.append(nums2[j])
            j += 1
        n = n1 + n2
        if n % 2 == 1:
            return overall[n//2]
        else:
            return (overall[n//2] + overall[(n//2) - 1])/2.0
```

#### LeetCode 2529 - Maximum Count of Positive Integer and Negative Integer

**Question:**

Given an array `nums` sorted in non-decreasing order, return the maximum between the number of positive integers and the number of negative integers in the array.

- Zero is neither positive nor negative.
- The array may contain negative numbers, zeros, and positive numbers.
- The array is sorted, so binary search can be used to locate the boundaries of negative and positive values.

**Example:**

- Input: `nums = [-2,-1,-1,1,2,3]`
- Output: `3`
- Explanation: There are 3 negative integers and 3 positive integers. The maximum count is 3.

- Input: `nums = [-3,-2,-1,0,0,1,2]`
- Output: `3`
- Explanation: There are 3 negative integers and 2 positive integers. The maximum count is 3.

- Input: `nums = [5,20,66,1314]`
- Output: `4`
- Explanation: There are 4 positive integers and 0 negative integers. The maximum count is 4.

**Algorithm:**

**Brute Force:**

- Initialize two counters, `neg_count` and `pos_count`, to zero.
- Traverse the array from left to right.
- If the current element is less than zero, increment `neg_count`.
- If the current element is greater than zero, increment `pos_count`.
- Ignore elements equal to zero because they belong to neither category.
- Return `max(neg_count, pos_count)`.
- This approach examines every element once and does not use the sorted property of the array.

**Optimal Approach:**

- Since the array is sorted in non-decreasing order, all negative numbers appear before zeros and positive numbers, while all positive numbers appear after zeros and negative numbers.
- Initialize `neg_end = -1` to represent the last negative index when no negative number exists.
- Initialize `pos_start = n` to represent the first positive index when no positive number exists.
- Use binary search to find the last negative number:
  - Initialize `low = 0` and `high = n - 1`.
  - Calculate `mid = (low + high) // 2`.
  - If `nums[mid] < 0`, record `mid` as `neg_end` and search the right half to find a later negative number.
  - Otherwise, search the left half because the last negative number must occur before `mid`.
  - After the search, the number of negative integers is `neg_end + 1`.
- Use another binary search to find the first positive number:
  - Reset `low = 0` and `high = n - 1`.
  - Calculate `mid = (low + high) // 2`.
  - If `nums[mid] > 0`, record `mid` as `pos_start` and search the left half for an earlier positive number.
  - Otherwise, search the right half because the first positive number must occur after `mid`.
  - After the search, the number of positive integers is `n - pos_start`.
- Return the maximum of the two counts.
- The boundary defaults ensure that the algorithm handles arrays containing no negative numbers or no positive numbers without needing separate special cases.

**Complexity:**

- Brute Force:
  - Time Complexity: O(N), because every element is checked once.
  - Space Complexity: O(1), because only two counters are used.

- Optimal Approach:
  - Time Complexity: O(log N), because two binary searches are performed.
  - Space Complexity: O(1), because only a constant number of variables are used.

**Common Mistakes:**

- Counting zero as either positive or negative.
- Forgetting that the array is sorted and using a linear scan when binary search is expected.
- Finding the first negative number instead of the last negative number.
- Finding the last positive number instead of the first positive number.
- Returning `neg_end` directly instead of `neg_end + 1`, since the count is one greater than the last negative index.
- Calculating the positive count as `n - pos_start`, not `n - pos_start - 1`, because the count includes the element at `pos_start`.
- Initializing `neg_end` to `-1` and `pos_start` to `n` so that missing categories correctly produce a count of zero.
- Adding special cases for arrays of length one or arrays containing only zeros when the binary searches and boundary defaults already handle them.

**Recognition Pattern:**

- The input array is sorted.
- The problem asks for counts of values satisfying a condition such as `< 0` or `> 0`.
- The valid values form a contiguous section of the sorted array.
- Binary search can locate the first or last index satisfying a condition, allowing the count to be calculated from the boundary.

**Key Observation:**

- The number of negative integers equals the index of the last negative element plus one.
- The number of positive integers equals the array length minus the index of the first positive element.
- Zeros are automatically excluded because they lie between the negative and positive sections.
- Your binary-search logic is correct, but the special cases for `n == 1` and arrays whose first and last elements are zero are unnecessary. Initializing `neg_end = -1` and `pos_start = n` handles these cases naturally.
- Using those boundary defaults also avoids `float("inf")` and `float("-inf")`, keeping both boundaries as integers.

**Code:**

```python
class Solution(object):
    def maximumCount(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        pos_start = float("inf")
        neg_end = float("-inf")
        n = len(nums)
        low, high = 0, n-1
        if n == 1:
            if nums[0] == 0:
                return 0
            else:
                return 1
        if nums[0] == 0 and nums[n-1] == 0:
            return 0
        while low <= high:
            mid = (low + high)//2
            if nums[mid] < 0:
                neg_end = mid
                low = mid + 1
            else:
                high = mid - 1 
        low, high = 0, n-1
        while low <= high:
            mid = (low + high)//2
            if nums[mid] > 0:
                pos_start = mid
                high = mid - 1
            else:
                low = mid + 1
        return max(neg_end + 1, n - pos_start)
```

**Complexity:**

- Define `count(x)` as the number of elements satisfying `nums[i] >= x`.
- As `x` increases, `count(x)` can only stay the same or decrease.
- If `count(x) > x`, search larger candidates because the count may decrease until it equals the candidate.
- If `count(x) < x`, search smaller candidates because lowering the candidate can increase the count.

**Code:**

```python
class Solution(object):
    def specialArray(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums)
        low,high = 1,max(nums)
        while low <= high:
            mid = (low + high)//2
            count = 0
            for i in range(n):
                if nums[i] >= mid:
                    count += 1
            if count == mid:
                return mid
            elif count > mid:
                low = mid + 1
            else:
                high = mid - 1
        return -1
```

- The dimensions must satisfy an ordering condition such as `L >= W`.
- The goal is to minimize the difference between the factors.

- Problems involving closest factors often have their best solution near the square root of the target number.

- For any valid factor pair `(L, W)`, where `L >= W` and `L * W == area`, the smaller factor cannot exceed `√area`.
- Therefore, checking factors up to the square root is sufficient to find the pair with the smallest difference.

**Code:**

```python
class Solution(object):
    def constructRectangle(self, area):
        """
        :type area: int
        :rtype: List[int]
        """
        limit = int(area**0.5)
        L_final,W_final = 0,0
        difference = float('inf')
        for i in range(1,limit + 1):
            L,W = 0,0
            if area % i == 0:
                L = max(i,area / i)
                W = area / L
                temp = L - W
                if temp < difference:
                    difference = temp
                    L_final,W_final = L,W
        return [L_final,W_final]
```

**Algorithm:**

**Brute Force:**

- Initialize `rows = 0` and `remaining = n`.
- Start with the first row, which requires 1 coin.
- While `remaining` is at least the number of coins required for the next row, subtract that number from `remaining`.
- Increment `rows` after completing each row.
- Increase the required coins by 1 for the next row.
- Stop when there are not enough coins to complete the next row.
- Return `rows`.
- This approach builds the staircase one row at a time and takes O(√N) time because completing `k` rows requires `k(k + 1) / 2` coins.

**Optimal Approach — Binary Search:**

- Initialize `low = 1` and `high = n` to represent the possible number of rows.
- Initialize `ans = 0` to store the best valid number of rows found.
- Calculate `mid = (low + high) // 2` to test a candidate number of complete rows.
- Calculate the total coins required to complete `mid` rows using the triangular-number formula:

  `total_coins = mid * (mid + 1) // 2`

- If `total_coins <= n`, the candidate number of rows is possible.
- Store `mid` in `ans` and search for a larger number of rows by setting `low = mid + 1`.
- Otherwise, the candidate requires too many coins, so search for fewer rows by setting `high = mid - 1`.
- Continue until `low > high`.
- Return `ans`, which contains the maximum number of complete rows that can be formed.
- Binary search works because the total coins required increase as the number of rows increases.

**Complexity:**

- Brute Force:
  - Time Complexity: O(√N), because the number of complete rows grows proportionally to the square root of the number of coins.
  - Space Complexity: O(1), because only a constant number of variables are used.

- Optimal Approach:
  - Time Complexity: O(log N), because binary search halves the candidate range in every iteration.
  - Space Complexity: O(1), because only a constant number of variables are used.

**Common Mistakes:**

- Using `mid * (mid - 1) // 2` when testing `mid` complete rows. The correct formula is `mid * (mid + 1) // 2`.
- Returning `ans - 1` after binary search when `ans` already stores the number of complete rows.
- Forgetting to update `ans` when a candidate is valid.
- Searching for a smaller candidate when the current candidate can be completed. Instead, search the right half for a larger answer.
- Using `/` instead of `//` for integer coin calculations in Python 3.
- Handling `n == 1` separately when the general binary-search algorithm already works for it.
- Setting `high` to an unnecessarily large value. `high = n` is valid, though the answer cannot exceed approximately `√(2n)`.

**Recognition Pattern:**

- The problem asks for the maximum number of complete groups or rows that can be formed with a limited resource.
- The resource requirement increases as the number of groups increases.
- The condition can be checked by calculating the total resource needed for a candidate answer.
- Binary search on the answer is useful when feasibility is monotonic: if a certain number of rows is possible, every smaller number is also possible.

**Key Observation:**

- The number of coins required to complete `k` rows is the sum `1 + 2 + ... + k`, which equals `k(k + 1) / 2`.
- If a candidate `k` can be completed, there may still be enough coins for more rows, so search the right half.
- If a candidate requires more than `n` coins, fewer rows must be considered.
- Your current code uses `mid * (mid - 1) / 2`, which calculates the coins needed for `mid - 1` rows rather than `mid` rows. This is why you subtracting 1 from `ans` is not a reliable fix.
- Use `mid * (mid + 1) // 2` and return `ans` directly. The special case for `n == 1` then becomes unnecessary.

**Code:**

```python
class Solution(object):
    def arrangeCoins(self, n):
        """
        :type n: int
        :rtype: int
        """
        low,high = 1,n
        ans = 0
        if n == 1:
            return 1
        while low <= high:
            mid = (low + high)//2
            total_coin = (mid*(mid - 1))/2
            if total_coin <= n:
                ans = mid 
                low = mid + 1
            else:
                high = mid - 1
        return ans - 1
```

- The closest heater determines the minimum radius needed for an individual house.
- Binary search is useful when finding the nearest value in a sorted array.
- The final answer is the maximum among the nearest-heater distances for all houses.

**Key Observation:**

- The minimum radius needed for one house is its distance to the nearest heater.
- The minimum radius needed for all houses is the maximum of those individual minimum distances.
- After sorting the heaters, `bisect_left` identifies the first heater at or to the right of a house.
- The nearest heater can only be this heater or its immediate predecessor. Any heater farther away on either side cannot be closer than these candidates.
- Your solution correctly handles houses outside the heater range by initializing both distances to infinity and checking whether each candidate exists.
- The key pattern is **sort + binary search for nearest candidates + minimize per item + maximize the overall answer**.

**Code:**

```python
class Solution(object):
    def findRadius(self, houses, heaters):
        """
        :type houses: List[int]
        :type heaters: List[int]
        :rtype: int
        """
        heaters.sort()
        radius = 0
        for house in houses:
            idx = bisect_left(heaters, house)
            left = float('inf')
            right = float('inf')
            if idx > 0:
                left = house - heaters[idx - 1]
            if idx < len(heaters):
                right = heaters[idx] - house
            radius = max(radius, min(left, right))
        return radius
```

  - This lets binary search eliminate half of the possible values at each iteration.

  - Following `nums[i]` creates a sequence of indices.

```python
class Solution(object):
    def singleNonDuplicate(self, nums):
        n = len(nums)
        low,high = 1,n-2
        if n == 1:
            return nums[0]
        if nums[0] != nums[1]:
            return nums[0]
        if nums[n-1] != nums[n-2]:
            return nums[n-1]
        while low <= high:
            mid =(low + high)//2
            if nums[mid] == nums[mid + 1]:
                if mid % 2 == 0:
                    low = mid + 1
                else:
                    high = mid - 1
            elif nums[mid] == nums[mid - 1]:
                if mid%2 == 1:
                    low = mid + 1
                else:
                    high = mid - 1
            else:
                return nums[mid]
```

```python
# Method 1 
class Solution(object):
    def searchRange(self, nums, target):
        result = []
        n = len(nums)
        low,high = 0,n-1
        first,last = -1,-1
        while low <= high :
            mid = (low + high)//2
            if nums[mid] >= target :
                if nums[mid] == target:
                    first = mid
                high = mid - 1
            else:
                low = mid + 1
        low,high = 0,n-1
        while low <= high :
            mid = (low + high)//2
            if nums[mid] > target :
                last = mid
                high = mid - 1
            else:
                low = mid + 1
        if first == -1:
            return [-1, -1]
        elif first != -1 and last == -1:
            return [first,n-1]
        else:
            last = last - 1
        result.append(first)
        result.append(last)
        return result

# Method 2
class Solution(object):
    def searchRange(self, nums, target):
        result = []
        n = len(nums)
        low,high = 0,n-1
        first,last = -1,-1
        while low <= high :
            mid = (low + high)//2
            if nums[mid] == target :
                first = mid
                high = mid - 1
            elif nums[mid] >= target:
                high = mid - 1
            else:
                low = mid + 1
        low,high = 0,n-1
        if first == -1 :
            return [-1,-1]
        while low <= high :
            mid = (low + high)//2
            if nums[mid] == target :
                last = mid
                low = mid + 1
            elif nums[mid] >= target:
                high = mid - 1
            else:
                low = mid + 1
        result.append(first)
        result.append(last)
        return result 
```

#### LeetCode 493 - Reverse Pairs

**Question:**

Given an integer array `nums`, return the number of reverse pairs in the array.

A reverse pair is a pair of indices `(i, j)` satisfying:

- `0 <= i < j < len(nums)`
- `nums[i] > 2 * nums[j]`

The solution should count all such pairs efficiently.

**Example:**

- Input: `nums = [1,3,2,3,1]`
- Output: `2`
- Explanation: The reverse pairs are `(3,1)` and `(3,1)` at their respective valid positions, corresponding to the values `3` at index `1` and `1` at index `4`, and `3` at index `3` and `1` at index `4`.

- Input: `nums = [2,4,3,5,1]`
- Output: `3`
- Explanation: The reverse pairs are `(4,1)`, `(3,1)`, and `(5,1)`.

**Algorithm:**

- Use merge sort to divide the array into smaller sections and count reverse pairs efficiently.
- Define a recursive function `merge_sort(arr)` that returns two values:
  - The array sorted in ascending order.
  - The number of reverse pairs within that array.
- If the array contains zero or one element, return the array itself and a count of `0`, because no pair can exist.
- Divide the array into two halves using `mid = n // 2`.
- Recursively sort the left half and obtain `left` and `count_left`.
- Recursively sort the right half and obtain `right` and `count_right`.
- Initialize `count = count_left + count_right` to include reverse pairs found entirely within either half.
- Count the reverse pairs that cross between the left and right halves.
- Since both halves are sorted, use a pointer `j = 0` to scan the right half.
- For every element `left[i]`, move `j` forward while `left[i] > 2 * right[j]`.
- Every right-half element before `j` satisfies the reverse-pair condition for the current `left[i]`.
- Add `j` to `count` because all right-half elements from index `0` to `j - 1` form valid reverse pairs with `left[i]`.
- Do not reset `j` for every element in `left`, because the left half is sorted and the valid range in the right half can only move forward.
- After counting cross-half pairs, merge `left` and `right` into a single sorted array.
- Initialize two pointers `i = 0` and `j = 0` for the merge operation.
- Compare `left[i]` and `right[j]`, appending the smaller element to `merged` and advancing its corresponding pointer.
- Append any remaining elements from either half to `merged`.
- Return `merged` and `count`.
- Finally, call `merge_sort(nums)` and return the total reverse-pair count.

**Brute Force:**

- Initialize `count = 0`.
- Iterate through every index `i` from `0` to `n - 1`.
- For every `i`, iterate through every index `j` greater than `i`.
- Check whether `nums[i] > 2 * nums[j]`.
- If the condition is satisfied, increment `count`.
- Return `count` after checking every possible pair.
- Time Complexity: O(N²), because every pair of indices may need to be checked.
- Space Complexity: O(1) auxiliary space, excluding the input array.

**Optimal Approach:**

- Use modified merge sort to count reverse pairs while sorting the array.
- Divide the array into two halves recursively.
- Count reverse pairs within the left half.
- Count reverse pairs within the right half.
- Count cross-half reverse pairs using the sorted order of the two halves.
- For each element in the left half, find how many elements in the right half satisfy `left[i] > 2 * right[j]`.
- Use a monotonically increasing right pointer so the cross-half counting step takes linear time.
- Merge the two sorted halves into a sorted array.
- Return the combined count of left-half pairs, right-half pairs, and cross-half pairs.
- Time Complexity: O(N log N), because merge sort has O(log N) levels and each level performs O(N) work.
- Space Complexity: O(N), because merging and recursive slicing require additional memory.

**Complexity:**

- Brute Force — Time Complexity: O(N²), because every pair of indices is examined.
- Brute Force — Space Complexity: O(1) auxiliary space.
- Optimal — Time Complexity: O(N log N), because the array is divided recursively and each merge level performs linear work.
- Optimal — Space Complexity: O(N), because sorted halves and merged arrays require additional storage.

**Common Mistakes:**

- Checking `nums[i] > nums[j]` instead of the required condition `nums[i] > 2 * nums[j]`.
- Counting only reverse pairs within each half and forgetting cross-half pairs.
- Counting cross-half pairs after merging the halves without preserving the original left-versus-right relationship.
- Resetting the right pointer to `0` for every left element, which can increase the counting step to O(N²).
- Forgetting that the left and right halves must be sorted before the two-pointer counting step.
- Using the ordinary merge comparison `left[i] <= right[j]` to count reverse pairs. The counting condition must specifically compare `left[i]` with `2 * right[j]`.
- Forgetting to merge the two halves after counting, which prevents the recursive caller from receiving a sorted array.
- Returning only the count without returning the sorted array from the recursive function.
- Modifying the original array unintentionally when using in-place sorting. Your implementation instead returns newly merged arrays.
- Assuming negative values can be handled by reasoning only about absolute magnitudes. The actual comparison `left[i] > 2 * right[j]` must be preserved for all integer values.

**Recognition Pattern:**

- The problem asks for the number of index pairs satisfying a comparison between two elements.
- A brute-force solution checks every pair in O(N²).
- The condition involves an inequality that can be evaluated efficiently after sorting.
- Merge sort can count pairs between two sorted halves before merging them.
- A monotonically increasing pointer can find the number of valid partners for each element without restarting the scan.
- Similar patterns appear in Count of Smaller Numbers After Self, Count of Range Sum, and inversion-counting problems.

**Key Observation:**

- A reverse pair requires `i < j` and `nums[i] > 2 * nums[j]`.
- Merge sort ensures the left and right halves are sorted before cross-half pairs are counted.
- For each `left[i]`, all right-half elements satisfying `left[i] > 2 * right[j]` form a valid prefix of the sorted right half.
- Because the left half is sorted in ascending order, the right pointer never needs to move backward as `i` increases.
- The total count consists of three parts:
  - Reverse pairs entirely within the left half.
  - Reverse pairs entirely within the right half.
  - Reverse pairs crossing from the left half to the right half.
- Your solution correctly counts cross-half pairs before merging, then returns the sorted merged array and combined count.
- The key pattern is **merge sort + count cross-half pairs with two pointers + merge**.

**Code:**

```python
class Solution(object):
    def reversePairs(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        def merge_sort(arr):
            n = len(arr)
            if n <= 1:
                return arr, 0
            mid = n // 2
            left, count_left = merge_sort(arr[:mid])
            right, count_right = merge_sort(arr[mid:])
            count = count_left + count_right
            j = 0
            for i in range(len(left)):
                while j < len(right) and left[i] > 2 * right[j]:
                    j += 1
                count += j
            merged = []
            i = j = 0
            while i < len(left) and j < len(right):
                if left[i] <= right[j]:
                    merged.append(left[i])
                    i += 1
                else:
                    merged.append(right[j])
                    j += 1
            merged.extend(left[i:])
            merged.extend(right[j:])
            return merged, count
        _, total_count = merge_sort(nums)
        return total_count    
```
