# Fractional Knapsack

```python
# Time Complexity : O(N log N) + O(N)
# Space Complexity : O(N) and O(1) if weight and values are given as an object of a class
def solve(arr,w):
    # arr.sort( key = lambda x: x.value / x.weight, reverse = True)
    # use this when the weight and values are given as an object of a class
    temp = []
    for i in range(len(arr)):
        temp.append([arr[i][0] / arr[i][1],i])
    temp.sort(key=lambda x: x[0], reverse=True)
    result = 0
    for i in range(len(arr)):
        if arr[temp[i][1]][1] <= w:
            result += arr[temp[i][1]][0]
            w -= arr[temp[i][1]][1]
        else:
            result += temp[i][0]*w
            break
    return result
    
arr = [(100,20),(60,10),(100,50),(200,50)]
w = 90
x = solve(arr,w)
print(x)
```

#### LeetCode 455 - Assign Cookies

**Question:**

- You are given two arrays, `g` and `s`.
- `g[i]` represents the minimum cookie size required to satisfy child `i`.
- `s[j]` represents the size of cookie `j`.
- Each child can receive at most one cookie, and each cookie can be assigned to at most one child.
- A child is satisfied if the assigned cookie's size is greater than or equal to the child's greed factor.
- Return the maximum number of satisfied children.

**Example:**

- Input: `g = [1,2,3]`, `s = [1,1]`
- Output: `1`
- Explanation: Only one child can be satisfied because both cookies have size `1`, and only the child with greed factor `1` can accept a cookie.

**Algorithm:**

**Brute Force:**

- For each child, search for an unused cookie whose size is at least the child's greed factor.
- If a suitable cookie is found, assign it to the child and mark the cookie as used.
- Repeat until all children have been considered or no suitable cookies remain.
- A naive implementation may take O(N × M) time, where `N` is the number of children and `M` is the number of cookies.
- A greedy assignment order is needed to guarantee the maximum number of satisfied children.

**Optimal Approach - Greedy With Two Pointers:**

- Sort the greed array `g` in ascending order so the child with the smallest requirement is considered first.
- Sort the cookie array `s` in ascending order so the smallest available cookie is considered first.
- Initialize two pointers, `l = 0` for the children and `r = 0` for the cookies.
- Initialize `result = 0` to track the number of satisfied children.
- Compare the greed factor `g[l]` with the current cookie size `s[r]`.
- If `g[l] <= s[r]`, the cookie can satisfy the current child.
  - Increment `result` because one more child is satisfied.
  - Move `l` forward to consider the next child.
  - Move `r` forward because the cookie has been used.
- If `g[l] > s[r]`, the cookie is too small for the current child.
  - Move `r` forward to try a larger cookie.
  - Keep `l` unchanged because the current child has not been satisfied.
- Continue until either all children or all cookies have been considered.
- Return `result`.

**Complexity:**

- Time Complexity: O(N log N + M log M + N + M)
  - Sorting the greed array takes O(N log N).
  - Sorting the cookie array takes O(M log M).
  - The two-pointer traversal takes O(N + M).
  - This is commonly simplified to O(N log N + M log M).
- Space Complexity: O(1) auxiliary space for the two-pointer traversal, excluding the space used internally by Python's sorting implementation.

**Common Mistakes:**

- Forgetting to sort both arrays before applying the greedy strategy.
- Moving the child pointer when the cookie is too small.
- Counting a child as satisfied when the cookie size is smaller than the greed factor.
- Reusing a cookie for multiple children.
- Trying to give the largest cookie to the child with the largest greed factor first without considering whether a smaller cookie could satisfy another child.
- Describing the total space complexity as strictly O(1) without mentioning that Python's sorting implementation may use additional memory.

**Recognition Pattern:**

- Use the greedy two-pointer pattern when matching two sorted collections under a minimum requirement.
- Examples include assigning cookies to children, matching jobs to workers, and assigning resources to tasks.
- When the goal is to maximize the number of successful matches, consider satisfying the smallest requirement with the smallest resource that can meet it.

**Key Observation:**

- If the current cookie is too small for the child with the smallest remaining greed factor, it cannot satisfy any of the remaining children because their greed factors are at least as large.
- Therefore, discard that cookie and try the next larger one.
- If the cookie satisfies the current child, assigning it to that child preserves larger cookies for children with higher requirements.
- Sorting allows the algorithm to make these greedy choices in linear time after sorting.

**Code:**

```python
# Time Complexity : O(N logN) + O(M logM) + O(M)
# Space Complexity : O(1)
class Solution:
    def findContentChildren(self, g: list[int], s: list[int]) -> int:
        result = 0
        g.sort()
        s.sort()
        l,r = 0,0
        while l < len(g) and r < len(s):
            if g[l] <= s[r]:
                result += 1
                l += 1
            r += 1
        return result 
```

#### LeetCode 860 - Lemonade Change

**Question:**

- At a lemonade stand, each lemonade costs 5 dollars.
- Customers pay using bills of 5, 10, or 20 dollars.
- Each customer buys exactly one lemonade and pays with one bill.
- Initially, you have no change.
- Return `True` if you can provide the correct change to every customer in order; otherwise, return `False`.

**Example:**

- Input: `bills = [5,5,5,10,20]`
- Output: `True`
- Explanation:
  - The first three customers pay with 5-dollar bills.
  - The fourth customer pays with a 10-dollar bill, so return one 5-dollar bill as change.
  - The fifth customer pays with a 20-dollar bill, so return one 10-dollar bill and one 5-dollar bill as change.

- Input: `bills = [5,5,10,10,20]`
- Output: `False`
- Explanation:
  - After serving the first four customers, you have two 10-dollar bills but no 5-dollar bills.
  - You cannot provide 15 dollars in change to the final customer.

**Algorithm:**

**Brute Force:**

- Maintain a collection of the bills received from customers.
- For each customer, calculate the change required by subtracting the 5-dollar lemonade cost from the bill.
- Find a combination of previously received bills that adds up to the required change.
- If no valid combination exists, return `False`.
- Otherwise, remove the bills used for change, store the current customer's bill, and continue.
- This approach can become inefficient if many combinations of bills must be checked.

**Optimal Approach - Greedy With a Hash Map:**

- Initialize an empty hash map `temp` to store the frequency of each bill denomination.
- Traverse the `bills` array in customer order.
- If the customer pays with a 5-dollar bill, no change is required.
- If the customer pays with a 10-dollar bill:
  - Check whether at least one 5-dollar bill is available.
  - If available, use one 5-dollar bill as change.
  - Otherwise, return `False` because the customer cannot be served.
- If the customer pays with a 20-dollar bill:
  - First, check whether both a 10-dollar bill and a 5-dollar bill are available.
  - If both are available, use one of each to provide 15 dollars in change.
  - Otherwise, check whether at least three 5-dollar bills are available.
  - If available, use three 5-dollar bills as change.
  - If neither combination is possible, return `False`.
- After providing change successfully, add the customer's bill to the hash map.
- Continue until all customers have been served.
- Return `True` if every customer can be served successfully.

**Why Prefer a 10-Dollar Bill and a 5-Dollar Bill Over Three 5-Dollar Bills?**

- Both combinations provide the required 15 dollars in change for a 20-dollar bill.
- A 10-dollar bill cannot provide change to a customer paying with a 10-dollar bill because that customer needs 5 dollars in change.
- Using a 10-dollar bill together with a 5-dollar bill preserves more 5-dollar bills for future customers who pay with 10-dollar bills.
- Therefore, when both combinations are available, prefer one 10-dollar bill and one 5-dollar bill.

**Complexity:**

- Time Complexity: O(N), where N is the number of customers.
- Space Complexity: O(1), because there are only three possible bill denominations: 5, 10, and 20 dollars.

**Common Mistakes:**

- Forgetting that a customer paying with a 10-dollar bill requires one 5-dollar bill as change.
- Forgetting that a customer paying with a 20-dollar bill requires 15 dollars in change.
- Using three 5-dollar bills when a 10-dollar bill and a 5-dollar bill are both available.
- Adding the current customer's bill to the hash map before checking whether sufficient change is available.
- Forgetting to return `False` immediately when no valid change combination exists.
- Assuming that having enough total money guarantees that the correct denominations are available.

**Recognition Pattern:**

- Use the greedy approach when processing transactions in order and making locally optimal choices that preserve resources for future transactions.
- This pattern applies to change-making problems where the available denominations and transaction rules are fixed.
- When multiple combinations can satisfy the current transaction, choose the combination that preserves the most useful resources for future transactions.

**Key Observation:**

- A 5-dollar bill is the most important resource because it is the only bill that can provide change to a customer paying with a 10-dollar bill.
- For a 20-dollar bill, prefer using one 10-dollar bill and one 5-dollar bill over three 5-dollar bills whenever possible.
- The algorithm only needs to track the counts of the three denominations, so a hash map is sufficient, although three individual counters would also work.
- Your implementation follows the correct greedy strategy and updates the bill counts only after successfully providing change.

**Code:**

```python
class Solution:
    def lemonadeChange(self, bills: list[int]) -> bool:
        temp = {}
        for i in bills:
            if i == 10:
                if temp.get(5,0) > 0:
                    temp[5] -= 1
                else:
                    return False
            if i == 20:
                if (temp.get(5,0) > 0 and temp.get(10,0) > 0):
                    temp[5] -= 1
                    temp[10] -= 1
                elif temp.get(5,0) >= 3:
                    temp[5] -= 3
                else:
                    return False
            temp[i] = temp.get(i,0) + 1
        return True
```

#### LeetCode 55 - Jump Game

**Question:**

- You are given an integer array `nums`.
- Each element `nums[i]` represents the maximum number of steps you can jump forward from index `i`.
- You start at index `0`.
- Return `True` if you can reach the last index; otherwise, return `False`.

**Example:**

- Input: `nums = [2,3,1,1,4]`
- Output: `True`
- Explanation: Jump from index `0` to index `1`, then jump to the last index.

- Input: `nums = [3,2,1,0,4]`
- Output: `False`
- Explanation: You eventually reach index `3`, where the maximum jump length is `0`. You cannot reach the last index.

**Algorithm:**

**Brute Force:**

- Start at index `0`.
- At each index, try every possible jump length from `1` to `nums[i]`.
- Recursively check whether any of these jumps can reach the last index.
- If at least one possible path reaches the last index, return `True`.
- If every possible path fails, return `False`.
- This approach can take exponential time because it explores many possible jump sequences.

**Optimal Approach - Greedy:**

- Initialize `max_len = 0` to track the farthest index reachable using the positions processed so far.
- Traverse the array from index `0` to the last index.
- Before processing index `i`, check whether `i > max_len`.
- If this condition is true, the current index is beyond the farthest reachable position, so return `False`.
- Otherwise, the current index is reachable.
- Calculate the farthest position reachable from the current index using `i + nums[i]`.
- Update `max_len = max(max_len, i + nums[i])` to retain the farthest reachable position found so far.
- Continue until the last index is processed.
- If the loop finishes without returning `False`, return `True` because the last index is reachable.

**Complexity:**

- Time Complexity: O(N), because each array element is processed once.
- Space Complexity: O(1), because only a constant amount of extra space is used.

**Common Mistakes:**

- Updating the farthest reachable position without first checking whether the current index is reachable.
- Returning `False` when `nums[i] == 0` without checking whether an earlier index can jump past it.
- Confusing the maximum jump length with the farthest index reachable.
- Using `i + nums[i]` as the new farthest position directly instead of taking the maximum with the previous `max_len`.
- Forgetting that reaching or passing the last index is sufficient; the jump does not need to land exactly on it.

**Recognition Pattern:**

- Use the greedy farthest-reach pattern when each position determines how far forward you can move.
- Look for problems asking whether a destination is reachable or how far you can progress through an array.
- Maintain the farthest reachable position instead of exploring every possible path.

**Key Observation:**

- `max_len` represents the farthest index reachable from any position processed so far.
- If the current index is greater than `max_len`, there is no possible way to reach that index, so the destination is also unreachable.
- If the current index is reachable, update the farthest reachable position using `i + nums[i]`.
- The greedy approach works because only the farthest reachable boundary matters; tracking every individual jump sequence is unnecessary.

**Code:**

```python
# Time Complexity : O(N)
# Space Complexity : O(1)
class Solution:
    def canJump(self, nums: list[int]) -> bool:
        n = len(nums)
        max_len = 0
        for i in range(len(nums)):
            if i > max_len:
                return False
            max_len = max(max_len, i + nums[i])
        return True
```

#### LeetCode 45 - Jump Game II

**Question:**

- You are given an integer array `nums`.
- Each element `nums[i]` represents the maximum number of steps you can jump forward from index `i`.
- You start at index `0`.
- Return the minimum number of jumps required to reach the last index.
- You can assume that the last index is always reachable.

**Example:**

- Input: `nums = [2,3,1,1,4]`
- Output: `2`
- Explanation: Jump from index `0` to index `1`, then jump from index `1` to index `4`. The minimum number of jumps is `2`.

- Input: `nums = [2,3,0,1,4]`
- Output: `2`
- Explanation: Jump from index `0` to index `1`, then jump from index `1` to index `4`.

**Algorithm:**

**Brute Force - Recursion:**

- Start at index `0` with `jump = 0`.
- If the current index reaches or passes the last index, return the number of jumps taken.
- Otherwise, try every possible jump length from `1` to `nums[index]`.
- For each possible jump, recursively calculate the minimum jumps required to reach the last index.
- Increment the jump count for every recursive move.
- Keep the minimum result among all possible jumps.
- Return the minimum number of jumps after exploring every possible path.
- This approach can take exponential time because many possible jump sequences are explored repeatedly.

**Better Approach - Dynamic Programming:**

- Define a DP state representing the minimum number of jumps needed to reach the destination from a particular index.
- Initialize the destination index with `0` jumps because no further movement is required.
- Process the remaining indices from right to left.
- For each index `i`, try every reachable next index from `i + 1` to `i + nums[i]`, without exceeding the last index.
- Calculate the minimum jumps by taking `1 + dp[next_index]` for each valid next index.
- Store the smallest result in `dp[i]`.
- Return `dp[0]`, which represents the minimum jumps required from the starting index.
- This bottom-up approach takes O(N²) time and O(N) space.
- Memoization can also avoid recomputing states, but the DP state should depend on the current index rather than both the index and the number of jumps already taken.

**Optimal Approach - Greedy:**

- Initialize `jump = 0` to count the jumps taken.
- Initialize `l = 0` and `r = 0` to represent the current range of indices reachable using the current number of jumps.
- Continue while `r` is before the last index.
- Initialize `farthest = 0` to track the farthest index reachable with one additional jump from the current range.
- Traverse every index `i` from `l` to `r`, inclusive.
- For each index, calculate `temp = i + nums[i]`.
- Update `farthest = max(farthest, temp)` to find the maximum reach from all indices in the current range.
- After processing the current range, set `l = r + 1` to begin the next range of indices.
- Set `r = farthest` because the next jump expands the reachable boundary to the farthest position found.
- Increment `jump` because one jump has been used to reach the next range.
- Repeat until the current range reaches or passes the last index.
- Return `jump`.

**Complexity:**

- Recursion:
  - Time Complexity: Exponential in the worst case, commonly bounded by O(2^N) for this branching process.
  - Space Complexity: O(N) for the maximum recursion depth.
- Dynamic Programming:
  - Time Complexity: O(N²).
  - Space Complexity: O(N).
- Greedy:
  - Time Complexity: O(N), because each index is processed at most once across the ranges.
  - Space Complexity: O(1), because only a few variables are maintained.

**Common Mistakes:**

- Confusing Jump Game I, which asks whether the destination is reachable, with Jump Game II, which asks for the minimum number of jumps.
- Incrementing `jump` for every index instead of once for each expanded reachable range.
- Updating `r` before calculating the farthest reachable index from the entire current range.
- Forgetting to include `r` in the loop by using `range(l, r + 1)`.
- Updating `farthest` using only the current index instead of comparing it with the maximum reach found so far.
- Assuming the recursive solution is efficient because it finds the minimum; it can explore many overlapping paths.
- Using a DP table indexed by both `index` and `jump` unnecessarily. The minimum jumps remaining can be represented using the current index alone.

**Recognition Pattern:**

- Use the greedy range-expansion pattern when the problem asks for the minimum number of jumps to reach a destination.
- Think of each jump as expanding the range of indices that can be reached.
- Within the current range, find the farthest position reachable with one more jump.
- This is similar to a level-by-level traversal: each range represents positions reachable using the same number of jumps.

**Key Observation:**

- `l` and `r` represent the range of indices reachable using the current number of jumps.
- Every index inside this range can be reached with the same number of jumps.
- To minimize jumps, examine every index in the current range and find the farthest position reachable from any of them.
- Expanding to that farthest position ensures the next range covers as much ground as possible.
- Each index is processed only once, giving O(N) time complexity.
- The greedy approach relies on the problem's guarantee that the last index is reachable.

**Code:**

```python
# Greedy Approach
# Time Complexity : O(N)
# Space Complexity : O(1)
class Solution:
    def jump(self, nums: list[int]) -> int:
        jump,l,r = 0,0,0
        while r < len(nums) - 1:
            farthest = 0
            for i in range(l,r + 1):
                temp = i + nums[i]
                farthest = max(farthest, temp)
            l = r + 1
            r = farthest
            jump += 1
        return jump

# Dynamic Programming Approach
# Time Complexity : O(N^2)
# Space Complexity : O(N)
class Solution:
    def solve(self, nums, index, jump):
        if index >= len(nums) - 1:
            return jump
        if dp[index][jump] != -1:
            return dp[index][jump]
        min_jump = float('inf')
        for i in range(1, nums[index] + 1):
            min_jump = min(min_jump, self.solve(nums, index + i, jump + 1))
        dp[index][jump] = min_jump
        return dp[index][jump]

    def jump(self, nums: list[int]) -> int:
        global dp
        n = len(nums)
        dp = [[-1] * n for _ in range(n)]
        x = self.solve(nums, 0, 0)
        return x

# Recursion Approach
# Time Complexity : O(N^N)
# Space Complexity : O(N)
class Solution:
    def solve(self, nums, index, jump):
        if index >= len(nums) - 1:
            return jump
        min_jump = float('inf')
        for i in range(1, nums[index] + 1):
            min_jump = min(min_jump, self.solve(nums, index + i, jump + 1))
        return min_jump

    def jump(self, nums: list[int]) -> int:
        x = self.solve(nums, 0, 0)
        return x
```
