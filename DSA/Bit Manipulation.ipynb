# Convert Decimal to Binary
- Time Complexity : O(log N)
- Space Complexity : O(log N)

````python
def convertBinary(num:int)->str:
    result = ""
    while num > 0:
        if num % 2 == 1:
            result += "1"
        else:
            result += "0"
        num = num // 2
    result = result[::-1]
    return result
num = 13
x = convertBinary(num)
print("Integer : ",num)
print("Binary : ",x)
````

Note : 
- In 1's compliment just flip the binary digits when a number is in binary form
- In 2's compliment flip the binary digits and add 1 to it when a number is in its binary form 

````python
def onesCompliment(s:str)->str:
    result = ""
    for i in range(len(s)):
        if s[i] == "1":
            result += "0"
        else:
            result += "1"
    return result

def twosCompliment(s:str)->str:
    temp = ""
    for i in range(len(s)):
        if s[i] == "1":
            temp += "0"
        else:
            temp += "1"
    carry,result = 1,list(temp)
    for i in range((len(result)-1),-1,-1):
        x = int(result[i]) + carry
        if x == 2:
            result[i] = "0"
            carry = 1
        else:
            result[i] = str(x)
            carry = 0
            break
    return "".join(result)
 
x = "1101"
num = onesCompliment(x)
num_2 = twosCompliment(x)
print("Binary : ",x)
print("1's Compliment : ",num)
print("2's Compliment : ",num_2)
````

# Find XOR of numbers from L to R

Problem Statement: Given two integers L and R. Find the XOR of the elements in the range [L , R].
- Example 1: Input : L = 3 , R = 5  Output : 2 Explanation : answer = (3 ^ 4 ^ 5) = 2.
- Example 2: Input : L = 1, R =  3 Output : 0 Explanation : answer = (1 ^ 2 ^ 3) = 0.

````python
# Brute Force
def xor(l,r):
    res = 0
    for i in range(l,r + 1):
        res = res ^ i
    return res

l,r = 3,5
x = xor(l,r)
print(x)
l,r = 1,3
x = xor(l,r)
print(x)
````

````python
# Optimal Solution
def xortillN(n):
    if n % 4 == 1:
        return 1
    if n % 4 == 2:
        return n + 1
    if n % 4 == 3:
        return 0
    return n

l,r = 3,5
x,y = xortillN(l - 1),xortillN(r)
print(x^y)
l,r = 1,3
x,y = xortillN(l - 1),xortillN(r)
print(x^y)
````

# XOR from 0 to N

## Repeating Pattern

The XOR of numbers from `0` to `N` follows a pattern that repeats every 4 numbers.

| N % 4 | XOR from 0 to N |
|-------|------------------|
| 0     | N                |
| 1     | 1                |
| 2     | N + 1            |
| 3     | 0                |

### Pattern to Remember

- `N % 4 == 0` → `N`
- `N % 4 == 1` → `1`
- `N % 4 == 2` → `N + 1`
- `N % 4 == 3` → `0`

## Why Does the Pattern Repeat?

The first four numbers cancel:

`0 ^ 1 ^ 2 ^ 3 = 0`

Similarly:

- `4 ^ 5 ^ 6 ^ 7 = 0`
- `8 ^ 9 ^ 10 ^ 11 = 0`
- `12 ^ 13 ^ 14 ^ 15 = 0`

Therefore, the XOR pattern repeats every 4 numbers.

To find:

`L ^ (L + 1) ^ ... ^ R`

use:

`XOR(0 to R) ^ XOR(0 to L - 1)`

### Why?

Numbers from `0` to `L - 1` occur twice and cancel because:

`x ^ x = 0`

#### LeetCode 231 - Power of Two

**Question:**

Given an integer `n`, return `True` if it is a power of two. Otherwise, return `False`.

An integer `n` is a power of two if there exists an integer `x` such that `n = 2^x`.

**Example:**

- Input: `n = 16`
- Output: `True`
- Explanation: `16 = 2^4`, so it is a power of two.

- Input: `n = 3`
- Output: `False`
- Explanation: `3` cannot be expressed as a power of two.

- Input: `n = 1`
- Output: `True`
- Explanation: `1 = 2^0`, so it is a power of two.

**Algorithm:**

- A positive power of two has exactly one set bit (`1`) in its binary representation.
- For example:
  - `8 = 1000`
  - `16 = 10000`
  - `32 = 100000`
- The bitwise expression `n & (n - 1)` removes the rightmost set bit of `n`.
- If `n` is a positive power of two, removing its only set bit produces `0`.
- Since `0` and negative numbers are not powers of two, ensure that `n > 0`.

**Brute Force:**

- If `n <= 0`, return `False`.
- Repeatedly divide `n` by `2` while it is divisible by `2`.
- After the loop:
  - If `n == 1`, return `True`.
  - Otherwise, return `False`.
- A power of two can be reduced to `1` by repeatedly dividing by `2`.
- A positive number that is not a power of two will eventually remain greater than `1` and odd.

**Optimal Approach:**

- **Step 1: Check zero**
  - If `n == 0`, return `False`.
  - More generally, checking `n <= 0` is safer and explicitly excludes negative values.

- **Step 2: Apply the bitwise operation**
  - Calculate `x = n & (n - 1)`.
  - Subtracting `1` changes the rightmost set bit to `0` and changes the trailing zeros to `1`.
  - Performing bitwise AND with the original number clears its rightmost set bit.

- **Step 3: Check the result**
  - If `x == 0`, return `True`.
  - Otherwise, return `False`.
  - For positive integers, this works because a power of two has exactly one set bit.

- **Example: `n = 8`**
  - Binary representation of `n`: `1000`
  - Binary representation of `n - 1`: `0111`
  - Bitwise AND: `1000 & 0111 = 0000`
  - Result: `True`

- **Example: `n = 10`**
  - Binary representation of `n`: `1010`
  - Binary representation of `n - 1`: `1001`
  - Bitwise AND: `1010 & 1001 = 1000`
  - Result: `False`

**Complexity:**

- **Brute Force:** Time Complexity: O(log N)  Space Complexity: O(1) auxiliary space.
- **Optimal Approach:** Time Complexity: O(1)  Space Complexity: O(1) auxiliary space.
- The optimal approach uses a constant number of bitwise operations without loops.

**Common Mistakes:**

- Checking only `n == 0` instead of `n <= 0` can be confusing when handling negative inputs. In Python, the bitwise expression can also produce `0` for certain negative values, such as `n = -1`, so the positivity check is essential.
- Forgetting that `1` is a power of two because `1 = 2^0`.
- Using `n & (n - 1) == 0` without checking positivity can incorrectly classify some negative integers.
- Confusing `n & (n - 1)` with `n & (n + 1)`. The former clears the rightmost set bit.
- In the brute-force approach, forgetting to handle `n <= 0` before entering the loop can lead to incorrect results or unnecessary processing.

**Recognition Pattern:**

- Problems involving powers of two or numbers with exactly one set bit.
- Problems where a bitwise operation can eliminate the rightmost set bit.
- Use `n & (n - 1)` to check whether a positive integer contains exactly one set bit.

**Key Observation:**

- Every positive power of two has exactly one set bit in binary.
- The expression `n & (n - 1)` clears the rightmost set bit.
- Therefore, a positive integer is a power of two if and only if `n & (n - 1) == 0`.
- Always check `n > 0` before applying this test.

**Code:**

````python
class Solution(object):
    def isPowerOfTwo(self, n):
        """
        :type n: int
        :rtype: bool
        """
        if n == 0:
            return False
        x = n & (n-1)
        if x == 0:
            return True
        return False

# Brute Force 
class Solution(object):
    def isPowerOfTwo(self, n):
        if n <= 0:
            return False
        while n % 2 == 0:
            n = n // 2
        if n == 1:
            return True
        else:
            return False
````

#### LeetCode 2220 - Minimum Bit Flips to Convert Number

**Question:**

Given two integers `start` and `goal`, return the minimum number of bit flips required to convert `start` into `goal`.

A bit flip changes a binary digit from `0` to `1` or from `1` to `0`.

**Example:**

- Input: `start = 10`, `goal = 7`
- Output: `3`
- Explanation:
  - `start = 10` → Binary: `1010`
  - `goal = 7` → Binary: `0111`
  - XOR: `1010 ^ 0111 = 1101`
  - The XOR result contains three set bits, so three bit flips are required.

**Algorithm:**

- Use the bitwise XOR operator (`^`) to identify the bit positions where `start` and `goal` differ.
- XOR produces `1` when two corresponding bits differ and `0` when they are equal.
- Count the number of set bits (`1`s) in the XOR result.
- The number of set bits is the minimum number of bit flips required.

**Brute Force:**

- Convert `start` and `goal` into binary strings.
- Pad the shorter binary representation with leading zeros so both strings have equal length.
- Compare the bits at corresponding positions.
- Increment the counter whenever the bits differ.
- Return the final counter.
- This approach works but requires binary string conversion and character-by-character comparison.

**Optimal Approach:**

- **Step 1: Find differing bits**
  - Calculate `temp = start ^ goal`.
  - Every set bit in `temp` represents a position where `start` and `goal` differ.

- **Step 2: Initialize the counter**
  - Set `count = 0` to track the number of differing bits.

- **Step 3: Examine the rightmost bit**
  - Check `temp & 1`.
  - If the result is `1`, the rightmost bit is set, so increment `count`.
  - If the result is `0`, do not increment the counter.

- **Step 4: Shift right**
  - Execute `temp = temp >> 1`.
  - This removes the rightmost bit from consideration by shifting all remaining bits one position to the right.

- **Step 5: Repeat until zero**
  - Continue checking and shifting while `temp > 0`.
  - When `temp` becomes `0`, all set bits have been counted.

- **Step 6: Return the result**
  - Return `count`, which is the minimum number of bit flips required.

**Brute Force:**

- **Time Complexity:** O(B), where B is the number of bits in the binary representation.
- **Space Complexity:** O(B) for the binary strings.

**Complexity:**

- **Optimal Approach:** Time Complexity: O(B)  Space Complexity: O(1) auxiliary space.
- Here, B is the number of bits in the XOR result.
- Each loop iteration processes one bit, so the algorithm takes time proportional to the number of bits.
- The algorithm uses only a few integer variables, requiring constant auxiliary space.

**Common Mistakes:**

- Using bitwise AND (`&`) instead of XOR (`^`) to compare `start` and `goal`. XOR directly identifies differing bit positions.
- Counting the number of set bits in `start` alone instead of counting them in `start ^ goal`.
- Forgetting to shift `temp` right inside the loop, which causes an infinite loop whenever `temp > 0`.
- Incrementing the counter for every bit instead of only when `temp & 1` is nonzero.
- Forgetting that leading zeros do not affect the answer because XOR automatically compares corresponding bit positions.

**Recognition Pattern:**

- Problems asking for the minimum number of bit changes between two integers.
- Problems involving the Hamming distance between binary representations.
- Problems where differing bit positions need to be counted.
- Use XOR followed by a set-bit counting technique.

**Key Observation:**

- XOR produces `1` exactly where the corresponding bits of `start` and `goal` differ.
- Every differing bit requires exactly one flip.
- Therefore, the answer is the number of set bits in `start ^ goal`.
- Your loop counts these set bits one at a time by checking the least significant bit and shifting right.

**Code:**

````python
class Solution:
    def minBitFlips(self, start: int, goal: int) -> int:
        temp = start ^ goal
        count = 0
        while temp > 0:
            if temp & 1:
                count += 1
            temp = temp >> 1
        return count 
````

#### LeetCode 137 - Single Number II

**Question:**

Given an integer array `nums` where every element appears three times except for one element that appears exactly once, find and return the element that appears only once.

The solution should ideally use linear time complexity and constant extra space.

**Example:**

- Input: `nums = [2,2,3,2]`
- Output: `3`
- Explanation:
  - The number `2` appears three times.
  - The number `3` appears only once.
  - Therefore, the answer is `3`.

**Algorithm:**

- **Brute Force:** Use a hash map to count the frequency of each number and return the number with frequency `1`.
- **My Solution:** Sort the array so that duplicate values appear consecutively, then compare adjacent elements to identify the unique number.
- **Optimal Approach:** Use two bit masks, `ones` and `twos`, to track bits that have appeared once and twice, respectively.
- The bitwise operations update the masks so that bits corresponding to values appearing three times are removed.

**Brute Force:**

- Initialize an empty dictionary named `hash_map`.
- Iterate through `nums` and count the frequency of each number.
- Use `hash_map.get(num, 0) + 1` to increment the frequency.
- Iterate through the dictionary entries.
- If a number has frequency `1`, return that number.

**My Solution:**

- **Step 1: Sort the array**
  - Execute `nums.sort()`.
  - Equal numbers become adjacent, allowing duplicates to be identified through neighboring elements.

- **Step 2: Initialize the index**
  - Set `i = 0` and iterate through the sorted array.

- **Step 3: Compare adjacent elements**
  - If `i + 1 < n`, compare `nums[i]` and `nums[i + 1]` using XOR.
  - If `nums[i] ^ nums[i + 1] == 0`, the two values are equal.
  - Since duplicates occur three times, skip the group by incrementing `i` by `3`.
  - Otherwise, return `nums[i]` because it is the unique number.

- **Step 4: Handle the final element**
  - If no next element exists, return `nums[i]`.

- **Important limitation:**
  - This approach relies on the problem guarantee that every repeated number occurs exactly three times and that the sorted groups remain intact.
  - Sorting changes the original array and increases the time complexity.
  - The adjacent comparison can be simplified to `nums[i] == nums[i + 1]` because XOR equals zero exactly when the two integers are equal.

**Optimal Approach:**

- **Step 1: Initialize two bit masks**
  - `ones = 0` tracks bits that have appeared once modulo three.
  - `twos = 0` tracks bits that have appeared twice modulo three.
  - Each bit position is tracked independently.

- **Step 2: Update `ones`**
  - Execute `ones = (ones ^ nums[i]) & ~(twos)`.
  - `ones ^ nums[i]` toggles the bits contributed by the current number.
  - `~twos` clears any bits that are already recorded in `twos`.
  - Therefore, bits that would enter the `twos` state are removed from `ones`.

- **Step 3: Update `twos`**
  - Execute `twos = (twos ^ nums[i]) & ~(ones)`.
  - `twos ^ nums[i]` updates the bits represented by `twos`.
  - `~ones` clears bits that are currently recorded in `ones`.
  - This maintains the intended bit-state transitions.

- **Step 4: Process all numbers**
  - Repeat both updates for every element in `nums`.
  - Each bit follows a cycle corresponding to its number of occurrences:
    - Appears once → recorded in `ones`.
    - Appears twice → recorded in `twos`.
    - Appears three times → removed from both masks.

- **Step 5: Return `ones`**
  - All numbers appearing three times cancel out at the bit level.
  - The number appearing once remains represented in `ones`.
  - Return `ones`.

**Complexity:**

- **Brute Force:** Time Complexity: O(N)  Space Complexity: O(N) for the hash map.
- **My Solution:** Time Complexity: O(N log N)  Space Complexity: O(1) auxiliary space if sorting is performed in place, excluding the sorting implementation's internal stack space.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space.
- The optimal approach performs a constant number of bitwise operations per element.

**Common Mistakes:**

- Assuming this is the same as Single Number I, where every other element appears twice. This problem requires handling elements that appear three times.
- In the sorting solution, incrementing the index by `2` instead of `3` after identifying a repeated value would break the grouping logic.
- Forgetting to check `i + 1 < n` before accessing `nums[i + 1]` causes an index-out-of-range error.
- Using a hash map when constant auxiliary space is required by the optimal solution.
- Updating `twos` before `ones` in this exact implementation changes the mask update semantics. Keep the order consistent with the given solution.
- Misunderstanding `~` as a logical NOT. In Python, `~` performs bitwise complement on an integer.
- Assuming `ones` simply XORs every number. The additional mask operations are essential to cancel bits after three occurrences.

**Recognition Pattern:**

- Every element appears three times except one.
- The problem asks for linear time and constant extra space.
- Bitwise state tracking can replace frequency counting.
- Use the `ones` and `twos` bit-mask technique when each repeated element appears three times.

**Key Observation:**

- Each bit position can be tracked independently according to how many times it has appeared, modulo three.
- The `ones` and `twos` masks represent the first and second occurrence states.
- When a bit reaches its third occurrence, it is cleared from both masks.
- Therefore, all numbers appearing three times cancel out, leaving the unique number in `ones`.

**Code:**

````python
# Optimal Solution
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        n = len(nums)
        ones,twos = 0,0
        for i in range(n):
            ones = (ones ^ nums[i]) & ~(twos)
            twos = (twos ^ nums[i]) & ~(ones)
        return ones 

# My Solution
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        n,i = len(nums),0
        nums.sort()
        while i < n:
            if i + 1 < n: 
                if nums[i]^nums[i + 1] == 0:
                    i += 3
                    continue
                else:
                    return nums[i]
            return nums[i]

# Brute Force
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        hash_map = {}
        for num in nums:
            hash_map[num] = hash_map.get(num, 0) + 1
        for key in hash_map:
            if hash_map[key] == 1:
                return key
````

#### LeetCode 260 - Single Number III

**Question:**

Given an integer array `nums` in which exactly two elements appear only once and all the other elements appear exactly twice, return the two elements that appear only once.

You must solve the problem in linear time and use constant extra space.

**Example:**

- Input: `nums = [1,2,1,3,2,5]`
- Output: `[3,5]`
- Explanation:
  - `1` appears twice.
  - `2` appears twice.
  - `3` and `5` appear only once.
  - Therefore, the answer contains `3` and `5`.

**Algorithm:**

- **Brute Force:** Use a hash map to count the frequency of every number and collect the two numbers whose frequency is `1`.
- **Optimal Approach:** Use XOR to eliminate duplicate numbers and isolate the two unique numbers.
- XOR has two important properties:
  - `a ^ a = 0`
  - `a ^ 0 = a`
- XORing the entire array cancels all duplicate pairs, leaving the XOR of the two unique numbers.
- Since the two unique numbers are different, their XOR contains at least one set bit. Use that bit to divide the array into two groups and recover each unique number independently.

**Brute Force:**

- Initialize an empty dictionary named `hash_map`.
- Iterate through `nums` and count the frequency of each number using `hash_map.get(num, 0) + 1`.
- Iterate through the dictionary keys.
- If a number has frequency `1`, append it to `result`.
- Return `result` after finding both unique numbers.

**Optimal Approach:**

- **Step 1: XOR all elements**
  - Initialize `temp = 0`.
  - Iterate through every number and calculate `temp = temp ^ num`.
  - All numbers appearing twice cancel each other.
  - At the end, `temp = a ^ b`, where `a` and `b` are the two unique numbers.

- **Step 2: Isolate the rightmost set bit**
  - Calculate `x = (temp & (temp - 1)) ^ temp`.
  - The expression `temp & (temp - 1)` clears the rightmost set bit.
  - XORing the result with `temp` isolates that bit.
  - Therefore, `x` contains exactly one set bit corresponding to a position where `a` and `b` differ.

- **Step 3: Initialize two groups**
  - Set `b1 = 0` and `b2 = 0`.
  - These variables will hold the two unique numbers.

- **Step 4: Divide numbers into two groups**
  - Iterate through every number in `nums`.
  - If `num & x == 0`, the number has a `0` at the selected bit position. XOR it into `b1`.
  - Otherwise, the number has a `1` at that position. XOR it into `b2`.
  - Equal numbers always belong to the same group because their bits at the selected position are identical.
  - Therefore, duplicates cancel within their respective groups.

- **Step 5: Collect the unique numbers**
  - After processing the array, `b1` contains one unique number and `b2` contains the other.
  - Append both values to `result` and return the list.
  - The order of the two returned numbers does not matter.

**Complexity:**

- **Brute Force:** Time Complexity: O(N)  Space Complexity: O(N) for the hash map.
- **Optimal Approach:** Time Complexity: O(N)  Space Complexity: O(1) auxiliary space, excluding the output list.
- The optimal approach makes two passes through the array and uses a constant number of integer variables.

**Common Mistakes:**

- Trying to recover both unique numbers from the total XOR alone. The XOR gives only `a ^ b`, not the individual numbers.
- Using `temp & (temp - 1)` directly as the selected bit. This expression clears the rightmost set bit; XORing it with `temp` isolates that bit.
- Forgetting that `temp` must be nonzero. Since the two unique numbers are distinct, their XOR is nonzero.
- Grouping numbers using an arbitrary bit without ensuring that the selected bit differs between the two unique numbers.
- Forgetting to XOR numbers within each group. Simply assigning the numbers to groups would not eliminate duplicates.
- Using a hash map when the optimal solution requires constant auxiliary space.
- Assuming the output must be sorted. The problem accepts either order.

**Recognition Pattern:**

- Exactly two numbers appear once while every other number appears twice.
- The problem requires linear time and constant extra space.
- XOR can cancel pairs, but the remaining XOR must be separated into two individual values.
- Use a set bit in the XOR result to partition the array into two groups.

**Key Observation:**

- XORing all elements leaves `a ^ b`, because every duplicated number cancels out.
- Since `a` and `b` are different, at least one bit differs between them.
- Isolating a set bit from `a ^ b` identifies a position where the two unique numbers have different bits.
- Partitioning the array using that bit places the two unique numbers in separate groups while keeping every duplicate pair together.
- XORing each group independently recovers both unique numbers.

**Code:**

````python
class Solution:
    def singleNumber(self, nums: List[int]) -> List[int]:
        result,temp = [],0
        for num in nums:
            temp = temp^num
        x = (temp & (temp-1)) ^ temp
        b1,b2 = 0,0
        for num in nums:
            if num & x == 0:
                b1 = b1 ^ num
            else:
                b2 = b2 ^ num
        result.append(b1)
        result.append(b2)
        return result

# Brute Force 
class Solution:
    def singleNumber(self, nums: List[int]) -> List[int]:
        result,hash_map = [],{}
        for num in nums:
            hash_map[num] = hash_map.get(num,0) + 1
        for key in hash_map:
            if hash_map[key] == 1:
                result.append(key)
        return result
````
